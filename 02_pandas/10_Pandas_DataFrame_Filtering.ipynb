<h3 style="color:#6FA8DC; font-weight:bold">10. Filtering a DataFrame</h3>

⭐⭐⭐⭐⭐ **VVVV Important**

Filtering means **selecting only those rows which satisfy a particular condition**.

This is one of the most important operations in Pandas because in real ML datasets we constantly need to find rows based on conditions.

We will use the IPL matches dataset.



<h5 style="color:#78B89A; font-weight:bold;">Load the IPL dataset</h5>

````python
import pandas as pd

df = pd.read_csv("datasets/ipl-matches.csv")

df.head()
````

<h5 style="color:#78B89A; font-weight:bold;">1. Basic DataFrame Filtering</h5>

We can create a condition using a relational operator.

For example, find all matches where the winning team is **Chennai Super Kings**.

```python
df["WinningTeam"] == "Chennai Super Kings"
```

This gives a Boolean Series containing `True` and `False`.



````python
df["WinningTeam"] == "Chennai Super Kings"
````

Now pass this Boolean condition back inside the DataFrame.

```python
df[df["WinningTeam"] == "Chennai Super Kings"]
```

Only the rows where the condition is `True` are returned.



````python
df[df["WinningTeam"] == "Chennai Super Kings"]
````

<h5 style="color:#78B89A; font-weight:bold;">2. Find all the Final Winners</h5>

We can filter the dataset where `MatchNumber` is `"Final"`.



````python
final_winners = df[df["MatchNumber"] == "Final"]

final_winners[["Season", "Team1", "Team2", "WinningTeam"]]
````

Here we first filter the rows:

```python
df["MatchNumber"] == "Final"
```

and then display only the useful columns.



<h5 style="color:#78B89A; font-weight:bold;">3. How many Super Over finishes have occurred?</h5>

The `SuperOver` column contains `Y` for matches that had a Super Over.

First filter them:

```python
df[df["SuperOver"] == "Y"]
```



````python
super_over_matches = df[df["SuperOver"] == "Y"]

super_over_matches
````

To find **how many** such matches occurred, use `shape[0]`:

```python
super_over_matches.shape[0]
```



````python
super_over_matches.shape[0]
````

We can also do it directly:

```python
df[df["SuperOver"] == "Y"].shape[0]
```

⭐ `shape[0]` → number of rows



````python
df[df["SuperOver"] == "Y"].shape[0]
````

<h5 style="color:#78B89A; font-weight:bold;">4. How many matches has CSK won in Kolkata?</h5>

Here we need **two conditions**:

1. City should be `"Kolkata"`
2. Winning team should be `"Chennai Super Kings"`

For multiple conditions, use `&`.

```python
(condition1) & (condition2)
```

⚠️ Put each condition inside parentheses.



````python
csk_kolkata = df[
    (df["City"] == "Kolkata") &
    (df["WinningTeam"] == "Chennai Super Kings")
]

csk_kolkata
````

Now count the rows:



````python
csk_kolkata.shape[0]
````

⭐ Multiple conditions:

```python
df[
    (condition1) &
    (condition2)
]
```

`&` means **AND**.



<h5 style="color:#78B89A; font-weight:bold;">5. Toss Winner is Match Winner → Percentage</h5>

We want to find the percentage of matches where:

```text
TossWinner == WinningTeam
```

First create the condition.



````python
toss_winner_match_winner = df["TossWinner"] == df["WinningTeam"]

toss_winner_match_winner.head()
````

Now count how many matches satisfy this condition.

Because `True` behaves like `1` and `False` behaves like `0`, we can use `.sum()`.



````python
toss_winner_match_winner.sum()
````

Now find the total number of matches:



````python
total_matches = df.shape[0]

total_matches
````

Finally calculate the percentage:

```text
percentage =
(matches where toss winner = match winner / total matches) × 100
```



````python
percentage = (toss_winner_match_winner.sum() / total_matches) * 100

percentage
````

We can round the result:



````python
round(percentage, 2)
````

<h5 style="color:#78B89A; font-weight:bold;">6. Filtering using Numeric Conditions</h5>

Filtering is not limited to text columns.

For example, find matches won by more than 100 runs.



````python
df[df["WonBy"] == "Runs"]
````

The `Margin` column contains the margin. We can combine conditions to find matches won by more than 100 runs.



````python
df[
    (df["WonBy"] == "Runs") &
    (df["Margin"] > 100)
][["Season", "Team1", "Team2", "WinningTeam", "Margin"]]
````

<h5 style="color:#78B89A; font-weight:bold;">7. Filtering using OR `|`</h5>

Sometimes we want rows satisfying **either** condition.

Use `|` for OR.

For example, find matches where either CSK or Mumbai Indians won.



````python
df[
    (df["WinningTeam"] == "Chennai Super Kings") |
    (df["WinningTeam"] == "Mumbai Indians")
][["Season", "Team1", "Team2", "WinningTeam"]]
````

⭐

```python
&   → AND
|   → OR
~   → NOT
```



<h5 style="color:#78B89A; font-weight:bold;">8. Filtering using NOT `~`</h5>

`~` reverses a Boolean condition.

For example, find matches that did **not** have a Super Over.



````python
df[~(df["SuperOver"] == "Y")].head()
````

<h5 style="color:#78B89A; font-weight:bold;">9. Filtering with isin()</h5>

When we want to check whether a value belongs to a list, `isin()` is very useful.

For example, find matches won by CSK or RCB.



````python
teams = ["Chennai Super Kings", "Royal Challengers Bangalore"]

df[df["WinningTeam"].isin(teams)][
    ["Season", "Team1", "Team2", "WinningTeam"]
]
````

<h5 style="color:#78B89A; font-weight:bold;">10. Filtering with multiple conditions + selecting columns</h5>

Filtering and column selection can be combined.

For example:



````python
df[
    (df["City"] == "Kolkata") &
    (df["WinningTeam"] == "Chennai Super Kings")
][["Season", "Date", "Team1", "Team2", "WinningTeam"]]
````

<h5 style="color:#78B89A; font-weight:bold;">Quick Revision</h5>

| Requirement | Code |
|---|---|
| Condition | `df["col"] == value` |
| Filter rows | `df[df["col"] == value]` |
| Count filtered rows | `df[condition].shape[0]` |
| AND | `(condition1) & (condition2)` |
| OR | `(condition1) | (condition2)` |
| NOT | `~(condition)` |
| Multiple values | `df["col"].isin([...])` |
| Compare two columns | `df["col1"] == df["col2"]` |

<br>

⭐⭐⭐⭐⭐ **VVVV Important**

The basic pattern is:

```python
df[condition]
```

Example:

```python
df[df["WinningTeam"] == "Chennai Super Kings"]
```

For multiple conditions:

```python
df[
    (condition1) &
    (condition2)
]
```

This filtering concept is extremely important for **data analysis and ML preprocessing**.
