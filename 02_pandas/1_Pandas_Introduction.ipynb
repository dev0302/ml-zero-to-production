<h3 style="color:#6FA8DC; font-weight:bold">Pandas Introduction</h3>

Pandas is one of the most important Python libraries for **data handling and data analysis**.

It is mainly used when we have data in a **structured/tabular form**, such as rows and columns.

<br>

<h5 style="color:#78B89A; font-weight:bold;">What is Pandas?</h5>

**Pandas** is an open-source Python library used for **working with, manipulating, cleaning and analyzing data**.

It provides easy-to-use data structures that allow us to work with data in a table-like format.

For example:

| Name | Age | Marks |
|---|---:|---:|
| Dev | 20 | 85 |
| Rahul | 21 | 91 |
| Aman | 20 | 78 |

Pandas is especially useful because real-world data is usually not perfectly clean. It may contain:
- Missing values
- Duplicate records
- Different data types
- Incorrect or unwanted values
- Large amounts of data

Pandas provides functions to handle these situations efficiently.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Why is Pandas used?</h5>

The main reason we use Pandas is to make **data manipulation and analysis easier**.

**1. Working with tabular data**  
We can easily work with data arranged in rows and columns.

**2. Data cleaning**  
We can find and handle missing values, duplicate values and incorrect data.

**3. Data selection and filtering**  
We can select specific rows or columns and filter data based on conditions.

**4. Data analysis**  
We can calculate statistics, group data and find useful patterns.

**5. Reading and writing data**  
Pandas can work with common data formats such as CSV and Excel files.

**6. Preparing data for Machine Learning**  
Before giving data to an ML model, we usually need to clean, transform and prepare it. Pandas is heavily used for this preprocessing step.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Practical Applications of Pandas</h5>

Pandas is used in many real-world data-related applications.

**1. Data Analysis**  
Analyze datasets and find useful information from them.

**2. Data Cleaning**  
Remove duplicates, handle missing values and correct data.

**3. Machine Learning**  
Prepare datasets before training ML models.

**4. Business Analytics**  
Analyze sales, customers, revenue and expenses.

**5. Financial Data Analysis**  
Work with stock prices, transactions and other financial datasets.

**6. CSV / Excel Data Processing**  
Read data from files, modify it and save the processed data.

**7. Data Visualization Preparation**  
Prepare and organize data before creating graphs and visualizations.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Pandas in Machine Learning</h5>

Pandas is generally used in the **data preparation stage** of an ML workflow.

```text
Raw Dataset
     ↓
Load Data using Pandas
     ↓
Understand the Data
     ↓
Clean the Data
     ↓
Handle Missing Values
     ↓
Select / Transform Features
     ↓
Prepare X and y
     ↓
Machine Learning Model
```

Pandas is usually **not the ML model itself**.

It helps us prepare the data so that the ML algorithm can work with it.

For example:

```text
Age | Salary | Experience | Purchased
```

Pandas can help us clean this dataset and separate:

```text
Features (X) → Age, Salary, Experience
Target (y)   → Purchased
```

This makes Pandas extremely important in practical Machine Learning.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Pandas Series</h3>

<h5 style="color:#78B89A; font-weight:bold;">What is a Series?</h5>

A **Pandas Series** is a **one-dimensional labeled data structure**.

Simply:

**Series = One column of data + labels (index)**

Example:

```text
0    Dev
1    Rahul
2    Aman
```

Here:
- `0, 1, 2` → index
- `Dev, Rahul, Aman` → values

A Series can contain integers, floats, strings, Boolean values and other Python objects.

A Series is similar to a **single column in a table**.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Series Example</h5>

Imagine we have the marks of three students:

```text
Student Marks

Dev       85
Rahul     91
Aman      78
```

This can be represented as a Pandas Series.

The important idea is:

```text
Series
  ↓
One-dimensional
  ↓
Values + Index
```

So whenever we need to work with a **single column / one-dimensional labeled data**, a Series can be used.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Pandas DataFrame</h3>

<h5 style="color:#78B89A; font-weight:bold;">What is a DataFrame?</h5>

A **Pandas DataFrame** is a **two-dimensional labeled data structure**.

Simply:

**DataFrame = Rows + Columns**

It looks similar to a table or an Excel spreadsheet.

Example:

```text
       Name  Age  Marks
0       Dev   20     85
1     Rahul   21     91
2      Aman   20     78
```

Here:
- Rows represent individual records.
- Columns represent different attributes/features.
- Index identifies each row.

A DataFrame can contain **multiple columns**, and different columns can have different data types.

<br>

<h5 style="color:#78B89A; font-weight:bold;">DataFrame in Machine Learning</h5>

In Machine Learning, datasets are very commonly represented using a **DataFrame**.

For example:

```text
Age | Salary | Experience | Purchased
--------------------------------------
20  | 30000  | 1           | No
25  | 50000  | 3           | Yes
30  | 70000  | 6           | Yes
```

Here:

- Each **row** → one observation / record
- Each **column** → one feature or attribute
- `Purchased` → target variable

This is why DataFrames are so important when working with ML datasets.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Series vs DataFrame</h3>

| Series | DataFrame |
|---|---|
| One-dimensional | Two-dimensional |
| Similar to one column | Similar to an entire table |
| Contains values + index | Contains rows, columns + index |
| Usually represents a single variable | Can represent an entire dataset |

Easy way to remember:

```text
Series    → Single column
DataFrame → Complete table
```

A DataFrame can be thought of as a **collection of Series**, where each column is a Series.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Simple Real-World Example</h3>

Suppose a company has employee data:

```text
Name    Age    Salary    Department
Dev     20     50000     IT
Aman    22     60000     HR
Riya    21     55000     IT
```

Using Pandas:

- The complete table → **DataFrame**
- `Name` column → **Series**
- `Age` column → **Series**
- `Salary` column → **Series**
- `Department` column → **Series**

```text
DataFrame
    │
    ├── Name        → Series
    ├── Age         → Series
    ├── Salary      → Series
    └── Department  → Series
```

<h3 style="color:#6FA8DC; font-weight:bold">NumPy vs Pandas</h3>

We have already learned NumPy, so it is useful to understand the basic difference.

**NumPy**

Mainly used for:
- Numerical computing
- Arrays
- Mathematical operations
- Fast numerical calculations

**Pandas**

Mainly used for:
- Tabular data
- Data cleaning
- Data analysis
- Data manipulation
- Working with datasets

Simple way to remember:

```text
NumPy  → Numerical arrays
Pandas → Data / Tables
```

Pandas is built on top of NumPy and uses NumPy internally for many numerical operations.
