<h3 style="color:#6FA8DC; font-weight:bold">Pandas DataFrame → Introduction</h3>

A **DataFrame** is a 2-dimensional labelled data structure in Pandas.

It can be thought of like a table with **rows and columns**, similar to an Excel spreadsheet or a database table.

In Machine Learning, datasets are commonly represented as DataFrames before they are cleaned, analyzed and passed to ML algorithms.
<br>


<h5 style="color:#78B89A; font-weight:bold;">Creating DataFrame → using lists</h5>

A DataFrame can be created using lists inside a dictionary. Each key becomes a column.


```python
import numpy as np
import pandas as pd

students = pd.DataFrame({
    "Name": ["Dev", "Rahul", "Aman", "Riya"],
    "Age": [20, 21, 20, 22],
    "Marks": [85, 90, 78, 92]
})

students
```

<h5 style="color:#78B89A; font-weight:bold;">Creating DataFrame → using dictionary</h5>

A dictionary can directly represent the columns of a DataFrame.


```python
data = {
    "Name": ["Dev", "Aman", "Riya"],
    "Age": [20, 21, 22],
    "City": ["Delhi", "Mumbai", "Pune"]
}

df = pd.DataFrame(data)
df
```

<h5 style="color:#78B89A; font-weight:bold;">Creating DataFrame → using read_csv()</h5>

The most common practical way of creating a DataFrame is by reading a CSV dataset.

We will use the actual datasets with the `datasets/filename` path.


```python
subs_df = pd.read_csv("datasets/subs.csv")
kohli_df = pd.read_csv("datasets/kohli_ipl.csv")
bollywood_df = pd.read_csv("datasets/bollywood.csv")

print(subs_df.head())
print(kohli_df.head())
print(bollywood_df.head())
```

<h5 style="color:#78B89A; font-weight:bold;">DataFrame attributes → shape, dtypes and index</h5>

`shape` → number of rows and columns

`dtypes` → data type of each column

`index` → row labels


```python
print(kohli_df.shape)
print(kohli_df.dtypes)
print(kohli_df.index)
```

<h5 style="color:#78B89A; font-weight:bold;">Viewing Data → head(), tail(), sample()</h5>

`head()` → first rows

`tail()` → last rows

`sample()` → random rows


```python
print(kohli_df.head())
print(kohli_df.tail())
print(kohli_df.sample(5))
```

<h5 style="color:#78B89A; font-weight:bold;">info() → VVVV Important</h5>

`info()` gives a quick summary of the DataFrame.

It tells us:
- number of rows
- column names
- non-null values
- data types
- memory usage

⭐ One of the first methods to use when exploring a new dataset.


```python
kohli_df.info()
```

<h5 style="color:#78B89A; font-weight:bold;">describe() → statistical summary</h5>

`describe()` gives statistical information for numerical columns such as count, mean, standard deviation, minimum, percentiles and maximum.

⭐ Very important for quickly understanding numerical data.


```python
print(kohli_df.describe())
```

<h5 style="color:#78B89A; font-weight:bold;">isnull() and isnull().sum() → missing values</h5>

`isnull()` checks every cell and returns `True` where a value is missing.

`isnull().sum()` gives the number of missing values in each column.

⭐ `isnull().sum()` is extremely important during data cleaning.


```python
print(bollywood_df.isnull())
print(bollywood_df.isnull().sum())
```

<h5 style="color:#78B89A; font-weight:bold;">duplicated() → finding duplicate rows</h5>

`duplicated()` returns `True` for rows that are duplicates of an earlier row.

This is useful while checking data quality before analysis or ML.


```python
print(bollywood_df.duplicated())
print("Duplicate rows:", bollywood_df.duplicated().sum())
```

<h5 style="color:#78B89A; font-weight:bold;">rename() → changing column names</h5>

`rename()` is used to change column names or index labels.

By default, `rename()` returns a new DataFrame and does not modify the original DataFrame.

Using `inplace=True` modifies the existing DataFrame directly.


```python
df = kohli_df.copy()

df = df.rename(columns={"runs": "Runs"})
print(df.head())

df = df.rename(columns={
    "match_no": "Match Number",
    "Runs": "Runs Scored"
})
print(df.head())
```

### `inplace=True` → modify the existing DataFrame

Instead of creating and assigning a new DataFrame, we can modify the current DataFrame directly.


```python
df = kohli_df.copy()

df.rename(columns={"runs": "Runs Scored"}, inplace=True)

print(df.head())
```

<h3 style="color:#6FA8DC; font-weight:bold">DataFrame Mathematical Methods</h3>

Mathematical and statistical methods can be applied directly to DataFrames.

These methods are useful for understanding datasets and preparing numerical features for Machine Learning.
<br>


<h5 style="color:#78B89A; font-weight:bold;">sum() → sum of values</h5>

For a DataFrame, `sum()` works column-wise by default.


```python
math_df = pd.DataFrame({
    "Math": [80, 90, 70],
    "Science": [75, 85, 95],
    "English": [88, 92, 78]
})

print(math_df.sum())
```

### `axis=0` and `axis=1`

`axis=0` → operate column-wise

`axis=1` → operate row-wise

For mathematical methods, the default is usually `axis=0`.


```python
print(math_df.sum(axis=0))
print(math_df.sum(axis=1))
```

<h5 style="color:#78B89A; font-weight:bold;">mean(), median(), mode()</h5>

`mean()` → average

`median()` → middle value

`mode()` → most frequently occurring value(s)


```python
print(math_df.mean())
print(math_df.median())
print(math_df.mode())
```

<h5 style="color:#78B89A; font-weight:bold;">min(), max()</h5>

`min()` → minimum value

`max()` → maximum value


```python
print(math_df.min())
print(math_df.max())
```

<h5 style="color:#78B89A; font-weight:bold;">std() and var()</h5>

`std()` → standard deviation

`var()` → variance

Both help us understand the spread or variation of numerical data.


```python
print(math_df.std())
print(math_df.var())
```

<h5 style="color:#78B89A; font-weight:bold;">count() and prod()</h5>

`count()` → counts non-null values

`prod()` → calculates the product of values


```python
print(math_df.count())
print(math_df.prod())
```

<h5 style="color:#78B89A; font-weight:bold;">describe() → complete mathematical summary</h5>

`describe()` combines many important statistical calculations into one summary.

It includes `count`, `mean`, `std`, `min`, percentiles and `max`.

⭐⭐⭐⭐⭐ **VVVV Important for ML data exploration.**


```python
math_df.describe()
```
