<h3 style="color:#6FA8DC; font-weight:bold">Series Methods</h3>

We will use our actual datasets to practice important Series methods.

```text
datasets/subs.csv
datasets/kohli_ipl.csv
datasets/bollywood.csv
```

<br>

````python
import pandas as pd

subs = pd.read_csv("datasets/subs.csv")["Subscribers gained"]
runs = pd.read_csv("datasets/kohli_ipl.csv")["runs"]
lead = pd.read_csv("datasets/bollywood.csv")["lead"]

print(subs.head())
````

<h5 style="color:#78B89A; font-weight:bold;">head() → first values</h5>

Returns the first 5 values by default.

```python
series.head()
series.head(n)
```


````python
print(subs.head())
print(subs.head(3))
````

<h5 style="color:#78B89A; font-weight:bold;">tail() → last values</h5>

Returns the last 5 values by default.

```python
series.tail()
series.tail(n)
```


````python
print(subs.tail())
print(subs.tail(3))
````

<h5 style="color:#78B89A; font-weight:bold;">sample() → random values</h5>

Returns random values from the Series.

```python
series.sample()
series.sample(n)
```


````python
print(subs.sample())
print(subs.sample(3))
````

<h5 style="color:#78B89A; font-weight:bold;">value_counts() → frequency</h5>

Counts how many times each unique value occurs.

```python
series.value_counts()
```

⭐ Very useful for categorical data and class distributions in ML.


````python
print(lead.value_counts().head(10))
````

Here we use the `lead` column from `bollywood.csv`.

It tells us how many movies are associated with each lead actor.

<br>

<h5 style="color:#78B89A; font-weight:bold;">value_count vs value_counts</h5>

The correct Series method is:

```python
value_counts()
```

There is no normal Series method called `value_count()`.

<br>

<h5 style="color:#78B89A; font-weight:bold;">sort_values() → sort by values</h5>

```python
series.sort_values()
series.sort_values(ascending=False)
```

Default order is ascending.


````python
print(runs.sort_values().head())

# highest runs first
print(runs.sort_values(ascending=False).head())
````

<h5 style="color:#78B89A; font-weight:bold;">sort_values(inplace=True)</h5>

By default, the original Series is not changed.

```python
sorted_runs = runs.sort_values()
```

To modify the original:

```python
runs.sort_values(inplace=True)
```

`inplace=True` means **change the original Series itself**.


````python
runs_copy = runs.copy()

runs_copy.sort_values(inplace=True)

print(runs_copy.head())
````

<h5 style="color:#78B89A; font-weight:bold;">sort_index() → sort by index</h5>

```python
series.sort_index()
series.sort_index(ascending=False)
```

Remember:

```text
sort_values() → values
sort_index()  → index
```

<br>

````python
print(runs.sort_index().head())
print(runs.sort_index(ascending=False).head())
````

<h3 style="color:#6FA8DC; font-weight:bold">Series Maths Methods</h3>

Now we use the numerical `runs` Series from `kohli_ipl.csv`.

<br>

<h5 style="color:#78B89A; font-weight:bold;">sum() → total</h5>

```python
series.sum()
```


````python
print(runs.sum())
````

<h5 style="color:#78B89A; font-weight:bold;">count() → non-missing values</h5>

```python
series.count()
```

It counts only non-missing values.


````python
print(runs.count())
````

<h5 style="color:#78B89A; font-weight:bold;">size vs count()</h5>

| `size` | `count()` |
|---|---|
| Total elements | Non-missing elements |
| Includes missing values | Does not count missing values |
| Attribute | Method |

Example:
```python
s = pd.Series([10, 20, None, 40])
s.size    # 4
s.count() # 3
```


````python
s = pd.Series([10, 20, None, 40])

print("size  =", s.size)
print("count =", s.count())
````

<h5 style="color:#78B89A; font-weight:bold;">prod() → product</h5>

```python
series.prod()
```


````python
small = pd.Series([2, 3, 4])

print(small.prod())  # 24
````

<h5 style="color:#78B89A; font-weight:bold;">mean() → average</h5>

```python
series.mean()
```

⭐ Very important for ML.


````python
print(runs.mean())
````

<h5 style="color:#78B89A; font-weight:bold;">median() → middle value</h5>

```python
series.median()
```

Median is less affected by extreme values than mean.

⭐ Important for ML.


````python
print(runs.median())
````

<h5 style="color:#78B89A; font-weight:bold;">mode() → most frequent value</h5>

```python
series.mode()
```


````python
print(runs.mode())
````

<h5 style="color:#78B89A; font-weight:bold;">std() → standard deviation</h5>

```python
series.std()
```

Standard deviation tells us how spread out values are around the mean.

⭐ Very important for ML and feature scaling.


````python
print(runs.std())
````

<h5 style="color:#78B89A; font-weight:bold;">var() → variance</h5>

```python
series.var()
```

```text
Variance = (Standard Deviation)²
```


````python
print(runs.var())
````

<h5 style="color:#78B89A; font-weight:bold;">min() → minimum</h5>

```python
series.min()
```


````python
print(runs.min())
````

<h5 style="color:#78B89A; font-weight:bold;">max() → maximum</h5>

```python
series.max()
```

`min()` and `max()` help us understand the range of a feature.

⭐ Important for ML.


````python
print(runs.max())
````

<h5 style="color:#78B89A; font-weight:bold;">describe() → statistical summary</h5>

```python
series.describe()
```

⭐ **VVVV Important for ML**


````python
print(runs.describe())
````

For a numerical Series, `describe()` commonly gives:

```text
count
mean
std
min
25%
50%
75%
max
```

It gives a quick idea about the **basic statistics and distribution of a feature**.
