<h3 style="color:#6FA8DC; font-weight:bold">9. Pandas DataFrame → Fetching Rows and Columns</h3>

In a DataFrame, we often need to select:

- one column
- multiple columns
- one row
- multiple rows
- specific rows and columns together

We can mainly use:

- `[]`
- `iloc[]`
- `loc[]`

<br>


<h5 style="color:#78B89A; font-weight:bold;">First → Load a DataFrame</h5>

We will use the actual Kohli IPL dataset.



````python
import pandas as pd

df = pd.read_csv("datasets/kohli_ipl.csv")

df.head()
````

<h5 style="color:#78B89A; font-weight:bold;">1. Selecting a Single Column</h5>

Use the column name inside `[]`.

```python
df["runs"]
```

A **single column selected from a DataFrame becomes a Pandas Series**.

⭐ This is very important because many Series methods can then be used on the selected column.


````python
runs = df["runs"]

print(runs)
print(type(runs))
````

### Output type

The type will be:

```text
<class 'pandas.core.series.Series'>
```

So:

**DataFrame → select one column → Series**



<h5 style="color:#78B89A; font-weight:bold;">2. Selecting Multiple Columns</h5>

For multiple columns, pass a **list of column names**.

```python
df[["match_no", "runs"]]
```

Notice the double square brackets:

- outer `[]` → selecting from DataFrame
- inner `[]` → list of columns

When multiple columns are selected, the result is still a **DataFrame**.


````python
selected_cols = df[["match_no", "runs"]]

print(selected_cols)
print(type(selected_cols))
````

⭐ **Remember**

```python
df["runs"]                    # Series
df[["match_no", "runs"]]      # DataFrame
```



<h5 style="color:#78B89A; font-weight:bold;">3. Selecting Rows using iloc[]</h5>

`iloc` means **integer-location based indexing**.

It selects rows/columns using their **position numbers**.

Position starts from `0`.

```python
df.iloc[0]
```

This selects the first row.


````python
first_row = df.iloc[0]

print(first_row)
print(type(first_row))
````

A single row selected using `iloc` is returned as a **Series**.



<h5 style="color:#78B89A; font-weight:bold;">4. Selecting Multiple Rows using iloc[]</h5>

We can use slicing to select multiple consecutive rows.

```python
df.iloc[0:5]
```

This selects rows from position `0` to `4`.

The ending position `5` is not included.


````python
df.iloc[0:5]
````

<h5 style="color:#78B89A; font-weight:bold;">5. Fancy Indexing with iloc[]</h5>

Fancy indexing means selecting **specific positions** by passing a list.

```python
df.iloc[[0, 2, 4]]
```

This selects rows at positions `0`, `2`, and `4`.


````python
df.iloc[[0, 2, 4]]
````

<h5 style="color:#78B89A; font-weight:bold;">6. Selecting Rows using loc[]</h5>

`loc` means **label-based indexing**.

It selects using the **index labels**.

For the current DataFrame, the default index is:

```text
0, 1, 2, 3, 4, ...
```

So:

```python
df.loc[0]
```

selects the row whose index label is `0`.


````python
df.loc[0]
````

<h5 style="color:#78B89A; font-weight:bold;">7. Selecting Multiple Rows using loc[]</h5>

We can use a slice with `loc`.

```python
df.loc[0:4]
```

⚠️ Important difference:

With `loc`, the ending label is **included**.

So `0:4` selects:

```text
0, 1, 2, 3, 4
```


````python
df.loc[0:4]
````

<h5 style="color:#78B89A; font-weight:bold;">8. Fancy Indexing with loc[]</h5>

We can also provide a list of index labels.

```python
df.loc[[0, 2, 4]]
```

This selects rows with index labels `0`, `2`, and `4`.


````python
df.loc[[0, 2, 4]]
````

<h5 style="color:#78B89A; font-weight:bold;">9. iloc vs loc → Important Difference</h5>

`iloc` → works with **positions**

`loc` → works with **index labels**

With the default index, they may look the same:

```python
df.iloc[2]
df.loc[2]
```

But conceptually they are different.



````python
print("Using iloc:")
print(df.iloc[2])

print("\nUsing loc:")
print(df.loc[2])
````

### Slicing difference

```python
df.iloc[0:4]    # 0, 1, 2, 3
df.loc[0:4]     # 0, 1, 2, 3, 4
```

⭐ `iloc` → ending position excluded

⭐ `loc` → ending label included



<h5 style="color:#78B89A; font-weight:bold;">10. Selecting Both Rows and Columns using iloc[]</h5>

`iloc[row, column]`

We can select both the row position and column position together.

For example:

```python
df.iloc[0, 1]
```

means:

- row position `0`
- column position `1`



````python
value = df.iloc[0, 1]

print(value)
print(type(value))
````

### Selecting multiple rows and columns

```python
df.iloc[0:5, 0:2]
```

means:

- rows → positions `0` to `4`
- columns → positions `0` to `1`



````python
df.iloc[0:5, 0:2]
````

### Fancy indexing with rows and columns

We can select specific row positions and specific column positions.

```python
df.iloc[[0, 2, 4], [0, 1]]
```



````python
df.iloc[[0, 2, 4], [0, 1]]
````

<h5 style="color:#78B89A; font-weight:bold;">11. Selecting Both Rows and Columns using loc[]</h5>

`loc` also allows us to select rows and columns together.

General form:

```python
df.loc[row_labels, column_labels]
```

For example:

```python
df.loc[0, "runs"]
```

means:

- row label → `0`
- column label → `"runs"`



````python
value = df.loc[0, "runs"]

print(value)
print(type(value))
````

### Selecting multiple rows and columns

```python
df.loc[0:4, ["match_no", "runs"]]
```

This selects:

- rows with labels `0` to `4`
- columns `"match_no"` and `"runs"`



````python
df.loc[0:4, ["match_no", "runs"]]
````

### Fancy indexing with loc[]

We can select specific row labels and specific column labels.

```python
df.loc[[0, 2, 4], ["match_no", "runs"]]
```



````python
df.loc[[0, 2, 4], ["match_no", "runs"]]
````

<h5 style="color:#78B89A; font-weight:bold;">12. Selecting Columns using iloc[]</h5>

`iloc` can also be used when we want columns by their **position**.

For example, column position `1` is the `runs` column.



````python
df.iloc[:, 1]
````

Here:

```python
:
```

means **all rows**.

So:

```python
df.iloc[:, 1]
```

means:

**all rows + column at position 1**

Similarly, multiple columns:

```python
df.iloc[:, [0, 1]]
```



````python
df.iloc[:, [0, 1]]
````

<h5 style="color:#78B89A; font-weight:bold;">13. Selecting Columns using loc[]</h5>

With `loc`, we select columns using their **labels**.

```python
df.loc[:, "runs"]
```

`:` means all rows.



````python
df.loc[:, "runs"]
````

For multiple columns:

```python
df.loc[:, ["match_no", "runs"]]
```



````python
df.loc[:, ["match_no", "runs"]]
````

<h5 style="color:#78B89A; font-weight:bold;">14. Quick Revision</h5>

| Task | Syntax | Result |
|---|---|---|
| Single column | `df["runs"]` | Series |
| Multiple columns | `df[["match_no", "runs"]]` | DataFrame |
| Single row | `df.iloc[0]` | Series |
| Multiple rows | `df.iloc[0:5]` | DataFrame |
| Fancy rows | `df.iloc[[0,2,4]]` | DataFrame |
| Single row by label | `df.loc[0]` | Series |
| Multiple rows by label | `df.loc[0:4]` | DataFrame |
| Fancy rows by label | `df.loc[[0,2,4]]` | DataFrame |
| Row + column | `df.iloc[0,1]` | Single value |
| Rows + columns | `df.iloc[0:5,0:2]` | DataFrame |
| Row + named column | `df.loc[0,"runs"]` | Single value |
| Rows + named columns | `df.loc[0:4,["match_no","runs"]]` | DataFrame |

<br>

⭐⭐⭐⭐⭐ **VVVV Important**

```python
df["col"]                    # single column → Series
df[["col1", "col2"]]         # multiple columns → DataFrame

df.iloc[...]                 # position based
df.loc[...]                  # label based
```



<h5 style="color:#78B89A; font-weight:bold;">15. One Simple Way to Remember</h5>

Think of a DataFrame like a table.

```text
             Columns →
        match_no     runs
Rows ↓
   0        1          1
   1        2         23
   2        3         13
   3        4         12
```

### `iloc`

Ask:

> **Where is it?**

Use row/column **positions**.

### `loc`

Ask:

> **What is its label/name?**

Use **index labels and column names**.
