<h3 style="color:#6FA8DC; font-weight:bold">Series with Python functionalities</h3>

A Pandas Series behaves a lot like a normal Python sequence, so many Python functions and operators can be used directly on it.

Here we will see Python functions, type conversions, membership operators, looping, arithmetic operators, relational operators and Boolean indexing on Series.
<br>

<h5 style="color:#78B89A; font-weight:bold;">Creating a Series for examples</h5>

We will use the actual datasets while learning these concepts.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Actual datasets
subs = pd.read_csv("datasets/subs.csv")["Subscribers gained"]
vk = pd.read_csv("datasets/kohli_ipl.csv")["runs"]
lead = pd.read_csv("datasets/bollywood.csv")["lead"]

print(subs.head())
print(vk.head())
print(lead.head())
```

<h5 style="color:#78B89A; font-weight:bold;">Python functions → len(), type(), dir(), sorted(), max(), min()</h5>

These normal Python functions can also work with a Pandas Series.
<br>

```python
# len() -> number of elements
print(len(vk))

# type() -> tells the type of object
print(type(vk))

# dir() -> shows available attributes and methods
print(dir(vk))

# sorted() -> returns a sorted Python list
print(sorted(vk)[:10])

# max() and min() -> maximum and minimum value
print(max(vk))
print(min(vk))
```

### Important difference → sorted() vs sort_values()

- `sorted(vk)` returns a **Python list**.
- `vk.sort_values()` returns a **Pandas Series**.

So, both can sort values, but their return types are different.

```python
print(type(sorted(vk)))
print(type(vk.sort_values()))
```

<h5 style="color:#78B89A; font-weight:bold;">Type conversions → converting Series into other useful types</h5>

A Series can be converted into different Python / NumPy / Pandas forms depending on the requirement.
<br>

```python
# Series -> Python list
vk_list = vk.tolist()
print(vk_list[:10])
print(type(vk_list))

# Series -> NumPy array
vk_array = vk.to_numpy()
print(vk_array[:10])
print(type(vk_array))

# Series -> dictionary
vk_dict = vk.to_dict()
print(list(vk_dict.items())[:5])
print(type(vk_dict))

# Series -> DataFrame
vk_df = vk.to_frame()
print(vk_df.head())
print(type(vk_df))
```

### `astype()` → changing the data type of Series values

`astype()` is used when we specifically want to change the dtype of a Series.

Common conversions:
- integer → float
- integer → string
- integer → boolean
- string → integer / float (when values are numeric)

⭐ `astype()` is very important when preparing data for ML.

```python
# int -> float
vk_float = vk.astype(float)
print(vk_float.head())
print(vk_float.dtype)

# int -> string
vk_str = vk.astype(str)
print(vk_str.head())
print(vk_str.dtype)

# int -> boolean
# 0 becomes False, non-zero values become True
vk_bool = vk.astype(bool)
print(vk_bool.head())
print(vk_bool.dtype)
```

```python
# String numbers -> integer / float
numbers = pd.Series(["10", "20", "30", "40"])

numbers_int = numbers.astype(int)
numbers_float = numbers.astype(float)

print(numbers_int)
print(numbers_int.dtype)
print(numbers_float)
print(numbers_float.dtype)
```

### Other useful conversion functions

`pd.to_numeric()` is useful when values may contain numbers stored as strings. It can also handle invalid values using `errors="coerce"`.

`pd.to_datetime()` converts date-like values into Pandas datetime values.

`pd.to_timedelta()` converts values into Pandas time durations.


```python
# String numbers -> numeric values
numbers = pd.Series(["10", "20", "30", "abc"])
numeric = pd.to_numeric(numbers, errors="coerce")
print(numeric)

# Strings -> datetime
dates = pd.Series(["2026-01-01", "2026-02-01", "2026-03-01"])
dates = pd.to_datetime(dates)
print(dates)
print(dates.dtype)

# Strings -> timedelta
duration = pd.Series(["1 day", "2 days", "3 days"])
duration = pd.to_timedelta(duration)
print(duration)
print(duration.dtype)
```

<h5 style="color:#78B89A; font-weight:bold;">Membership operator → `in` operator</h5>

The `in` operator checks **index labels** of a Series by default, not its values.

For checking whether a value exists, use `value in series.values` or `value in series.to_numpy()`.
<br>

```python
s = pd.Series([10, 20, 30], index=["a", "b", "c"])

# Checks INDEX labels
print("a" in s)
print("x" in s)

# Checks VALUES
print(20 in s.values)
print(50 in s.values)
```

<h5 style="color:#78B89A; font-weight:bold;">Looping → values and indexes</h5>

A Series can be looped through like a normal Python sequence.

When we use `for value in s`, we get **values**.
To loop through indexes, use `s.index`.
<br>

```python
s = pd.Series([10, 20, 30], index=["a", "b", "c"])

# Loop through values
for value in s:
    print(value)

# Loop through indexes
for index in s.index:
    print(index)

# Loop through both index and value
for index, value in s.items():
    print(index, value)
```

<h5 style="color:#78B89A; font-weight:bold;">Arithmetic operators → broadcasting</h5>

Arithmetic operators such as `+`, `-`, `*`, `/`, `//`, `%` and `**` can be directly applied to a Series.

When a single number is used with a Series, Pandas applies that operation to **every element**. This is called **broadcasting**.
<br>

```python
vk_small = vk.head(5)
print(vk_small)

print(vk_small + 10)   # add 10 to every value
print(vk_small - 10)   # subtract 10 from every value
print(vk_small * 2)    # multiply every value by 2
print(vk_small / 2)    # divide every value by 2
print(vk_small // 2)   # floor division
print(vk_small % 2)    # remainder
print(vk_small ** 2)   # square every value
```

### Arithmetic between two Series

When two Series are used together, Pandas performs the operation using their **index labels**.

This is one of the important differences between Pandas and a normal NumPy array.

```python
a = pd.Series([10, 20, 30], index=["a", "b", "c"])
b = pd.Series([1, 2, 3], index=["a", "b", "c"])

print(a + b)
print(a * b)
```

<h5 style="color:#78B89A; font-weight:bold;">Relational operators → comparing Series values</h5>

Operators like `>`, `<`, `>=`, `<=`, `==` and `!=` compare every value of a Series and return a **Boolean Series**.
<br>

```python
vk_small = vk.head(10)

print(vk_small > 20)
print(vk_small < 20)
print(vk_small >= 20)
print(vk_small <= 20)
print(vk_small == 20)
print(vk_small != 20)
```

<h3 style="color:#6FA8DC; font-weight:bold">Boolean Indexing on Series</h3>

Boolean indexing means using a Boolean Series (`True` / `False`) as a filter to select only the required values.

This is one of the **most important Pandas concepts for ML and data analysis**.
<br>

<h5 style="color:#78B89A; font-weight:bold;">`vk >= 50` → create a Boolean condition</h5>

The comparison does not immediately give only the matching values. It first creates a Boolean Series of the same length.

```python
vk >= 50
```

<h5 style="color:#78B89A; font-weight:bold;">`vk[vk >= 50]` → filter the Series</h5>

The Boolean Series is passed inside `[]` to select only the rows where the condition is `True`.

```python
vk[vk >= 50]
```

<h5 style="color:#78B89A; font-weight:bold;">Finding ducks → `vk[vk == 0]`</h5>

In cricket, a score of `0` is called a duck.

So `vk[vk == 0]` gives all the innings in which Virat Kohli scored 0 runs.

To find the **number of ducks**, use `.size` or `.count()` on the filtered Series.

```python
# All ducks
ducks = vk[vk == 0]
print(ducks)

# Number of ducks
print("Number of ducks:", ducks.size)

# Same result using count()
print("Number of ducks:", ducks.count())
```

<h5 style="color:#78B89A; font-weight:bold;">More Boolean indexing examples</h5>

We can create many useful filters by changing the condition.

```python
# Runs greater than 100
print(vk[vk > 100])

# Runs between 50 and 100
print(vk[(vk >= 50) & (vk <= 100)])

# Runs less than 10
print(vk[vk < 10])

# Count innings with 50 or more runs
print((vk >= 50).sum())

# Count innings with 0 runs
print((vk == 0).sum())
```

### Multiple conditions → `&` and `|`

For Pandas Series:
- `&` → AND
- `|` → OR
- `~` → NOT

Put each condition inside parentheses.

❌ Do not use Python `and` / `or` for element-wise Series conditions.

```python
# AND -> runs between 50 and 100
print(vk[(vk >= 50) & (vk <= 100)])

# OR -> runs are either 0 or greater than 100
print(vk[(vk == 0) | (vk > 100)])

# NOT -> runs that are not 0
print(vk[~(vk == 0)])
```

<h3 style="color:#6FA8DC; font-weight:bold">Plotting graph on Series</h3>

A Series can be directly plotted using Pandas' `.plot()` method.

For a Series, the **index is generally used on the x-axis** and the **values are used on the y-axis**.

⭐ Plotting is very useful in ML/data analysis for understanding trends, distributions and outliers.
<br>

```python
# Line graph of subscribers gained
subs.plot()
plt.title("Subscribers Gained")
plt.xlabel("Day")
plt.ylabel("Subscribers")
plt.show()
```

### Plotting a selected part of a Series

For large Series, plotting the complete data may look crowded. We can select a smaller portion first and then plot it.

```python
# First 30 IPL matches
vk.head(30).plot()
plt.title("Virat Kohli Runs - First 30 Matches")
plt.xlabel("Match Index")
plt.ylabel("Runs")
plt.show()
```

### Bar graph on a Series

A Series can also be displayed as a bar graph using `.plot(kind="bar")`.

This is useful when comparing values for individual categories.

```python
# Plot a small sample so that labels remain readable
vk.head(10).plot(kind="bar")
plt.title("Virat Kohli Runs - First 10 Matches")
plt.xlabel("Match Index")
plt.ylabel("Runs")
plt.show()
```
