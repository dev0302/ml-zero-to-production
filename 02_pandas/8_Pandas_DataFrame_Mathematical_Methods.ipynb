<h3 style="color:#6FA8DC; font-weight:bold">DataFrame Mathematical Methods</h3>

Pandas provides many mathematical and statistical methods that can be directly applied to a DataFrame.

These methods are useful for understanding numerical data and are frequently used during data analysis and Machine Learning.

We will use a small DataFrame first so that the calculations are easy to understand.
<br>


<h5 style="color:#78B89A; font-weight:bold;">Creating a numerical DataFrame</h5>

We will use marks of students in different subjects.


````python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Math": [80, 90, 70, 60, 100],
    "Science": [75, 85, 95, 65, 90],
    "English": [88, 92, 78, 72, 85]
})

df
````

<h5 style="color:#78B89A; font-weight:bold;">sum() → sum of values</h5>

`sum()` adds the values of each column.

By default, `axis=0`, so the operation is performed **column-wise**.


````python
df.sum()
````

### `sum(axis=1)` → row-wise sum

When `axis=1` is used, the values are added **row-wise**.

So each student gets a total of all subject marks.


````python
df.sum(axis=1)
````

### `axis=0` vs `axis=1`

```text
axis=0 → column-wise → down the rows
axis=1 → row-wise    → across the columns
```

This concept is extremely important when working with DataFrames.


````python
print("Column-wise:")
print(df.sum(axis=0))

print("\nRow-wise:")
print(df.sum(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">mean() → average</h5>

`mean()` calculates the average of each column by default.

With `axis=1`, it can calculate the average for each row.


````python
print("Column-wise mean:")
print(df.mean())

print("\nRow-wise mean:")
print(df.mean(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">median() → middle value</h5>

`median()` returns the middle value after sorting the data.

It is less affected by extreme values compared with the mean.


````python
print("Column-wise median:")
print(df.median())

print("\nRow-wise median:")
print(df.median(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">mode() → most frequent value</h5>

`mode()` returns the most frequently occurring value.

A column can have more than one mode.


````python
print(df.mode())

print("\nRow-wise mode:")
print(df.mode(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">min() → minimum value</h5>

`min()` finds the smallest value in each column.

It can also be applied row-wise using `axis=1`.


````python
print("Column-wise minimum:")
print(df.min())

print("\nRow-wise minimum:")
print(df.min(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">max() → maximum value</h5>

`max()` finds the largest value in each column.

It can also be applied row-wise.


````python
print("Column-wise maximum:")
print(df.max())

print("\nRow-wise maximum:")
print(df.max(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">count() → count of non-null values</h5>

`count()` counts the number of **non-null values** in each column.

This is useful when checking how much usable data is available.


````python
print(df.count())
print("\nRow-wise count:")
print(df.count(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">prod() → product of values</h5>

`prod()` multiplies all values together.

It is less common in ML than `sum()` or `mean()`, but it is still a useful mathematical aggregation method.


````python
print(df.prod())

print("\nRow-wise product:")
print(df.prod(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">std() → standard deviation</h5>

`std()` measures how spread out the values are from their mean.

A higher standard deviation generally means greater variation.


````python
print("Column-wise standard deviation:")
print(df.std())

print("\nRow-wise standard deviation:")
print(df.std(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">var() → variance</h5>

`var()` calculates variance.

Variance is related to standard deviation:

```text
variance = (standard deviation)²
```


````python
print("Column-wise variance:")
print(df.var())

print("\nRow-wise variance:")
print(df.var(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">quantile() → percentile values</h5>

`quantile()` returns a specified percentile/quantile of the data.

For example:
- `0.25` → 25th percentile
- `0.50` → 50th percentile
- `0.75` → 75th percentile


````python
print("25th percentile:")
print(df.quantile(0.25))

print("\n50th percentile:")
print(df.quantile(0.50))

print("\n75th percentile:")
print(df.quantile(0.75))
````

<h5 style="color:#78B89A; font-weight:bold;">cumsum() → cumulative sum</h5>

`cumsum()` calculates a running cumulative sum.

Each value is added to all previous values.


````python
print(df.cumsum())

print("\nRow-wise cumulative sum:")
print(df.cumsum(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">cumprod() → cumulative product</h5>

`cumprod()` calculates the running product of values.


````python
print(df.cumprod())

print("\nRow-wise cumulative product:")
print(df.cumprod(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">cummin() and cummax()</h5>

`cummin()` → running minimum

`cummax()` → running maximum


````python
print("Cumulative minimum:")
print(df.cummin())

print("\nCumulative maximum:")
print(df.cummax())
````

<h5 style="color:#78B89A; font-weight:bold;">diff() → difference between consecutive values</h5>

`diff()` calculates the difference between the current value and the previous value.

The first value is `NaN` because there is no previous value.


````python
print(df.diff())

print("\nRow-wise difference:")
print(df.diff(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">pct_change() → percentage change</h5>

`pct_change()` calculates the percentage change from the previous value.

It is useful for analyzing how values increase or decrease over time.


````python
print(df.pct_change())

print("\nRow-wise percentage change:")
print(df.pct_change(axis=1))
````

<h5 style="color:#78B89A; font-weight:bold;">describe() → complete statistical summary</h5>

`describe()` gives a compact statistical summary of numerical columns.

It includes:

- `count`
- `mean`
- `std`
- `min`
- `25%`
- `50%`
- `75%`
- `max`

⭐⭐⭐⭐⭐ **VVVV Important for Machine Learning data analysis.**


````python
df.describe()
````

<h3 style="color:#6FA8DC; font-weight:bold">Using Mathematical Methods on Actual Dataset</h3>

Now we will apply some of the most useful mathematical methods to the actual `kohli_ipl.csv` dataset.


````python
kohli_df = pd.read_csv("datasets/kohli_ipl.csv")

kohli_df.head()
````

````python
print("Total runs:", kohli_df["runs"].sum())
print("Average runs:", kohli_df["runs"].mean())
print("Median runs:", kohli_df["runs"].median())
print("Minimum runs:", kohli_df["runs"].min())
print("Maximum runs:", kohli_df["runs"].max())
print("Standard deviation:", kohli_df["runs"].std())
````
