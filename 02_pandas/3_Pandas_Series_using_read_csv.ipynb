<h3 style="color:#6FA8DC; font-weight:bold">Series using read_csv()</h3>

In real-world projects, data is usually stored in CSV files.

For our examples we will use:

```text
datasets/subs.csv
datasets/kohli_ipl.csv
datasets/bollywood.csv
```

<br>

````python
import pandas as pd
````

<h5 style="color:#78B89A; font-weight:bold;">Project Folder Structure</h5>

We assume:

```text
ML/
└── 02_Pandas/
    ├── 3_Pandas_Series_using_read_csv.ipynb
    └── datasets/
        ├── subs.csv
        ├── kohli_ipl.csv
        └── bollywood.csv
```

Therefore the path format is:

```text
datasets/filename
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">How to import a CSV file?</h5>

**Syntax:**

```python
pd.read_csv("file_path")
```

Example:

```python
df = pd.read_csv("datasets/subs.csv")
```

`read_csv()` normally returns a **DataFrame**.


````python
subs_df = pd.read_csv("datasets/subs.csv")

print(subs_df.head())
````

Our `subs.csv` dataset contains the column:

```text
Subscribers gained
```

So we can select that column to get a Series.

<br>

<h5 style="color:#78B89A; font-weight:bold;">CSV → DataFrame → Series</h5>

```python
df = pd.read_csv("datasets/subs.csv")

subs = df["Subscribers gained"]
```

Flow:

```text
CSV
 ↓
pd.read_csv()
 ↓
DataFrame
 ↓
df["column"]
 ↓
Series
```


````python
subs_df = pd.read_csv("datasets/subs.csv")

subs = subs_df["Subscribers gained"]

print(subs.head())
print(type(subs))
````

<h5 style="color:#78B89A; font-weight:bold;">Reading kohli_ipl.csv</h5>

The dataset contains:

```text
match_no | runs
```


````python
kohli = pd.read_csv("datasets/kohli_ipl.csv")

print(kohli.head())

runs = kohli["runs"]

print(runs.head())
print(type(runs))
````

Here:

```text
kohli → DataFrame
runs  → Series
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Reading bollywood.csv</h5>

The dataset contains:

```text
movie | lead
```


````python
bollywood = pd.read_csv("datasets/bollywood.csv")

print(bollywood.head())

lead = bollywood["lead"]

print(lead.head())
print(type(lead))
````

Here:

```text
bollywood → DataFrame
lead      → Series
```

<h5 style="color:#78B89A; font-weight:bold;">If dataset folder is outside 02_Pandas</h5>

For comparison, if the structure were:


then we would use:


`..` means **go one folder up**.

But for our current project we use:

```python

```

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Quick Summary</h3>

Use these paths in our current project:

```python
pd.read_csv("datasets/subs.csv")
pd.read_csv("datasets/kohli_ipl.csv")
pd.read_csv("datasets/bollywood.csv")
```

To get a Series:

```python
df = pd.read_csv("datasets/filename.csv")
series = df["column_name"]
```

⭐ This **CSV → DataFrame → Series** workflow is very important for working with real datasets and ML.
