<br>

<h3 style="color:#6FA8DC; font-weight:bold">Series Indexing</h3>

Indexing means **accessing values from a Pandas Series**.

```text
s[index]       → label-based indexing
s.iloc[pos]    → position-based indexing
```

We will also learn:
- Slicing
- Negative slicing
- Fancy indexing
- Indexing with labels
- Editing Series using indexing, slicing, fancy indexing and labels

<br>

````python
import pandas as pd

# Actual datasets
subs = pd.read_csv("datasets/subs.csv")["Subscribers gained"]
vk = pd.read_csv("datasets/kohli_ipl.csv")["runs"]
lead = pd.read_csv("datasets/bollywood.csv")["lead"]

print(subs.head())
print(vk.head())
print(lead.head())
````

<h5 style="color:#78B89A; font-weight:bold;">s[index] → label-based indexing</h5>

When a Series has custom labels, `s[index]` accesses a value using its **index label**.

**Syntax:**

```python
s[index]
```


````python
print(s["a"])  # 10
print(s["c"])  # 30
print(s["e"])  # 50
````

Here `"a"`, `"c"` and `"e"` are **labels**, not positions.

```text
Index → a   b   c   d   e
Value → 10  20  30  40  50
```

So:

```python
s["c"]
```

means **find label `"c"` and return its value**.

<br>

<h5 style="color:#78B89A; font-weight:bold;">s.iloc[pos] → position-based indexing</h5>

`iloc` is used when we want to access a value according to its **position**.

**Syntax:**

```python
s.iloc[pos]
```

Positions start from `0`.


````python
print(s.iloc[0])  # 10
print(s.iloc[2])  # 30
print(s.iloc[4])  # 50
````

So:

```python
s.iloc[2]
```

means **go to position `2` and return the value there**.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Positive and Negative Positional Indexing</h5>

Use `iloc` for reliable **positional indexing**.

```python
s.iloc[0]   # first element
s.iloc[1]   # second element
s.iloc[2]   # third element
```

Negative positional indexing also works:

```python
s.iloc[-1]  # last element
s.iloc[-2]  # second-last element
```


````python
print(s.iloc[-1])  # 50
print(s.iloc[-2])  # 40
````

### Important

Do **not** rely on:

```python
s[-1]
```

for positional indexing.

Use:

```python
s.iloc[-1]
```

when you mean **last position**.

This avoids confusion between **labels and positions**.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Series Slicing</h3>

Slicing is used to access a **range of values**.

For positional slicing, use `iloc`.

**Syntax:**

```python
s.iloc[start:stop]
```

The `stop` position is **not included**.


````python
s = pd.Series([10, 20, 30, 40, 50])

print(s.iloc[1:4])
````

Here:

```python
s.iloc[1:4]
```

selects positions `1, 2, 3`.

So we get:

```text
20
30
40
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Negative Slicing</h5>

We can use negative positions while slicing.

```python
s.iloc[-3:]   # last 3 elements
s.iloc[:-1]   # everything except the last element
```


````python
print(s.iloc[-3:])
print(s.iloc[:-1])
````

Negative positions count from the end:

```text
Position →  0   1   2   3   4
Negative → -5  -4  -3  -2  -1
Value    → 10  20  30  40  50
```

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Fancy Indexing</h3>

Fancy indexing means selecting **multiple specific positions or labels** at once.

For positional fancy indexing:

```python
s.iloc[[pos1, pos2, pos3]]
```


````python
s = pd.Series([10, 20, 30, 40, 50])

print(s.iloc[[0, 2, 4]])
````

Here we selected positions:

```text
0 → 10
2 → 30
4 → 50
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Fancy Indexing with Labels</h5>

We can select multiple labels by passing a list of labels.

**Syntax:**

```python
s[["label1", "label2"]]
```


````python
s = pd.Series(
    [10, 20, 30, 40, 50],
    index=["a", "b", "c", "d", "e"]
)

print(s[["a", "c", "e"]])
````

Here we directly selected the labels:

```text
a → 10
c → 30
e → 50
```

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Indexing with Labels</h3>

When a Series has meaningful labels, we can directly use those labels to access values.

```python
s["a"]
s["c"]
```

We can also use `loc` for explicit label-based indexing:

```python
s.loc["a"]
s.loc[["a", "c"]]
```

Remember:

```text
s["label"]        → label-based access
s.loc["label"]    → explicitly label-based
s.iloc[position]  → position-based
```

⭐ `loc` and `iloc` become very important when working with DataFrames and ML datasets.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Editing a Series</h3>

Indexing can also be used to **modify values** in a Series.

We can edit using:
- Indexing
- Slicing
- Fancy indexing
- Index labels

<br>

<h5 style="color:#78B89A; font-weight:bold;">Editing using Indexing</h5>

We can change a value at a particular position using `iloc`.

**Syntax:**

```python
s.iloc[position] = new_value
```


````python
# Work on a copy of the actual Kohli Series
s = vk.head(5).copy()
print(s)

# Editing using position
s.iloc[2] = 100
print(s)

# Editing using a slice
s.iloc[1:4] = 0
print(s)

# Editing using fancy indexing
s.iloc[[0, 2, 4]] = 100
print(s)

# Editing using index label
s.loc[1] = 200
print(s)
````

The value at position `2` was `30`, and we changed it to `100`.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Editing using Slicing</h5>

We can change multiple consecutive values using slicing.

**Syntax:**

```python
s.iloc[start:stop] = new_value
```


````python
s = pd.Series([10, 20, 30, 40, 50])

s.iloc[1:4] = 0

print(s)
````

Positions `1, 2, 3` are changed:

```text
Before → 10  20  30  40  50
After  → 10   0   0   0  50
```

We can also assign different values:

```python
s.iloc[1:4] = [200, 300, 400]
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Editing using Fancy Indexing</h5>

We can modify multiple specific positions at once.

**Syntax:**

```python
s.iloc[[pos1, pos2]] = new_value
```


````python
s = pd.Series([10, 20, 30, 40, 50])

s.iloc[[0, 2, 4]] = 100

print(s)
````

Only positions `0`, `2` and `4` are changed.

```text
Before → 10  20  30  40  50
After  → 100 20  100 40  100
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Editing using Index Labels</h5>

We can directly modify a value using its index label.

**Syntax:**

```python
s["label"] = new_value
```

or:

```python
s.loc["label"] = new_value
```


````python
s = pd.Series(
    [10, 20, 30, 40],
    index=["a", "b", "c", "d"]
)

s["b"] = 200
s.loc["d"] = 400

print(s)
````

Here:

```python
s["b"] = 200
```

changes the value associated with label `"b"`.

And:

```python
s.loc["d"] = 400
```

changes the value associated with label `"d"`.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Quick Summary</h3>

### Accessing values

```text
s[label]          → label-based indexing
s.iloc[position]  → position-based indexing

s.iloc[-1]        → last element
s.iloc[-2]        → second-last element
```

Do not rely on:

```python
s[-1]
```

for positional indexing. Use `s.iloc[-1]`.

### Slicing

```python
s.iloc[1:4]
s.iloc[-3:]
s.iloc[:-1]
```

### Fancy indexing

```python
s.iloc[[0, 2, 4]]
s[["a", "c", "e"]]
```

### Editing

```python
s.iloc[2] = 100
s.iloc[1:4] = 0
s.iloc[[0, 2]] = 100
s["a"] = 500
s.loc["a"] = 500
```

⭐ **Important for ML:** Understanding `loc`, `iloc`, slicing and fancy indexing is essential because later we will use them heavily with **DataFrames, feature selection and dataset preprocessing**.
