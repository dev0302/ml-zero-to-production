<h3 style="color:#6FA8DC; font-weight:bold">Pandas Series Basics</h3>

A **Pandas Series** is a one-dimensional labeled data structure.

We will learn:
- Creating Series from lists
- Creating Series from NumPy arrays
- Array + index
- Array + index + name
- Series from dictionary
- Series attributes

<br>

````python
import numpy as np
import pandas as pd
````

<h3 style="color:#6FA8DC; font-weight:bold">Creating Series</h3>

<h5 style="color:#78B89A; font-weight:bold;">Series from List</h5>

**Syntax:**
```python
pd.Series(list)
```

Pandas automatically creates the index starting from `0`.


````python
marks = [85, 91, 78, 88]

s1 = pd.Series(marks)

print(s1)
````

Here the list values become the Series values and Pandas creates the default index `0, 1, 2, 3`.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Series from NumPy Array</h5>

We can create a Series directly from a NumPy array.

**Syntax:**
```python
pd.Series(numpy_array)
```


````python
a = np.array([10, 20, 30, 40])

s2 = pd.Series(a)

print(s2)
````

The NumPy array values become the values of the Series.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Series from Array + Index</h5>

We can provide our own index.

**Syntax:**
```python
pd.Series(data, index=index)
```


````python
marks = [85, 91, 78, 88]
students = ["Dev", "Rahul", "Aman", "Riya"]

s3 = pd.Series(marks, index=students)

print(s3)
````

Now the labels are meaningful:

```text
Dev      → 85
Rahul    → 91
Aman     → 78
Riya     → 88
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Series from Array + Index + Name</h5>

We can also give a name to the Series.

**Syntax:**
```python
pd.Series(data, index=index, name="name")
```


````python
marks = [85, 91, 78, 88]
students = ["Dev", "Rahul", "Aman", "Riya"]

s4 = pd.Series(marks, index=students, name="Marks")

print(s4)
````

Here:
- `marks` → values
- `students` → index
- `"Marks"` → Series name

<br>

<h5 style="color:#78B89A; font-weight:bold;">Series from Dictionary</h5>

Dictionary keys become the index and dictionary values become the Series values.

**Syntax:**
```python
pd.Series(dictionary)
```


````python
marks = {
    "Dev": 85,
    "Rahul": 91,
    "Aman": 78,
    "Riya": 88
}

s5 = pd.Series(marks)

print(s5)
````

<h3 style="color:#6FA8DC; font-weight:bold">Series Attributes</h3>

Important attributes:

- `size`
- `dtype`
- `name`
- `is_unique`
- `index`
- `values`

<br>

<h5 style="color:#78B89A; font-weight:bold;">size → total elements</h5>

```python
series.size
```


````python
s = pd.Series([10, 20, 30, 40, 50])

print(s.size)  # 5
````

<h5 style="color:#78B89A; font-weight:bold;">dtype → data type</h5>

```python
series.dtype
```


````python
s = pd.Series([10, 20, 30, 40])

print(s.dtype)
````

<h5 style="color:#78B89A; font-weight:bold;">name → Series name</h5>

```python
series.name
```


````python
s = pd.Series([85, 91, 78], name="Marks")

print(s.name)  # Marks
````

<h5 style="color:#78B89A; font-weight:bold;">is_unique → checks uniqueness</h5>

```python
series.is_unique
```


````python
s1 = pd.Series([10, 20, 30, 40])
s2 = pd.Series([10, 20, 20, 40])

print(s1.is_unique)  # True
print(s2.is_unique)  # False
````

<h5 style="color:#78B89A; font-weight:bold;">index → index labels</h5>

```python
series.index
```


````python
s = pd.Series([85, 91, 78], index=["Dev", "Rahul", "Aman"])

print(s.index)
````

<h5 style="color:#78B89A; font-weight:bold;">values → actual values</h5>

```python
series.values
```


````python
print(s.values)
````

So:

```text
s.index  → labels
s.values → data
```

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Quick Summary</h3>

| Attribute | Meaning |
|---|---|
| `size` | Total number of elements |
| `dtype` | Data type of values |
| `name` | Name of the Series |
| `is_unique` | Checks whether values are unique |
| `index` | Returns index labels |
| `values` | Returns actual values |
