<h3 style="color:#6FA8DC; font-weight:bold">01. PowerTransformer — Introduction</h3>

<h5 style="color:#78B89A; font-weight:bold;">What is PowerTransformer?</h5>

`PowerTransformer` is a Scikit-learn transformer used to transform numerical features so their distribution becomes **more Gaussian-like** and less skewed.

```text
Original feature
      ↓
PowerTransformer
      ↓
Less skewed / more Gaussian-like feature
      ↓
ML Model
```

**Main idea:** it changes the **shape of the distribution**, not just its scale.

<h5 style="color:#78B89A; font-weight:bold;">Why use it?</h5>

Some numerical features are heavily skewed:

```text
Right-skewed

████████████
██████
████
██
█
──────────────────→
```

A long tail can make some models harder to fit effectively. Power transformation can reduce this skewness.

It is particularly useful when EDA shows strong skewness and the chosen model can benefit from a more regular feature distribution.

<h5 style="color:#78B89A; font-weight:bold;">When to use it?</h5>

```text
EDA
 ↓
Check distribution
 ↓
Check skewness
 ↓
Strongly skewed?
   ↙       ↘
 YES       NO
  ↓         ↓
Try Power   Usually
Transform   no need
  ↓
Check result
  ↓
Train model
  ↓
Compare performance
```

Do **not** apply it automatically to every feature. Transformation should be based on the data and model.

<h5 style="color:#78B89A; font-weight:bold;">Main use cases</h5>

**1. Reduce strong skewness**

```text
Skewed feature
      ↓
PowerTransformer
      ↓
More symmetric distribution
```

**2. Make features more Gaussian-like**

Useful when the chosen ML method benefits from features being closer to a normal/Gaussian shape.

**3. Compress long tails**

Large values in a heavily skewed feature can be compressed by a suitable power transformation.

**4. Use inside Pipeline / ColumnTransformer**

```text
Raw data
   ↓
PowerTransformer
   ↓
Model
```

or:

```text
ColumnTransformer
   ├── selected numerical → PowerTransformer
   └── other columns → other preprocessing
   ↓
Model
```

This is especially useful because the transformer can be fitted only on training data and reused consistently at prediction time.

<h5 style="color:#78B89A; font-weight:bold;">Two methods</h5>

```python
PowerTransformer(method="yeo-johnson")
PowerTransformer(method="box-cox")
```

### Yeo-Johnson

```python
PowerTransformer(method="yeo-johnson")
```

Can handle:

```text
positive + zero + negative
```

It is the more generally applicable option and the Scikit-learn default.

### Box-Cox

```python
PowerTransformer(method="box-cox")
```

Requires:

```text
strictly positive values
```

So it cannot directly handle zero or negative values.

**Easy memory trick:**

```text
Yeo-Johnson → positive + zero + negative
Box-Cox     → positive only
```

<h5 style="color:#78B89A; font-weight:bold;">Standardization</h5>

By default:

```python
PowerTransformer(standardize=True)
```

The transformed output is also standardized.

If you want only the power transformation:

```python
PowerTransformer(standardize=False)
```

Then you can add a separate scaler in a Pipeline if needed.

```text
PowerTransformer
      ↓
distribution transformation
      +
(default)
standardization
```

<h5 style="color:#78B89A; font-weight:bold;">PowerTransformer vs other transformers</h5>

| Situation | Consider |
|---|---|
| Strong skew, want Gaussian-like distribution | `PowerTransformer` |
| Simple custom/log transformation | `FunctionTransformer` |
| Want values between 0 and 1 | `MinMaxScaler` |
| Want standardized scale | `StandardScaler` |
| Want a uniform distribution | `QuantileTransformer(output_distribution="uniform")` |
| Tree models | Scaling/transformation is often unnecessary |

```text
StandardScaler
→ changes scale

MinMaxScaler
→ changes scale/range

PowerTransformer
→ changes distribution shape
  + standardizes by default

QuantileTransformer
→ maps the distribution
```

<h5 style="color:#78B89A; font-weight:bold;">No Data Leakage</h5>

❌ Wrong:

```text
Complete dataset
      ↓
fit PowerTransformer
      ↓
Train/Test split
```

✅ Correct:

```text
Dataset
   ↓
Train / Test Split
   ↓
fit transformer on TRAIN only
   ↓
transform TRAIN
transform TEST
```

Best practice:

```text
Pipeline
   ↓
PowerTransformer
   ↓
Model
```

With cross-validation, the preprocessing can be fitted separately inside each training fold.
