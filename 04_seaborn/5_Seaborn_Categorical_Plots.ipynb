<h3 style="color:#6FA8DC; font-weight:bold">1. Categorical Plots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are categorical plots?</h5>

Categorical plots are used when one or more variables represent **categories/groups**.

Examples of categorical variables:

```text
sex
smoker
day
time
species
```

For example:

```text
day → Thur, Fri, Sat, Sun
sex → Male, Female
species → setosa, versicolor, virginica
```

Categorical plots help us compare values or distributions **between different categories**.

### Main Classification

```text
Categorical Plots
       |
       |-------------------------------|
       |               |               |
Categorical       Categorical      Categorical
Scatter           Distribution     Estimate
       |               |               |
 stripplot         boxplot         barplot
 swarmplot         violinplot      pointplot
                                  countplot
```

Figure-level function:

```python
sns.catplot()
```

⭐ **VVVV Important**

Categorical plots are very useful in **EDA** when we want to compare groups.

<h3 style="color:#6FA8DC; font-weight:bold">3. Categorical Scatter Plots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are categorical scatter plots?</h5>

A categorical scatter plot shows **individual observations** for different categories.

The two important plots are:

- **Stripplot**
- **Swarmplot**

Example question:

```text
How are total bills distributed for each day?
```

We have:

```text
X-axis → day (categorical)
Y-axis → total_bill (numerical)
```

Each observation is shown as a point.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">4. Stripplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.stripplot()</h5>

A **stripplot** displays individual observations as points along a categorical axis.

Basic syntax:

```python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill"
)
```

### When to use?

Use stripplot when you want to:

- See individual observations
- Compare groups
- See concentration of values
- See spread
- Notice gaps
- Spot possible outliers

It is useful when the number of observations is not so large that the points become unreadable.


````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Total Bill by Day")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">5. Stripplot → Jitter ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">why do points spread sideways?</h5>

In a categorical plot, many observations can have the same category.

For example, many points belong to:

```text
Sat
```

Without jitter, points can overlap exactly.

### Jitter

**Jitter means adding a small random displacement to points** so overlapping observations become easier to see.

Example:

```python
jitter=True
```

This spreads points slightly around their categorical position.

```text
Without jitter:

   •
   •
   •
   •

With jitter:

 • •
   •
 •   •
    •
```

⭐ **VVVV Important**

**Jitter does NOT change the actual data value.**

It only changes the visual position slightly to make overlapping observations visible.


````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=True
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Stripplot with Jitter")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">6. Stripplot → jitter value</h3>

<h5 style="color:#78B89A; font-weight:bold;">controlling jitter amount</h5>

`jitter` can also accept a numeric value.

For example:

```python
jitter=0.2
```

A larger value gives more horizontal spreading.

The exact amount should be chosen so that points become readable without making categories confusing.


````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=0.2
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Stripplot with Controlled Jitter")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">7. Stripplot → hue and dodge</h3>

<h5 style="color:#78B89A; font-weight:bold;">comparing another category</h5>

We can use:

```python
hue="sex"
```

to separate observations by sex.

When there are multiple hue groups, `dodge=True` can separate them within each category.


````python
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    jitter=True,
    dodge=True
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Total Bill by Day and Sex")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">8. Swarmplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.swarmplot()</h5>

A **swarmplot** also shows individual observations.

The main difference is that Seaborn tries to position points so they **do not overlap**, while still representing their values.

Basic syntax:

```python
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill"
)
```

### When to use?

Use swarmplot when:

- You want to see individual observations
- Point overlap is a problem
- You want a clearer view of the distribution within categories
- The dataset is reasonably small/moderate

⚠️ Swarmplot can become slow or crowded with very large datasets.


````python
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Swarmplot of Total Bill by Day")

plt.show()
````

### Stripplot vs Swarmplot

| Stripplot | Swarmplot |
|---|---|
| Shows individual points | Shows individual points |
| Can overlap | Tries to avoid overlap |
| `jitter` controls spreading | Automatically arranges points |
| Usually faster | Can be slower for large data |
| Good for larger/simple views | Good for detailed smaller datasets |

⭐ Easy memory:

```text
Stripplot → points + optional jitter
Swarmplot → points arranged to avoid overlap
```

<h3 style="color:#6FA8DC; font-weight:bold">11. Boxplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.boxplot()</h5>

A boxplot summarizes a distribution using:

```text
Minimum
Q1
Median
Q3
Maximum
```

It is especially useful for comparing distributions between categories.

### When to use?

Use boxplot when you want to:

- Compare spread between groups
- Find the median
- Understand quartiles
- Identify possible outliers
- Compare many categories

⭐ **VVVV Important**

Boxplot is very useful for spotting **possible outliers**.


````python
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Boxplot of Total Bill by Day")

plt.show()
````

### Boxplot structure

```text
       ─── upper whisker
       │
    ┌───────┐
    │       │
    │ Median│
    │       │
    └───────┘
       │
       ─── lower whisker

       •  possible outlier
```

The box represents the **interquartile range (IQR)**:

```text
IQR = Q3 - Q1
```

Points outside the whiskers may be shown as potential outliers according to the boxplot rule.

<h3 style="color:#6FA8DC; font-weight:bold">13. Violinplot ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.violinplot()</h5>

A violinplot combines ideas from a **boxplot and KDE-style density visualization**.

It shows:

- Central tendency
- Spread
- Distribution shape
- Density
- Possible multimodal patterns

### When to use?

Use violinplot when you want more information about the **shape of the distribution** than a basic boxplot gives.

For example, it can show whether values are concentrated around one region or whether there are multiple dense regions.

⭐ **VVVV Important**

```text
Boxplot  → summary + outliers
Violin   → summary + distribution shape/density
```


````python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Violinplot of Total Bill by Day")

plt.show()
````

### Violinplot parameters

| Parameter | Purpose |
|---|---|
| `data` | DataFrame |
| `x` | X variable |
| `y` | Y variable |
| `hue` | Additional grouping |
| `order` | Category order |
| `hue_order` | Hue order |
| `cut` | Extend KDE beyond observations |
| `bw_adjust` | KDE smoothing |
| `density_norm` | How violin width is normalized |
| `inner` | What to draw inside violin |
| `split` | Split violin between hue groups |
| `fill` | Fill violin |
| `width` | Violin width |
| `dodge` | Separate hue groups |
| `gap` | Gap between dodged violins |
| `scale` | Older parameter; use current density normalization options |
| `palette` | Colors |
| `linewidth` | Edge width |
| `ax` | Specific Axes |

⭐ Focus first on:

```text
hue
inner
split
cut
bw_adjust
width
dodge
```


````python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    split=True,
    inner="box"
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Violinplot by Day and Sex")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">14. inner parameter in Violinplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">what appears inside the violin?</h5>

`inner` controls the information drawn inside the violin.

Common choices include:

```python
inner="box"
inner="quart"
inner="point"
inner="stick"
inner=None
```

Example:

```python
inner="box"
```

shows a small boxplot inside each violin.

```python
inner=None
```

removes the internal summary.


````python
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    inner="box"
)

plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.title("Violinplot with Inner Box")

plt.show()
````

These plots help summarize categorical data.

<h3 style="color:#6FA8DC; font-weight:bold">16. Barplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.barplot()</h5>

A barplot shows an **estimated numerical value** for each category.

By default, the estimator is typically the **mean**.

Example:

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)
```

This answers:

**What is the average total bill for each day?**

### When to use?

Use barplot when you want to:

- Compare a numerical summary across categories
- Compare means
- Use another estimator such as median
- Show uncertainty/error bars


````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.title("Average Total Bill by Day")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">17. Barplot → estimator</h3>

<h5 style="color:#78B89A; font-weight:bold;">what summary should the bar represent?</h5>

The `estimator` determines the summary statistic.

For example:

```python
estimator="median"
```

or using a Python function:

```python
estimator="mean"
```

Modern Seaborn also supports callable functions such as:

```python
estimator="median"
```

or:

```python
estimator=lambda x: x.mean()
```

Example with median:

```python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator="median"
)
```

⭐ **VVVV Important**

```text
estimator → what value is represented by each bar?
```


````python
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    estimator="median"
)

plt.xlabel("Day")
plt.ylabel("Median Total Bill")
plt.title("Median Total Bill by Day")

plt.show()
````


Examples:

```python

kind="box"

kind="bar"

```

So:

<h3 style="color:#6FA8DC; font-weight:bold">27. catplot → Facet Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">row, col and col_wrap</h5>

Like `relplot()` and `displot()`, `catplot()` can create facets.

Example:

```python
col="time"
```

creates separate panels for lunch and dinner.

```python
row="smoker"
```

creates separate rows for smoker/non-smoker.

```python
col_wrap=2
```

wraps the columns into multiple rows.


````python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col="time",
    kind="box"
)

plt.show()
````

````python
sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    col="day",
    col_wrap=2,
    kind="violin"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">30. Which Categorical Plot Should I Use?</h3>

<h5 style="color:#78B89A; font-weight:bold;">easy decision tree</h5>

```text
Want individual observations?
          |
      ------------
      |          |
   Strip      Swarm
      |          |
   overlap    less overlap


Want distribution?
          |
      -----------
      |         |
    Box       Violin
      |         |
 summary    shape/density


Want a numerical estimate?
          |
      -----------
      |         |
     Bar      Point
      |
   central
   tendency


Want category counts?
          |
      Countplot
```

⭐ **Quick rule**

```text
Individual points → strip / swarm
Distribution      → box / violin
Estimate           → bar / point
Counts             → count
Multiple panels    → catplot
```

<h3 style="color:#6FA8DC; font-weight:bold">31. Complete Categorical EDA Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">using different plots on the same dataset</h5>

Suppose we want to understand:

```text
How is total_bill distributed across days?
```

We can use different plots depending on what we want to know.

```text
Stripplot  → every observation
```
