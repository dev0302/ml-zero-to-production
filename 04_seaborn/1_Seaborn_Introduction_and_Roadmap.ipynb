<h3 style="color:#6FA8DC; font-weight:bold">1. Seaborn Introduction</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is Seaborn?</h5>

**Seaborn** is a Python library used for **data visualization**.

It is built on top of **Matplotlib** and provides a simpler way to create many statistical and data-analysis graphs.

Seaborn is especially useful for:

- Data Analysis
- Exploratory Data Analysis (EDA)
- Statistics
- Machine Learning
- Understanding relationships between variables
- Understanding distributions
- Visualizing categorical data

⭐ **VVVV Important**

**Seaborn → statistical/data visualization library built on Matplotlib.**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Why Seaborn?</h3>

<h5 style="color:#78B89A; font-weight:bold;">why do we need Seaborn when Matplotlib already exists?</h5>

Matplotlib is very powerful and gives us detailed control over graphs.

But for **data analysis**, we often need graphs such as:

- Relationship between variables
- Distribution of data
- Regression lines
- Categorical comparisons
- Correlation heatmaps
- Multiple statistical plots

Seaborn provides many of these visualizations with **less code** and useful default styling.

It also works naturally with **Pandas DataFrames**.

### Main advantages

- Simple syntax
- Works well with Pandas
- Good default appearance
- Many statistical plots are built in
- Very useful for EDA
- Easy to compare variables and categories

⭐ **VVVV Important**

**Matplotlib → detailed/general plotting control**

**Seaborn → easier statistical/data visualization**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. When to Use Seaborn?</h3>

<h5 style="color:#78B89A; font-weight:bold;">when is Seaborn useful?</h5>

Seaborn is especially useful when working with a **dataset** and trying to understand the data before Machine Learning.

For example, suppose our dataset contains:

We may want to ask:

Seaborn provides visualizations that help answer these questions.

This makes it very useful during:

```text

   ↓
EDA
   ↓
Visualization
   ↓
Feature Understanding
   ↓
Machine Learning
```

<h3 style="color:#6FA8DC; font-weight:bold">7. Basic Seaborn Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">first plot</h5>

Seaborn works very naturally with Pandas DataFrames.


````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Runs": [120, 180, 150, 220, 200]
})

sns.lineplot(
    data=df,
    x="Month",
    y="Runs"
)

plt.title("Runs by Month")

plt.show()
````

Notice that we can provide the complete DataFrame:

```python
data=df
```

and then specify its column names:

```python
x="Month"
y="Runs"
```

This is one of the main reasons Seaborn is convenient for EDA.

<h3 style="color:#6FA8DC; font-weight:bold">8. Seaborn Roadmap</h3>

<h5 style="color:#78B89A; font-weight:bold;">types of functions</h5>

There are two important levels of Seaborn functions:

### 1. Figure Level

These functions manage the **overall figure** and can create complete figures, including multiple facets/subplots.

Examples:

```python
sns.relplot()
sns.displot()
sns.catplot()
sns.lmplot()
```

### 2. Axes Level

These functions draw a plot on a particular **Axes**.

Examples:

```python
sns.scatterplot()
sns.lineplot()
sns.histplot()
sns.barplot()
sns.boxplot()
sns.heatmap()
```

⭐ **VVVV Important**

**Figure-level → manages the complete figure**

**Axes-level → works with a particular plotting area**

<h3 style="color:#6FA8DC; font-weight:bold">9. Main Classification of Seaborn</h3>

<h5 style="color:#78B89A; font-weight:bold;">our Seaborn roadmap</h5>

### 1. Relational Plot

Used to understand the **relationship between variables**.

Examples:

```python
sns.scatterplot()
sns.lineplot()
sns.relplot()
```

Questions:

```text
Are two variables related?

How does one variable change with another?
```

### 2. Distribution Plot

Used to understand the **distribution of data**.

Examples:

```python
sns.histplot()
sns.kdeplot()
sns.displot()
```

Questions:

```text
Where are most values?

How is the data distributed?

Is the data skewed?
```

### 3. Categorical Plot

Used when one or more variables represent **categories**.

Examples:

```python
sns.barplot()
sns.countplot()
sns.boxplot()
sns.violinplot()
sns.stripplot()
sns.swarmplot()
```

Questions:

```text
Which category has higher values?

How are values distributed within each category?
```

### 4. Regression Plot

Used to understand a relationship along with a **regression/trend line**.

Examples:

```python
sns.regplot()
sns.lmplot()
```

### 5. Matrix Plot

Used when data is naturally represented as a **matrix**.

Important example:

```python
sns.heatmap()
```

A common use is visualizing a **correlation matrix**.

### 6. Multiplots

Used to create **multiple related plots** in one figure.

Examples include:

```python
sns.relplot()
sns.displot()
sns.catplot()
sns.lmplot()
```

<h3 style="color:#6FA8DC; font-weight:bold">11. Figure Level vs Axes Level — Easy Understanding</h3>

<h5 style="color:#78B89A; font-weight:bold;">simple way to remember</h5>

Imagine a notebook page.

### Figure

The **whole page/container**.

### Axes

One **graph area** on that page.

```text
Figure
┌─────────────────────────────┐
│                             │
│        Axes / Plot          │
│                             │
└─────────────────────────────┘
```

With multiple plots:

```text
Figure
┌──────────────┬──────────────┐
│    Axes 1    │    Axes 2    │
├──────────────┼──────────────┤
│    Axes 3    │    Axes 4    │
└──────────────┴──────────────┘
```
