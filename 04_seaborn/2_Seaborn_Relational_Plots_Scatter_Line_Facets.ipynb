<h3 style="color:#6FA8DC; font-weight:bold">1. Relational Plots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are relational plots?</h5>

Relational plots are used to understand the **relationship between variables**.

For example:

```text
Total Bill  ↔  Tip
```

We may ask:

- Does tip increase when total bill increases?
- Is there a pattern?
- Does the pattern change for different groups?

The two main relational plots we will learn are:

- **Scatter Plot**
- **Line Plot**

The main Seaborn figure-level function is:

```python
sns.relplot()
```

⭐ **VVVV Important**

**Relational Plot → relationship between variables**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Scatter Plot Dataset → tips</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.load_dataset("tips")</h5>

Seaborn provides built-in datasets for learning.

We can load the `tips` dataset:

```python
tips = sns.load_dataset("tips")
```

The dataset contains columns such as:

```text
total_bill
tip
sex
smoker
day
time
size
```

For our first graph:

```text
X-axis → total_bill
Y-axis → tip
```


````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

tips.head()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. Scatter Plot → Axes Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.scatterplot()</h5>

Basic syntax:

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

Here:

```text
data → DataFrame
x    → X-axis column
y    → Y-axis column
```

Each row becomes one point on the graph.


````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Total Bill vs Tip")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">4. Scatter Plot → Figure Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.relplot()</h5>

The figure-level equivalent is:

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

### Difference

```text
Axes Level
    ↓
sns.scatterplot()

Figure Level
    ↓
sns.relplot()
```

**Axes-level → draws on a particular Axes**

**Figure-level → creates/manages the complete figure**


````python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">5. Scatter Plot Features → hue, style, size</h3>

<h5 style="color:#78B89A; font-weight:bold;">showing more variables in one graph</h5>

Seaborn can map different columns to different visual properties.

### hue

```python
hue="sex"
```

→ different colors for different sexes.

### style

```python
style="time"
```

→ different marker styles for different times.

### size

```python
size="size"
```

→ point size depends on table size.

⭐ **VVVV Important**

```text
hue   → color
style → marker/style
size  → point size
```


````python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="time",
    size="size",
    sizes=(30, 200),
    alpha=0.7
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Total Bill vs Tip")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">6. Scatter Plot → Figure Level + Features</h3>

<h5 style="color:#78B89A; font-weight:bold;">relplot with hue, style and size</h5>

The same semantic mappings can be used with the figure-level function:

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="time",
    size="size"
)
```

The important difference is not the meaning of `hue`, `style` or `size`.

The main difference is the **level of the function**:

```text
scatterplot() → Axes level
relplot()     → Figure level
```


````python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="time",
    size="size",
    sizes=(30, 200)
)

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">7. Facet Plots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is faceting?</h5>

A **facet plot** divides the dataset into groups and creates a separate plot for each group.

This is mainly done using **figure-level functions**.

For example:

```python
col="time"
```

creates separate plots for:

```text
Lunch
Dinner
```

Example:

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="time"
)
```

⭐ **VVVV Important**

`col` → separate plots column-wise.

`row` → separate plots row-wise.


````python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="time"
)

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">8. row, col and col_wrap</h3>

### row

```python
row="sex"
```

→ creates separate rows for each category.

### col

```python
col="day"
```

→ creates separate columns/plots for each category.

### col_wrap

```python
col_wrap=2
```

→ maximum 2 facet plots in each row.

This is useful when there are many categories.

Example:

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    col_wrap=2
)
```


````python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    col_wrap=2
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">9. Line Plot Dataset → Gapminder</h3>

<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. Line Plot → Axes Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.lineplot()</h5>

A line plot is useful when the X-axis has an **ordered progression**, such as year or time.

Let's see India's life expectancy over the years.

First filter India:

```python
india = gap[gap["country"] == "India"]
```

Then use:

```python
sns.lineplot(
    data=india,
    x="year",
    y="lifeExp"
)
```


````python
india = gap[
    gap["country"] == "India"
]

sns.lineplot(
    data=india,
    x="year",
    y="lifeExp"
)

plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.title("India Life Expectancy Over Time")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">11. Line Plot → Figure Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">relplot(kind="line")</h5>

The figure-level version uses:

```python
sns.relplot(
    data=india,
    x="year",
    y="lifeExp",
    kind="line"
)
```

Remember:

```text
Axes Level
    ↓
sns.lineplot()

Figure Level
    ↓
sns.relplot(kind="line")
```

`relplot()` is the figure-level interface for relational plots.

```python
kind="scatter"
```

→ scatter plot

```python
kind="line"
```

→ line plot


````python
sns.relplot(
    data=india,
    x="year",
    y="lifeExp",
    kind="line"
)

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">12. Line Plot → hue</h3>

<h5 style="color:#78B89A; font-weight:bold;">comparing multiple groups</h5>

We can use `hue` to create different lines for different countries.

```python
hue="country"
```

This allows us to compare their trends.

Other useful semantic parameters include:

```text
hue   → color/group
style → line style
size  → line width
```


````python
selected = gap[
    gap["country"].isin(
        ["India", "China", "Japan"]
    )
]

sns.lineplot(
    data=selected,
    x="year",
    y="lifeExp",
    hue="country"
)

plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy Comparison")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">13. Line Plot Parameters</h3>

<h5 style="color:#78B89A; font-weight:bold;">important parameters</h5>

| Parameter | Purpose |
|---|---|
| `data` | DataFrame |
| `x` | X-axis column |
| `y` | Y-axis column |
| `hue` | Color/group |
| `style` | Line style/group |
| `size` | Line width/group |
| `markers` / `marker` | Show markers |
| `dashes` | Control dash styles |
| `palette` | Color palette |
| `legend` | Legend control |
| `errorbar` | Error representation |
| `ax` | Specific Matplotlib Axes |

When `hue` is used, Seaborn normally creates a legend automatically.

For example:

```python
legend=False
```

can be used to hide it.


````python
sns.lineplot(
    data=selected,
    x="year",
    y="lifeExp",
    hue="country",
    style="country",
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy by Country")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">14. Line Plot + Facets</h3>

<h5 style="color:#78B89A; font-weight:bold;">relplot + col</h5>

Figure-level `relplot()` becomes especially useful when combined with facets.

Example:

```python
sns.relplot(
    data=selected,
    x="year",
    y="lifeExp",
    hue="country",
    col="country",
    col_wrap=2,
    kind="line"
)
```

Here:

```text
x         → year
y         → lifeExp
hue       → country
col       → separate plot for each country
col_wrap  → 2 plots per row
kind      → line
```

This creates multiple related plots in one figure.


````python
selected = gap[
    gap["country"].isin(
        ["India", "China", "Japan", "United States"]
    )
]

sns.relplot(
    data=selected,
    x="year",
    y="lifeExp",
    hue="country",
    col="country",
    col_wrap=2,
    kind="line"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">15. Scatter vs Line Plot</h3>

| Scatter Plot | Line Plot |
|---|---|
| Shows individual observations | Connects observations |
| Good for relationships | Good for ordered trends |
| Points are separate | Points are connected |
| `sns.scatterplot()` | `sns.lineplot()` |
| Figure level → `relplot(kind="scatter")` | Figure level → `relplot(kind="line")` |

### Easy memory trick

```text
Scatter → points
Line    → points + connection
```

Use a line when the X-axis has a meaningful **order/progression**.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">16. Figure Level vs Axes Level — Final</h3>

| Axes Level | Figure Level |
|---|---|
| `sns.scatterplot()` | `sns.relplot(kind="scatter")` |
| `sns.lineplot()` | `sns.relplot(kind="line")` |
| Draws on an Axes | Creates/manages a complete figure |
| Can receive `ax=` | Very useful for facets |
| Good for custom Matplotlib layouts | Good for multiplots |

### Semantic mappings

```text
hue   → color
style → marker / line style
size  → point size / line width
```

### Faceting

```text
row       → separate rows
col       → separate columns
col_wrap  → wrap columns into multiple rows
```

⭐ **FINAL MEMORY TRICK**

```text
Relational
    ↓
Scatter → relationship / points
Line    → ordered trend
    ↓
Axes Level
    ↓
scatterplot(), lineplot()
    ↓
Figure Level
    ↓
relplot()
    ↓
Facets
    ↓
row + col + col_wrap
```
