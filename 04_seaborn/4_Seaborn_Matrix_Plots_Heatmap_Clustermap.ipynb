<h3 style="color:#6FA8DC; font-weight:bold">1. Matrix Plots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are matrix plots?</h5>

Matrix plots are used when data is arranged in a **matrix / table-like form** and we want to visualize relationships or values using colors.

The two important matrix plots we will learn are:

- **Heatmap**
- **Cluster Map**

⭐ **VVVV Important**

```text
Matrix Plot
     ↓
Heatmap       → values using colors
Cluster Map   → values + clustering
```


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Heatmap</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a heatmap?</h5>

A heatmap represents values using **different colors**.

For example:

```text
        A    B    C
    A  1.0  0.8  0.2
    B  0.8  1.0  0.4
    C  0.2  0.4  1.0
```

The values are converted into colors.

Usually:

```text
darker / stronger color → larger value
lighter / weaker color  → smaller value
```

The exact meaning depends on the selected colormap.

### When to use Heatmap?

Heatmaps are useful for:

- Correlation matrices
- Comparing many values at once
- Finding patterns
- Finding high/low values
- Showing relationships between variables
- EDA

⭐ A very common ML/Data Science use:

**Correlation heatmap**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. Basic Heatmap Syntax</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.heatmap()</h5>

Basic syntax:

```python
sns.heatmap(data)
```

Usually `data` is a **2D table / matrix**.

Example:

```python
data = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

sns.heatmap(data)
```


````python
import seaborn as sns
import matplotlib.pyplot as plt

data = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

sns.heatmap(data)

plt.title("Basic Heatmap")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">4. Heatmap with a DataFrame</h3>

<h5 style="color:#78B89A; font-weight:bold;">practical EDA example</h5>

A very common use is creating a correlation matrix.

We can use the `tips` dataset:

```python
tips = sns.load_dataset("tips")
```

Then select numerical columns and calculate correlation:

```python
corr = tips.corr(numeric_only=True)
```

Now visualize it:

```python
sns.heatmap(corr)
```

⭐ **VVVV Important**

```text
DataFrame
   ↓
corr()
   ↓
Correlation Matrix
   ↓
Heatmap
```


````python
tips = sns.load_dataset("tips")

corr = tips.corr(numeric_only=True)

sns.heatmap(corr)

plt.title("Correlation Heatmap")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">5. annot → Show Values ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">writing values inside cells</h5>

By default, a heatmap mainly shows values through color.

To display the actual values inside every cell:

```python
annot=True
```

Example:

```python
sns.heatmap(
    corr,
    annot=True
)
```

Now each cell contains its numerical value.

⭐ **VVVV Important**

```text
annot=False → only colors
annot=True  → colors + values
```


````python
sns.heatmap(
    corr,
    annot=True
)

plt.title("Correlation Heatmap with Values")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">6. fmt → Formatting Annotated Values</h3>

<h5 style="color:#78B89A; font-weight:bold;">controlling decimal places</h5>

When `annot=True`, we can control the displayed format using:

```python
fmt=".2f"
```

Meaning:

```text
.2f → 2 digits after decimal
```

Example:

```python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)
```


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">7. cmap → Color Map ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">changing the color scheme</h5>

`cmap` means **colormap**.

It controls how numerical values are converted into colors.

Example:

```python
cmap="Blues"
```

Other commonly used colormaps include:

```text
Blues
Reds
Greens
YlOrRd
coolwarm
viridis
magma
```

For correlation matrices, a diverging colormap such as:

```python
cmap="coolwarm"
```

is often useful because negative and positive values can be visually separated.


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">8. linewidths → Lines Between Cells</h3>

<h5 style="color:#78B89A; font-weight:bold;">creating separation between cells</h5>

We can add lines between heatmap cells using:

```python
linewidths=0.5
```

Example:

```python
sns.heatmap(
    corr,
    annot=True,
    linewidths=0.5
)
```

`linewidths` controls the width of the lines separating cells.


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Heatmap with Cell Lines")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">9. linecolor</h3>

<h5 style="color:#78B89A; font-weight:bold;">color of cell separation lines</h5>

Along with `linewidths`, we can specify:

```python
linecolor="white"
```

Example:

```python
sns.heatmap(
    corr,
    linewidths=0.5,
    linecolor="white"
)
```

So:

```text
linewidths → thickness of lines
linecolor  → color of lines
```


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    linecolor="white"
)

plt.title("Styled Correlation Heatmap")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. vmin and vmax</h3>

<h5 style="color:#78B89A; font-weight:bold;">controlling the value range</h5>

We can control the minimum and maximum values represented by the colormap.

```python
vmin=-1
vmax=1
```

This is especially useful for correlation matrices because correlation normally ranges from:

```text
-1 → +1
```

Example:

```python
sns.heatmap(
    corr,
    vmin=-1,
    vmax=1
)
```


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap with Fixed Range")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">11. center</h3>

<h5 style="color:#78B89A; font-weight:bold;">setting the center of the colormap</h5>

`center` can be used to specify the value that should be considered the center of the colormap.

For correlation:

```python
center=0
```

is useful because:

```text
negative → 0 → positive
```

Example:

```python
sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)
```


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Centered Correlation Heatmap")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">12. annot_kws</h3>

<h5 style="color:#78B89A; font-weight:bold;">customizing annotation text</h5>

`annot_kws` allows us to pass text formatting options for annotations.

Example:

```python

```

This changes the size of the values written inside cells.

Example:

```python
sns.heatmap(
    corr,
    annot=True,
    annot_kws={"size": 10}

```


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    annot_kws={"size": 10}
)

plt.title("Heatmap with Formatted Annotations")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">13. cbar → Color Bar</h3>

<h5 style="color:#78B89A; font-weight:bold;">understanding the colors</h5>

The color bar tells us which colors correspond to which values.

By default it is shown.

To hide it:

```python
cbar=False
```

Example:

```python
sns.heatmap(
    corr,
    cbar=False
)
```

⭐ **Color bar = scale/reference for the colors.**


````python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    cbar=False
)

plt.title("Heatmap without Color Bar")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">16. Cluster Map</h3>

<h5 style="color:#78B89A; font-weight:bold;">heatmap + hierarchical clustering</h5>

A **cluster map** is similar to a heatmap, but it also performs **clustering**.

Function:

```python
sns.clustermap()
```

It groups similar rows and columns together.

This makes patterns and groups easier to identify.

### Main difference

```text
Heatmap
   ↓
shows values using colors

Cluster Map
   ↓
shows values using colors
+
reorders similar rows/columns using clustering
```

⭐ **VVVV Important**

**Clustermap = Heatmap + hierarchical clustering**

<h3 style="color:#6FA8DC; font-weight:bold">18. Basic Cluster Map</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.clustermap()</h5>

Basic syntax:

```python
sns.clustermap(
    iris_features
)
```

The cluster map shows:

- A heatmap
- Row dendrogram
- Column dendrogram
- Reordered rows/columns according to similarity

A **dendrogram** is the tree-like diagram showing how observations are grouped.


````python
sns.clustermap(
    iris_features
)

plt.show()
````


Example:


````python
sns.clustermap(
    iris_features,
    cmap="viridis"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">20. Cluster Map → standard_scale ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">important for iris features</h5>

The iris features have different ranges.

For example, different measurements may have different scales.

We can standardize the values before clustering using:

```python
standard_scale=1
```

Common meanings:

```text
standard_scale=0 → scale each row
standard_scale=1 → scale each column
```

For iris features, column-wise scaling is often useful:

```python
sns.clustermap(
    iris_features,
    standard_scale=1
)
```

⭐ Scaling can be important when variables are measured on different scales.


````python
sns.clustermap(
    iris_features,
    standard_scale=1,
    cmap="viridis"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">22. Heatmap vs Cluster Map</h3>

| Heatmap | Cluster Map |
|---|---|
| Shows values using colors | Shows values using colors |
| Does not automatically cluster rows/columns | Clusters rows/columns |
| `sns.heatmap()` | `sns.clustermap()` |
| Simpler | More advanced |
| Good for correlation matrices | Good for finding similar groups/patterns |
| No dendrogram | Has dendrograms |

### Easy memory

```text
Heatmap
→ color representation

Clustermap
→ color representation
+ clustering
+ dendrogram
```

<h3 style="color:#6FA8DC; font-weight:bold">23. Heatmap vs Clustermap → When to Use?</h3>

<h5 style="color:#78B89A; font-weight:bold;">practical decision</h5>

### Use Heatmap when:

You mainly want to:

- View a matrix
- See correlation
- Compare values
- Find high/low regions
- Understand relationships quickly

Example:

```python
corr = tips.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True
)
```

### Use Cluster Map when:

You also want to:

- Find similar observations
- Find similar variables
- Automatically group similar rows/columns
- Explore clusters/patterns

Example:

```python
sns.clustermap(
    iris_features,
    standard_scale=1
)
```
