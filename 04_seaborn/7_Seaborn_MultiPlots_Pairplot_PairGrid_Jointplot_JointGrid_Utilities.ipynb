<br><h3 style="color:#6FA8DC; font-weight:bold">MULTI-PLOTS</h3>

<h5 style="color:#78B89A; font-weight:bold;">2. What are Multi-Plots? → many plots in one figure</h5>

A **multi-plot** means showing multiple graphs together so that we can compare different parts of the data.

For example:

- different categories
- different variables
- different groups
- different relationships

Seaborn provides different ways to create multi-plots.

The most important ones in this notebook are:

| Function | Main use |
|---|---|
| `pairplot()` | Relationships between many numerical variables |
| `PairGrid()` | Custom pair-plot layout |
| `jointplot()` | Relationship between 2 variables + their distributions |
| `JointGrid()` | Custom joint-plot layout |

⭐ These are especially useful during **EDA (Exploratory Data Analysis)**.

<h5 style="color:#78B89A; font-weight:bold;">3. Pairplot → quick overview of many variables</h5>

`pairplot()` creates a **matrix of plots** for numerical columns.

It shows:

- scatter plots between different variables
- distribution of each variable on the diagonal

### Syntax

```python
sns.pairplot(data)
```

### Why use it?

Suppose a dataset has:

`sepal_length`, `sepal_width`, `petal_length`, `petal_width`

Instead of making many scatter plots manually, `pairplot()` creates them together.

⭐ **VVVV Important:** Pairplot is one of the most useful Seaborn functions for quickly understanding relationships in a dataset.

````python
sns.pairplot(iris)
plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">4. Understanding the Pairplot → diagonal and off-diagonal</h5>

Suppose we have 4 numerical columns.

The pairplot creates a 4 × 4 matrix.

### Diagonal

The diagonal normally shows the **distribution of each variable**.

Example:

```text
sepal_length → distribution
sepal_width  → distribution
petal_length → distribution
petal_width  → distribution
```

### Off-diagonal

The other cells show relationships between two variables.

Usually:

```text
x = one variable
y = another variable
```

and a scatter plot is shown.

So one pairplot can help us answer:

- Which variables are related?
- Which variables have different distributions?
- Are there possible clusters?
- Are there possible outliers?

<h5 style="color:#78B89A; font-weight:bold;">6. Important Pairplot parameters → remember these</h5>

### `data`


### `x_vars`
Variables used on the x-axis.

```python
    x_vars=["sepal_length", "petal_length"],
    y_vars=["sepal_width", "petal_width"]
```


### `diag_kind`
Plot used on the diagonal.

Common choices:

- `"hist"`
- `"kde"`

```python
sns.pairplot(
    iris,
    diag_kind="kde"
)
```

### `corner`
If `True`, show only the lower triangle.

```python
sns.pairplot(
    iris,
```

This avoids showing duplicate relationships.

Height of each subplot.

Width/height ratio of each subplot.

### `palette`
Controls colors used for `hue`.

### `markers`
Marker style for different hue categories.

### `plot_kws`

### `diag_kws`

<h3 style="color:#6FA8DC; font-weight:bold">PAIRGRID</h3>

<h5 style="color:#78B89A; font-weight:bold;">8. What is PairGrid? → custom version of Pairplot</h5>

`PairGrid` creates the same type of matrix structure as a pairplot, but gives us **much more control**.

### Simple idea

```text
pairplot()
    ↓
ready-made and easy

PairGrid()
    ↓
customizable
```

With `PairGrid`, we can decide:

- what appears below the diagonal
- what appears above the diagonal
- what appears on the diagonal

### Basic syntax

```python
g = sns.PairGrid(data)
g.map(...)
```

````python
g = sns.PairGrid(iris)
g.map(sns.scatterplot)

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">9. PairGrid with different plots → upper, lower and diagonal</h5>

One of the biggest advantages of `PairGrid` is that we can use different functions in different parts.

```python
g.map_upper(...)
g.map_lower(...)
g.map_diag(...)
```

For example:

- upper triangle → scatter plot
- lower triangle → regression plot
- diagonal → histogram

This gives much more control than a normal pairplot.

````python
g = sns.PairGrid(iris)

g.map_upper(sns.scatterplot)
g.map_lower(sns.regplot)
g.map_diag(sns.histplot)

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">10. PairGrid with hue → category-wise comparison</h5>

We can also use `hue`.

```python
g = sns.PairGrid(
    iris,
    hue="species"
)

g.map(sns.scatterplot)
g.add_legend()
```

`add_legend()` adds the legend for the hue categories.

````python
g = sns.PairGrid(
    iris,
    hue="species"
)

g.map(sns.scatterplot)
g.add_legend()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">JOINTPLOT</h3>

<h5 style="color:#78B89A; font-weight:bold;">12. What is Jointplot? → relationship + distributions</h5>

`jointplot()` studies the relationship between **two variables**.

It combines:

- a main plot in the center
- distribution of x on the top
- distribution of y on the side

Example:

```text
          X distribution
       ───────────────
       |             |
Y dist |  Main plot  |
       |             |
       ───────────────
```

### Why use it?

It is useful when we want to study:

- relationship between two numerical variables
- distribution of each variable
- possible correlation
- possible outliers
- concentration of observations

### Syntax

```python
sns.jointplot(
    data=data,
    x="column1",
    y="column2"
)
```

````python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">13. Jointplot kinds → change the central relationship plot</h5>

`kind` decides the type of main plot.

Important values:

- `"scatter"` → scatter plot
- `"kde"` → KDE
- `"hist"` → histogram
- `"hex"` → hexagonal bins
- `"reg"` → regression
- `"resid"` → residual plot

Examples:

```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="reg"
)
```

or

```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="kde"
)
```

⭐ `kind="reg"` is useful when we want to see the relationship together with a regression line.

````python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="reg"
)

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">14. Jointplot with hue → compare categories</h5>


This allows us to compare the relationship for different categories.

<h3 style="color:#6FA8DC; font-weight:bold">JOINTGRID</h3>

<h5 style="color:#78B89A; font-weight:bold;">16. What is JointGrid? → custom version of Jointplot</h5>

`JointGrid` gives manual control over the central plot and the two marginal plots.

### Basic structure

```python
g = sns.JointGrid(
    data=tips,
    x="total_bill",
    y="tip"
)

g.plot(
    sns.scatterplot,
    sns.histplot
)
```

Here we can decide what function is used for:

- joint/main plot
- marginal plots

````python
g = sns.JointGrid(
    data=tips,
    x="total_bill",
    y="tip"
)

g.plot(
    sns.scatterplot,
    sns.histplot
)

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">17. JointGrid with separate mappings → maximum control</h5>

We can separately control the joint plot and marginal plots.

```python
g = sns.JointGrid(
    data=tips,
    x="total_bill",
    y="tip"
)

g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.histplot)
```

### Useful methods

`plot_joint()` → controls the central plot

`plot_marginals()` → controls both marginal plots

`plot_marginals(x=...)` → customize x marginal

`plot_marginals(y=...)` → customize y marginal

This is useful when we need a customized joint visualization.

````python
g = sns.JointGrid(
    data=tips,
    x="total_bill",
    y="tip",
    height=6
)

g.plot_joint(sns.regplot)
g.plot_marginals(sns.histplot)

plt.show()
````
