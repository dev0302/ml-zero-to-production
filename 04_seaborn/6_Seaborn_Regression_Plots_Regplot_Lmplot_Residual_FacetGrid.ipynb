<h3 style="color:#6FA8DC; font-weight:bold">1. Regression Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is regression?</h5>

Regression is used to understand the **relationship between numerical variables** and to estimate how one variable changes when another variable changes.

For example:

```text
X → total_bill
Y → tip
```

We can ask:

**As the total bill increases, does the tip also tend to increase?**

A regression plot shows:

```text
Actual data points
        +
Regression line
```

The line represents the **general trend** in the data.

⭐ **VVVV Important**

Regression plots are useful for:

- Understanding relationships
- Finding trends
- Seeing whether variables are positively/negatively related
- Visualizing a fitted regression model
- Checking how closely observations follow a trend


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Basic Idea of Regression Line</h3>

<h5 style="color:#78B89A; font-weight:bold;">line of best fit</h5>

A regression line is a line that represents the overall trend of the observations.

```text
Y
↑
│                 •
│             •
│          •
│       •
│    •
│  /
│ /
└──────────────────→ X
```

For simple linear regression:

```text
ŷ = b₀ + b₁x
```

Where:

```text
ŷ  → predicted value
b₀ → intercept
b₁ → slope
x  → input variable
```

The fitted line tries to make the prediction errors/residuals as small as possible according to the least-squares method.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. Positive and Negative Regression</h3>

<h5 style="color:#78B89A; font-weight:bold;">understanding the direction</h5>

### Positive relationship

When X increases, Y generally increases.

```text
Y ↑
│       •
│     •
│   •
│ •
└────────────→ X
```

Regression line slopes upward.

### Negative relationship

When X increases, Y generally decreases.

```text
Y ↑
│ •
│   •
│     •
│       •
└────────────→ X
```

Regression line slopes downward.

### No clear linear relationship

Points do not follow a clear straight-line trend.

⭐ A regression line helps us visually identify the **linear trend**, but a fitted line alone does not prove that one variable causes another.

<h3 style="color:#6FA8DC; font-weight:bold">4. Regression Plot → Seaborn</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.regplot()</h5>

Seaborn provides:

```python
sns.regplot()
```

Basic syntax:

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

It creates:

```text
Scatter points
+
Regression line
```

This is an **Axes-level** function.


````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Regression Plot: Total Bill vs Tip")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">5. regplot → scatter and line separately</h3>

<h5 style="color:#78B89A; font-weight:bold;">scatter points + fitted line</h5>

```text

```

We can control them.

### Hide scatter points

```python
scatter=False
```

### Hide regression line

```python
fit_reg=False
```

Example:

```python
sns.regplot(

    y="tip",

)
```


````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter=False
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Regression Line Only")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">7. regplot → ci</h3>

<h5 style="color:#78B89A; font-weight:bold;">confidence interval</h5>

By default, Seaborn can show a shaded **confidence interval** around the regression estimate.

We can control it with:

```python
ci=95
```

This requests a 95% confidence interval.

To hide it:

```python
ci=None
```

Example:

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ci=None
)
```

⭐ The shaded area represents uncertainty around the estimated regression relationship; it is not a region containing 95% of individual observations.


````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ci=None
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Regression Plot without Confidence Interval")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">8. regplot → order</h3>

<h5 style="color:#78B89A; font-weight:bold;">non-linear polynomial regression</h5>

Normally, `regplot()` fits a straight line.

```python
order=1
```

means linear regression.

We can fit a polynomial relationship:

```python
order=2
```

Example:

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    order=2
)
```

⭐ Use higher-order regression only when the relationship genuinely appears non-linear. A higher order can overfit.


````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    order=2
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Polynomial Regression Plot")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">9. regplot → jitter</h3>

<h5 style="color:#78B89A; font-weight:bold;">visualizing overlapping observations</h5>

`x_jitter` and `y_jitter` add small random visual displacement to observations.

Example:

```python
x_jitter=0.2
```

This is useful when several points overlap.

Remember:

```text
Jitter → visual adjustment
       → helps show overlapping points
       → does not represent a new measured value
```


````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    x_jitter=0.2
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Regression Plot with X Jitter")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. regplot → scatter_kws and line_kws</h3>

<h5 style="color:#78B89A; font-weight:bold;">customizing points and line</h5>

We can pass dictionaries to customize the scatter and regression line.

### Scatter

```python

    "s": 50,
    "alpha": 0.6
}
```

### Line

```python

    "linewidth": 2
```

These parameters are passed to the underlying Matplotlib plotting functions.


````python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter_kws={"s": 50, "alpha": 0.6},
    line_kws={"linewidth": 2}
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Customized Regression Plot")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">16. Residual Plot ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">checking regression errors</h5>

A **residual** is the difference between the actual value and predicted value.

```text
Residual = Actual - Predicted
```

If:

```text
Actual    = 10
Predicted = 8
```

then:

```text
Residual = 10 - 8 = 2
```

A residual plot shows residuals against the predictor variable.

### Why use a residual plot?

It helps us check whether the regression model is a reasonable fit.

We look for:

- Random-looking spread
- Patterns
- Curves
- Changing spread
- Unusual observations

Ideally, residuals should be scattered around zero without a clear systematic pattern.

⭐ **VVVV Important**

```text
Residual plot
      ↓
Check errors of regression model
```


<br>

<h3 style="color:#6FA8DC; font-weight:bold">17. Residual Plot → sns.residplot()</h3>

<h5 style="color:#78B89A; font-weight:bold;">basic syntax</h5>

Seaborn provides:

```python
sns.residplot()
```

Basic syntax:

```python
sns.residplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

It fits the regression relationship and plots the residuals.

The horizontal reference line is:

```text
y = 0
```

Residuals above zero → actual is above prediction.

Residuals below zero → actual is below prediction.


````python
sns.residplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.axhline(0, linestyle="--")

plt.xlabel("Total Bill")
plt.ylabel("Residual")
plt.title("Residual Plot")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">18. How to Read Residual Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">good vs problematic pattern</h5>

### Good-looking residual pattern

```text
Residual
   ↑
 + |  •    •
 0 |----•------•----→ X
 - |    •  •
```

Points are reasonably scattered around zero.

### Pattern

```text
Residual
   ↑
 + | •
 0 |   •
 - |      •   •
 └──────────────→ X
```

A clear pattern can indicate that a simple linear relationship may not capture the data well.

### Funnel shape

If the spread of residuals becomes larger or smaller as X changes, this can indicate **non-constant variance** (heteroscedasticity).

<h3 style="color:#6FA8DC; font-weight:bold">20. FacetGrid</h3>

<h5 style="color:#78B89A; font-weight:bold;">creating a grid of plots</h5>

`FacetGrid` is a Seaborn object used to create a **grid of subplots based on categorical variables**.

It is useful when we want to compare the same relationship or distribution across multiple groups.

Basic idea:

```text
          Group A    Group B
        ┌─────────┬─────────┐
Group 1 │  Plot   │  Plot   │
        ├─────────┼─────────┤
Group 2 │  Plot   │  Plot   │
        └─────────┴─────────┘
```

Basic syntax:

```python
g = sns.FacetGrid(
    tips,
    col="time"
)
```

Then map a plotting function:

```python
g.map_dataframe(
    sns.scatterplot,
    x="total_bill",
    y="tip"
)
```

⭐ `FacetGrid` gives more low-level control than simply calling `lmplot()`.


````python
g = sns.FacetGrid(
    tips,
    col="time"
)

g.map_dataframe(
    sns.scatterplot,
    x="total_bill",
    y="tip"
)

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">25. Complete Regression Workflow</h3>

<h5 style="color:#78B89A; font-weight:bold;">EDA → regression → residuals</h5>

A simple workflow can be:

```text
1. Scatter Plot
       ↓
   See relationship

2. Regression Plot
       ↓
   See fitted trend

3. Residual Plot
       ↓
   Check model errors

4. Facets / Groups
       ↓
   Compare relationship across categories
```

Example:

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

Then:

```python
sns.residplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

This gives us both the relationship and the residual/error view.
