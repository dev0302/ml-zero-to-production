<h3 style="color:#6FA8DC; font-weight:bold">1. Distribution Plots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are distribution plots?</h5>

Distribution plots help us understand the **distribution of data**.

They are mainly used for **univariate analysis** — studying one variable.

They help us find:

- Range of observations
- Spread of data
- Central tendency
- Shape of distribution
- Whether data is **bimodal** (two major peaks)
- Possible outliers
- Where most observations are concentrated

### Main plots

- **Histplot**
- **Kdeplot**
- **Rugplot**

⭐ **VVVV Important**

**Distribution plot → How is the data distributed?**


<br><h3 style="color:#6FA8DC; font-weight:bold">2. Univariate vs Bivariate</h3>

<h5 style="color:#78B89A; font-weight:bold;">important for all distribution plots</h5>

### Univariate

**Uni = one variable**

```python
sns.histplot(data=tips, x="total_bill")
```

Only `total_bill` is being studied.

### Bivariate

**Bi = two variables**

```python
sns.histplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

Now two numerical variables are involved.

⭐ Remember:

```text
x only → Univariate
x + y  → Bivariate
```

A `hue` variable can also be used to compare the distribution of one numerical variable across categories.


<br><h3 style="color:#6FA8DC; font-weight:bold">3. Histplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a histogram?</h5>

A histogram divides numerical data into **bins** and shows how many observations fall in each bin.

It is useful for understanding:

- Distribution
- Frequency
- Range
- Concentration
- Skewness
- Possible outliers
- Multiple peaks

### Practical example

Suppose we want to know:

**How are restaurant total bills distributed?**

```python
sns.histplot(
    data=tips,
    x="total_bill"
)
```


````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.histplot(
    data=tips,
    x="total_bill"
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Distribution of Total Bill")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">4. Histplot → bins ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">controlling intervals</h5>

A **bin** is an interval/range into which observations are grouped.

Example:

```text
10–15
15–20
20–25
25–30
```

Use:

```python
bins=20
```

to control the number of bins.

More bins → more detailed

Fewer bins → simpler

⭐ **VVVV Important**

`bins` is one of the most important histogram parameters.


````python
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Total Bill Distribution")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">5. Histplot → binwidth and binrange</h3>

Instead of the number of bins:

```python
bins=10
```

we can specify the width:

```python
binwidth=5
```

We can also specify the range:

```python
binrange=(0, 60)
```

### Easy difference

```text
bins=10
→ approximately 10 bins

binwidth=5
→ each bin has width 5

binrange=(0,60)
→ use this range for binning
```


````python
sns.histplot(
    data=tips,
    x="total_bill",
    binwidth=5,
    binrange=(0, 60)
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Controlled Histogram")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">6. Histplot → hue</h3>

<h5 style="color:#78B89A; font-weight:bold;">compare distributions by category</h5>

```python
hue="sex"
```

uses different colors for different groups.

The numerical variable is still:

```text
total_bill
```

but we compare its distribution for different values of `sex`.


````python
sns.histplot(
    data=tips,
    x="total_bill",
    hue="sex",
    bins=20
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Total Bill Distribution by Sex")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">7. Histplot → kde=True</h3>

<h5 style="color:#78B89A; font-weight:bold;">histogram + smooth curve</h5>

We can add a KDE curve directly:

```python
kde=True
```

```python
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)
```

The histogram shows binned frequency, while KDE shows a smooth estimate of the distribution.


````python
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Histogram with KDE")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">8. Histplot → stat</h3>

<h5 style="color:#78B89A; font-weight:bold;">what does bar height represent?</h5>

By default, histogram bars represent **count**.

Common `stat` values:

```python
stat="count"
stat="frequency"
stat="probability"
stat="percent"
stat="density"
```

Example:

```python
sns.histplot(
    data=tips,
    x="total_bill",
    stat="density"
)
```

```text
count       → number of observations
frequency   → frequency
probability → probability
percent     → percentage
density     → density
```


````python
sns.histplot(
    data=tips,
    x="total_bill",
    stat="density"
)

plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.title("Density Histogram")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">9. Histplot → multiple</h3>

<h5 style="color:#78B89A; font-weight:bold;">how multiple hue groups are displayed</h5>

When `hue` is used, `multiple` controls the arrangement.

Common options:

```python
multiple="layer"
multiple="stack"
multiple="dodge"
multiple="fill"
```

- `layer` → groups overlap
- `stack` → groups stack
- `dodge` → groups appear side-by-side
- `fill` → each bin shows proportions


````python
sns.histplot(
    data=tips,
    x="total_bill",
    hue="sex",
    multiple="stack",
    bins=15
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Stacked Histplot")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">10. Histplot → Axes Level vs Figure Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">important Seaborn distinction</h5>

### Axes level

```python
sns.histplot()
```

It draws on a particular Axes.

```python
fig, ax = plt.subplots()

sns.histplot(
    data=tips,
    x="total_bill",
    ax=ax
)

plt.show()
```

### Figure level

```python
sns.displot(
    data=tips,
    x="total_bill",
    kind="hist"
)
```

⭐ **VVVV Important**

```text
histplot() → Axes level
displot()  → Figure level
```


````python
fig, ax = plt.subplots(figsize=(8, 5))

sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    ax=ax
)

ax.set_xlabel("Total Bill")
ax.set_ylabel("Count")
ax.set_title("Axes-Level Histplot")

plt.show()
````

````python
sns.displot(
    data=tips,
    x="total_bill",
    bins=20,
    kind="hist"
)

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">11. Histplot → Facet Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">separate distributions</h5>

Figure-level `displot()` supports:

```python
row=
col=
col_wrap=
```

Example:

```python
sns.displot(
    data=tips,
    x="total_bill",
    col="day",
    col_wrap=2,
    kind="hist"
)
```

This creates a separate histogram for each day.

⭐ **Facet → same type of plot, separate panel for each group.**


````python
sns.displot(
    data=tips,
    x="total_bill",
    col="day",
    col_wrap=2,
    bins=15,
    kind="hist"
)

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">12. Bivariate Histplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">two numerical variables</h5>

Histplot can also represent two numerical variables:

```python
sns.histplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

This creates a **2D binned distribution**.

```text
x → total_bill
y → tip
```

So this is **bivariate analysis**.


````python
sns.histplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Bivariate Histogram")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">13. KDE Plot ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">Kernel Density Estimate</h5>

KDE stands for **Kernel Density Estimate**.

It gives a **smooth estimate of the distribution** of numerical data.

Unlike a histogram:

```text
Histogram → bars + bins
KDE       → smooth curve
```

KDE helps us understand:

- Shape
- Spread
- Concentration
- Skewness
- Multiple peaks
- Group differences

⭐ **VVVV Important**

**KDE → smooth estimate of probability density.**


<br><h3 style="color:#6FA8DC; font-weight:bold">14. KDE → Axes Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">sns.kdeplot()</h5>

Basic syntax:

```python
sns.kdeplot(
    data=tips,
    x="total_bill"
)
```


````python
sns.kdeplot(
    data=tips,
    x="total_bill"
)

plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.title("KDE of Total Bill")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">15. KDE → Figure Level</h3>

<h5 style="color:#78B89A; font-weight:bold;">displot(kind="kde")</h5>

Figure-level distribution plotting uses:

```python
sns.displot(
    data=tips,
    x="total_bill",
    kind="kde"
)
```

So:

```text
Axes Level
    ↓
sns.kdeplot()

Figure Level
    ↓
sns.displot(kind="kde")
```


````python
sns.displot(
    data=tips,
    x="total_bill",
    kind="kde"
)

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">16. KDE → hue and fill</h3>

<h5 style="color:#78B89A; font-weight:bold;">compare distributions</h5>

Use:

```python
hue="sex"
```

to compare groups.

Use:

```python
fill=True
```

to fill the area under the KDE curves.


````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="sex",
    fill=True
)

plt.xlabel("Total Bill")
plt.ylabel("Density")
plt.title("Filled KDE by Sex")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">17. KDE → bw_adjust ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">controlling smoothness</h5>

`bw_adjust` controls the amount of KDE smoothing.

```python
bw_adjust=0.5
```

→ less smoothing / more detail

```python
bw_adjust=2
```

→ more smoothing

⭐ Remember:

```text
smaller bw_adjust → less smooth
larger bw_adjust  → more smooth
```


````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    bw_adjust=0.5
)

plt.title("KDE with Less Smoothing")

plt.show()
````

````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    bw_adjust=2
)

plt.title("KDE with More Smoothing")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">18. Bivariate KDE ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">x + y</h5>

KDE can estimate the density of two numerical variables:

```python
sns.kdeplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

With:

```python
fill=True
```

we can see the density regions more clearly.

This is **bivariate KDE**.


````python
sns.kdeplot(
    data=tips,
    x="total_bill",
    y="tip",
    fill=True
)

plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Bivariate KDE")

plt.show()
````



<br><h3 style="color:#6FA8DC; font-weight:bold">20. KDE vs Histogram</h3>

| Histogram | KDE |
|---|---|
| Uses bins | Smooth curve |
| Shows count/statistic | Shows density |
| Depends on binning | Depends on smoothing |
| Direct frequency view | Smooth distribution shape |
| `sns.histplot()` | `sns.kdeplot()` |

They can also be combined:

```python
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True
)
```

⭐ Easy memory:

```text
Histogram → blocks/bins
KDE       → smooth curve
```


<br><h3 style="color:#6FA8DC; font-weight:bold">21. Rugplot</h3>

<h5 style="color:#78B89A; font-weight:bold;">individual observations</h5>

A rug plot represents each observation as a small mark on an axis.

```text
| |  || |    | ||| |    ||
──────────────────────────→
           values
```

It helps us see:

- Individual observations
- Concentration
- Gaps
- Range
- Possible outliers

Function:

```python
sns.rugplot()
```


````python
sns.rugplot(
    data=tips,
    x="total_bill"
)

plt.xlabel("Total Bill")
plt.title("Rug Plot of Total Bill")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">22. Rugplot with KDE</h3>

<h5 style="color:#78B89A; font-weight:bold;">using rug as supporting information</h5>

Rugplot is often useful together with KDE:

```python
sns.kdeplot(
    data=tips,
    x="total_bill"
)

sns.rugplot(
    data=tips,
    x="total_bill"
)
```

```text
KDE  → smooth distribution
Rug  → actual observations
```


````python
sns.kdeplot(
    data=tips,
    x="total_bill"
)

sns.rugplot(
    data=tips,
    x="total_bill"
)

plt.xlabel("Total Bill")
plt.title("KDE + Rug Plot")

plt.show()
````

<br><h3 style="color:#6FA8DC; font-weight:bold">25. Figure-Level Distribution Function → displot()</h3>

<h5 style="color:#78B89A; font-weight:bold;">one figure-level interface</h5>

`displot()` is the main **figure-level** distribution interface.

Examples:

```python
sns.displot(
    data=tips,
    x="total_bill",
    kind="hist"
)
```

→ histogram

```python
sns.displot(
    data=tips,
    x="total_bill",
    kind="kde"
)
```

→ KDE

It is especially useful for facets:

```python
row=
col=
col_wrap=
```

⭐ `rugplot()` is an Axes-level supporting plot; it is not a `displot(kind=...)` option.


<br><h3 style="color:#6FA8DC; font-weight:bold">26. Complete Distribution EDA Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">histogram + KDE + rug</h5>

Suppose we want to understand restaurant bill distribution.

```text
Histogram → frequency
KDE       → smooth shape
Rug       → individual observations
```


````python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True
)

sns.rugplot(
    data=tips,
    x="total_bill"
)

plt.xlabel("Total Bill")
plt.ylabel("Count")
plt.title("Complete Distribution Analysis")

plt.show()
````
