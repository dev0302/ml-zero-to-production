<h5 style="color:#78B89A; font-weight:bold;">1. What is a CSV file? → Comma Separated Values</h5>

CSV stands for **Comma Separated Values**.

- Rows represent observations.
- Columns represent features.
- Values are generally separated by commas.

In Machine Learning:
- rows → records/observations
- columns → features
- target column → output to predict

Our dataset is `aug_train.csv`.

<h5 style="color:#78B89A; font-weight:bold;">5. `sep` → separator between columns</h5>

By default, Pandas expects commas.


Other common separators:

| File | Separator |
|---|---|
| CSV | `,` |
| TSV | `\t` |
| Semicolon file | `;` |
| Pipe-separated file | `|` |

For a TSV file:

```python

```

<h3 style="color:#6FA8DC; font-weight:bold">HEADER AND INDEX</h3>

<h5 style="color:#78B89A; font-weight:bold;">6. `header` → select the row containing column names</h5>

Normally, the first row is treated as the header.

```python
df = pd.read_csv("file.csv", header=0)
```

If the file has no header:

```python
df = pd.read_csv("file.csv", header=None)
```

If the third row is the header:

```python
df = pd.read_csv("file.csv", header=2)
```

Remember: indexing starts from zero.

````python
# File has no header
# df = pd.read_csv("file.csv", header=None)

# Third row is header
# df = pd.read_csv("file.csv", header=2)
````

<h5 style="color:#78B89A; font-weight:bold;">7. If the first data row gets treated as the header → fix it</h5>

Use `header=None` and provide custom names.

```python
df = pd.read_csv(
    "file.csv",
    header=None,
    names=["Name", "Age", "City"]
)
```

This prevents the first data row from becoming column names.

<h5 style="color:#78B89A; font-weight:bold;">8. `index_col` → make a column the index</h5>

By default, Pandas creates an index:

<h3 style="color:#6FA8DC; font-weight:bold">SQUEEZE</h3>

<h5 style="color:#78B89A; font-weight:bold;">10. `squeeze` → convert one-column DataFrame into Series</h5>

Older Pandas tutorials may show:

```python
s = pd.read_csv(
    "file.csv",
    usecols=["training_hours"],
    squeeze=True
)
```

⚠️ `squeeze` was removed from modern Pandas.

Modern replacement:

```python
s = pd.read_csv(
    "file.csv",
    usecols=["training_hours"]
).squeeze("columns")
```

Or simply:

```python
s = pd.read_csv("file.csv")["training_hours"]
```

<h3 style="color:#6FA8DC; font-weight:bold">HANDLING DATES</h3>

<h5 style="color:#78B89A; font-weight:bold;">16. `parse_dates` → convert date columns while reading</h5>

CSV files often store dates as strings.

```python
df = pd.read_csv(

```

After conversion, we can:
- extract year/month/day
- sort by date
- calculate time differences
- perform time-series analysis

```python
df["Date"].dt.year
```

<h3 style="color:#6FA8DC; font-weight:bold">CONVERTERS</h3>

<h5 style="color:#78B89A; font-weight:bold;">17. `converters` → apply a custom function during import</h5>

<h3 style="color:#6FA8DC; font-weight:bold">NA_VALUES</h3>

<h5 style="color:#78B89A; font-weight:bold;">18. `na_values` → define custom missing values</h5>

Missing values may appear as:

```text
NA, N/A, ?, unknown, missing, -
```

Use:

```python
df = pd.read_csv(
    "file.csv",
    na_values=["?", "unknown", "missing", "-"]
)
```

Column-specific values:

```python
df = pd.read_csv(
    "file.csv",
    na_values={
        "gender": ["?", "unknown"],
        "experience": ["-", "missing"]
    }
)
```

⭐ Very important before data cleaning.

<h3 style="color:#6FA8DC; font-weight:bold">LOADING HUGE DATASETS</h3>

<h5 style="color:#78B89A; font-weight:bold;">19. `chunksize` → read a large CSV in parts</h5>

A huge dataset may not fit comfortably into RAM.

```python
chunks = pd.read_csv(
    "large_file.csv",
    chunksize=10000
)
```

This returns an iterable of DataFrame chunks.

```python
for chunk in chunks:
    print(chunk.shape)
```

Advantages:
- saves memory
- handles very large datasets
- supports batch processing
- useful in data pipelines

⭐ VVVV Important for real-world Data Engineering and ML.

````python
# Example
chunks = pd.read_csv("aug_train.csv", chunksize=1000)

for chunk in chunks:
    print(chunk.shape)
    break
````

<h5 style="color:#78B89A; font-weight:bold;">20. Processing chunks → count target values</h5>

We can process each chunk separately instead of loading the complete file.

```python
total = 0

for chunk in pd.read_csv("aug_train.csv", chunksize=1000):
    total += (chunk["target"] == 1).sum()

print(total)
```

````python
total = 0

for chunk in pd.read_csv("aug_train.csv", chunksize=1000):
    total += (chunk["target"] == 1).sum()

print("Total target=1 rows:", total)
````
