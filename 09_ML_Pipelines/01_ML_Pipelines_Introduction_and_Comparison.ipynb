<h3 style="color:#6FA8DC; font-weight:bold">01. ML Pipelines — Introduction</h3>

<h5 style="color:#78B89A; font-weight:bold;">ML Pipeline → short notes</h5>

**ML Pipeline = a sequence of ML steps connected together in one workflow.**

```text
RAW DATA
   ↓
Preprocessing
   ↓
Feature Transformation
   ↓
ML Model
   ↓
Prediction
```

### Why do we need it?

Without a Pipeline, we manually perform and remember every preprocessing step.

With a Pipeline, preprocessing + model are packaged together.

```text
WITHOUT PIPELINE

Raw data
   ↓
Impute manually
   ↓
Encode manually
   ↓
Scale manually
   ↓
Model
   ↓
Prediction
```

```text
WITH PIPELINE

Raw data
   ↓
┌─────────────────────────┐
│       PIPELINE          │
│ Imputer → Encoder →     │
│ Scaler → Model          │
└─────────────────────────┘
   ↓
Prediction
```

### Main difference

| Without Pipeline | With Pipeline |
|---|---|
| Steps handled separately | Steps connected together |
| More manual code | Cleaner workflow |
| Prediction preprocessing is manual | `pipeline.predict()` handles it |
| Easier to accidentally use inconsistent preprocessing | Same fitted preprocessing is reused |
| Harder to manage many steps | Easier to manage complex workflows |
| Leakage must be carefully prevented manually | Fits naturally into leakage-safe cross-validation when used correctly |

<h5 style="color:#78B89A; font-weight:bold;">Most important idea</h5>

> **Training and prediction must use the same preprocessing logic.**

```text
TRAINING:
raw → imputer → encoder → scaler → model

PREDICTION:
new raw → SAME imputer → SAME encoder → SAME scaler → model
```

A fitted Pipeline keeps this workflow together.

<h5 style="color:#78B89A; font-weight:bold;">Pipeline vs ColumnTransformer</h5>

**Pipeline:** what happens **one after another**.

```text
Imputer → Scaler → Model
```

**ColumnTransformer:** which columns get **which preprocessing**.

```text
Numerical   → Imputer → Scaler
Categorical → Imputer → OneHotEncoder
```

In real ML:

```text
ColumnTransformer
       ↓
Preprocessing
       ↓
Pipeline
       ↓
Model
```

<h5 style="color:#78B89A; font-weight:bold;">Modern scikit-learn workflow</h5>

A common structure is:

```python
Pipeline([
    ("preprocessing", ColumnTransformer(...)),
    ("model", estimator)
])
```

Useful for:
- reproducible training
- cross-validation
- hyperparameter tuning
- deployment/inference
- keeping preprocessing and model together

**Important:** Pipeline does not automatically fix bad preprocessing. You still need a correct train/test split and must avoid data leakage.
