<h3 style="color:#6FA8DC; font-weight:bold">1. 3D Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a 3D plot?</h5>

A normal graph usually works with:

```text
X → horizontal
Y → vertical
```

A **3D plot** adds another dimension:

```text
X → first dimension
Y → second dimension
Z → third dimension
```

So instead of visualizing only two variables, we can visualize a relationship involving **three dimensions**.

In Matplotlib, 3D plots are created using:

```python
projection="3d"
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">why do we need 3D plots?</h5>

3D plots can help us understand:

- Relationships between 3 numerical variables
- Shapes/surfaces
- Mathematical functions
- Optimization landscapes
- Spatial data
- How a target changes with two input variables

<br>

<h5 style="color:#78B89A; font-weight:bold;">when to use?</h5>

Use a 3D plot when the **third dimension adds useful information**.

Examples:

- `X`, `Y` → input features and `Z` → target
- Latitude, longitude, elevation
- Temperature over location
- Mathematical surfaces
- Loss/error as a function of two parameters

⚠️ **Important**

3D is not automatically better than 2D.

If the same information can be understood more clearly using a 2D plot, a 2D plot is often preferable.

⭐ **VVVV Important**

**3D Plot → visualize relationships involving X, Y and Z dimensions.**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Creating a 3D Axes</h3>

<h5 style="color:#78B89A; font-weight:bold;">projection="3d"</h5>

First create a figure:

```python
fig = plt.figure()
```

Then create a 3D axes:

```python
ax = fig.add_subplot(
    111,
    projection="3d"
)
```

Here:

```text
111 → 1 row, 1 column, 1st plot
```

The important part is:

```python
projection="3d"
```

Without it, we get a normal 2D Axes.


````python
fig = plt.figure(figsize=(8, 5))

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("Basic 3D Axes")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. 3D Line Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">ax.plot()</h5>

A 3D line plot connects points using a line in three dimensions.

Syntax:

```python
ax.plot(x, y, z)
```

So we provide **three arrays** instead of two.


````python
t = np.linspace(0, 10, 100)

x = t
y = np.sin(t)
z = np.cos(t)

fig = plt.figure(figsize=(8, 5))

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.plot(
    x,
    y,
    z
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Line Plot")

plt.show()
````

Useful when the points have an **ordered path or progression**.

Examples:

- 3D trajectory
- Movement through space
- Time-based 3-variable path
- Mathematical curves

The main syntax to remember:

```python
ax.plot(x, y, z)
```

<h3 style="color:#6FA8DC; font-weight:bold">4. 3D Surface Plot ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a surface plot?</h5>

A **3D surface plot** represents a surface where:

```text
Z = f(X, Y)
```

This means the Z value depends on X and Y.

For example:

```python
Z = sin(sqrt(X² + Y²))
```

A surface plot is one of the most useful 3D visualizations for understanding a function of **two input variables**.

⭐ **VVVV Important**

**3D Surface Plot → extremely useful when Z depends on X and Y.**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">5. meshgrid()</h3>

Before making many surface plots, we commonly use:

```python
np.meshgrid()
```

Example:

```python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)
```

`meshgrid()` creates a grid of X and Y coordinates.

Then we calculate:

```python
Z = f(X, Y)
```

So:

```text
X + Y
 ↓
function
 ↓
Z
 ↓
surface
```


````python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)

Z = np.sin(
    np.sqrt(X**2 + Y**2)
)

print(X.shape)
print(Y.shape)
print(Z.shape)
````

All three arrays have compatible 2D shapes, allowing Matplotlib to construct the surface.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">6. Creating a 3D Surface Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">plot_surface()</h5>

Syntax:

```python
ax.plot_surface(X, Y, Z)
```



````python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)

Z = np.sin(
    np.sqrt(X**2 + Y**2)
)

fig = plt.figure(figsize=(8, 5))

ax = fig.add_subplot(
    111,
    projection="3d"
)

ax.plot_surface(
    X,
    Y,
    Z
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Surface Plot")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">7. Surface Plot with cmap ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">colour represents Z values</h5>

We can use a colormap:

```python
cmap="viridis"
```

Example:

```python
surface = ax.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis"
)
```

Then add a colorbar:

```python
fig.colorbar(surface)
```

This helps us understand which colors correspond to which Z values.


````python
fig = plt.figure(figsize=(8, 5))

ax = fig.add_subplot(
    111,
    projection="3d"
)

surface = ax.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis"
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Surface Plot with Colormap")

fig.colorbar(
    surface,
    ax=ax,
    shrink=0.65,
    label="Z value"
)

plt.show()
````

This is an important combination:

```text
X → position
Y → position
Z → height
Color → Z value
```

So color gives us an additional visual cue about the surface values.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">8. 3D Color Plot / 3D Scatter</h3>

<h5 style="color:#78B89A; font-weight:bold;">ax.scatter()</h5>

We can also represent 3D observations as points.

Syntax:

```python
ax.scatter(x, y, z)
```

And we can make the point colors depend on another numerical value:

```python
ax.scatter(
    x,
    y,
    z,
    c=z,
    cmap="viridis"
)
```


````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 15, 25, 30])
z = np.array([5, 12, 8, 18, 25])

fig = plt.figure(figsize=(8, 5))

ax = fig.add_subplot(
    111,
    projection="3d"
)

scatter = ax.scatter(
    x,
    y,
    z,
    c=z,
    cmap="plasma",
    s=80
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Color Scatter Plot")

fig.colorbar(
    scatter,
    ax=ax,
    shrink=0.65,
    label="Z value"
)

plt.show()
````


```python

```

<h3 style="color:#6FA8DC; font-weight:bold">9. Contour Plot ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a contour plot?</h5>

A **contour plot** represents a 3D surface in **2D using contour lines**.

Each contour line connects points having approximately the same Z value.

Think about a geographical map:

```text
same elevation
     ↓
one contour line
```

So instead of viewing the surface from an angle, we look at it from above.

⭐ **VVVV Important**

**Contour Plot → 2D representation of a 3D surface.**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. Creating a Contour Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.contour()</h5>

Syntax:

```python
plt.contour(X, Y, Z)
```



````python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)

Z = np.sin(
    np.sqrt(X**2 + Y**2)
)

fig, ax = plt.subplots(figsize=(8, 5))

contour = ax.contour(
    X,
    Y,
    Z,
    levels=15
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Contour Plot")

plt.colorbar(
    contour,
    ax=ax,
    label="Z value"
)

plt.show()
````

````python
display(Image(filename="matplotlib_3d_heatmap_images/contour.png"))
````

### levels

```python
levels=15
```

controls approximately how many contour levels are drawn.

More levels:

→ more detailed contour representation

Fewer levels:

→ simpler visualization


<br>

<h3 style="color:#6FA8DC; font-weight:bold">11. clabel() → Values on Contour Lines</h3>

<h5 style="color:#78B89A; font-weight:bold;">labeling contour lines</h5>

We can write the numerical value of each contour line using:

```python
ax.clabel(contour)
```

Example:

```python
ax.clabel(
    contour,
    inline=True,
    fontsize=8
)
```

This makes the contour plot easier to interpret.


````python
fig, ax = plt.subplots(figsize=(8, 5))

contour = ax.contour(
    X,
    Y,
    Z,
    levels=15
)

ax.clabel(
    contour,
    inline=True,
    fontsize=8
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Labeled Contour Plot")

plt.colorbar(
    contour,
    ax=ax,
    label="Z value"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">12. Heatmap ⭐</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a heatmap?</h5>

A **heatmap** represents numerical values using colors.

Instead of showing height like a 3D surface, we use:

```text
low value  → one shade
high value → another shade
```

A heatmap is usually viewed in **2D**.

It is very useful for:

- Correlation matrices
- Feature relationships
- Confusion matrices
- Frequency tables
- Missing-value patterns
- Matrix data
- EDA

⭐ **VVVV Important**

**Heatmap → values represented by color intensity.**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">13. Creating a Heatmap with imshow()</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.imshow()</h5>

A simple Matplotlib heatmap can be created using:

```python
plt.imshow(data)
```

For example:


````python
data = np.array([
    [10, 20, 30, 40],
    [20, 35, 45, 50],
    [30, 45, 60, 70],
    [40, 50, 70, 90]
])

plt.figure(figsize=(8, 5))

heatmap = plt.imshow(
    data,
    cmap="viridis"
)

plt.xlabel("Column")
plt.ylabel("Row")
plt.title("Heatmap")

plt.colorbar(
    heatmap,
    label="Value"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">14. Contour Plot vs Heatmap vs Surface Plot</h3>

| Plot | Main idea |
|---|---|
| 3D Surface | X + Y + Z shown as a 3D surface |
| Contour | X + Y + Z represented using contour lines |
| Heatmap | Values represented mainly using color |
| 3D Scatter | Individual observations shown as 3D points |

### Easy memory trick

**Surface → shape/height**

**Contour → levels/lines**

**Heatmap → color intensity**

**3D Scatter → individual points**

<h3 style="color:#6FA8DC; font-weight:bold">18. Complete Example — Surface + Contour + Heatmap</h3>

<h5 style="color:#78B89A; font-weight:bold;">same mathematical data, different visualizations</h5>

One of the best ways to understand these plots is to use the **same X, Y and Z data** and visualize it in different ways.

```text
Same Z = f(X, Y)

        ↓

3D Surface
Contour
Heatmap
```

Each visualization gives a different perspective on the same data.


````python
x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)

X, Y = np.meshgrid(x, y)

Z = np.sin(
    np.sqrt(X**2 + Y**2)
)

# 3D Surface
fig = plt.figure(figsize=(8, 5))

ax = fig.add_subplot(
    111,
    projection="3d"
)

surface = ax.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis"
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("Surface: Z = f(X, Y)")

fig.colorbar(
    surface,
    ax=ax,
    shrink=0.65,
    label="Z value"
)

plt.show()
````
