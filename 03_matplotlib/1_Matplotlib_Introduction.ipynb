<h3 style="color:#6FA8DC; font-weight:bold">1. Matplotlib → Introduction</h3>

<h5 style="color:#78B89A; font-weight:bold;">What is Matplotlib?</h5>

**Matplotlib** is a Python library used for **data visualization**.

It helps us represent data using graphs and charts so that patterns, trends, relationships and distributions are easier to understand.

Instead of looking at a large amount of numerical data in a table, we can visualize it using graphs.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Why do we need Data Visualization?</h5>

When a dataset contains hundreds or thousands of values, it can be difficult to understand the data just by looking at numbers.

Visualization helps us identify:

- increasing or decreasing trends
- highest and lowest values
- relationships between variables
- distribution of values
- unusual values or outliers
- patterns in the data

A graph can make these patterns much easier to see.

<br>


<h5 style="color:#78B89A; font-weight:bold;">What can we create using Matplotlib?</h5>

Matplotlib can create many types of visualizations:

- **Line Plot** → shows trends and changes
- **Bar Chart** → compares categories
- **Histogram** → shows distribution of numerical data
- **Scatter Plot** → shows relationship between two variables
- **Pie Chart** → shows proportions
- **Box Plot** → helps understand distribution and outliers

There are many other specialized plots as well.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Matplotlib in Data Science</h5>

Matplotlib is an important library in the Python data science ecosystem.

A common workflow is:

```text
Data
  ↓
NumPy / Pandas
  ↓
Data Cleaning
  ↓
Data Analysis
  ↓
Matplotlib / Seaborn
  ↓
Visualization
  ↓
Understand Patterns
```

We generally use **NumPy and Pandas** to work with data and **Matplotlib** to visualize it.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Matplotlib in Machine Learning</h5>

Matplotlib is useful throughout the ML workflow.

Before training a model, visualization can help us understand:

- feature distributions
- relationships between features
- outliers
- class distributions
- trends in data

After training a model, graphs can also help us visualize predictions and model performance.

Matplotlib itself is **not a Machine Learning library**. It is a visualization library that supports the ML workflow.

⭐⭐⭐⭐⭐ **VVVV Important for ML**

Visualization helps us **understand the data before applying Machine Learning algorithms**.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Matplotlib vs Pandas</h5>

Both can be used for plotting, but their main purposes are different.

| Pandas | Matplotlib |
|---|---|
| Data manipulation and analysis | Data visualization |
| Series and DataFrame | Graphs and charts |
| Filtering, sorting, cleaning | Plotting and customization |
| `df.head()` | `plt.plot()` |
| `df.describe()` | `plt.hist()` |

Pandas also provides simple plotting methods which internally use Matplotlib.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Matplotlib vs Seaborn</h5>

**Matplotlib** is a general-purpose visualization library that gives detailed control over plots.

**Seaborn** is built on top of Matplotlib and provides a higher-level interface for statistical visualizations.

A simple way to remember:

```text
Matplotlib → Base / detailed control
Seaborn    → Easier statistical visualization
```

We will first learn Matplotlib because it gives us a strong foundation for Python plotting.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Basic Matplotlib Terminology</h5>

Before learning the functions, understand these terms:

**Figure** → The complete canvas containing one or more plots.

**Axes** → The actual area where data is plotted.

**X-axis** → Usually represents the independent variable or categories.

**Y-axis** → Usually represents the measured or dependent value.

**Title** → Describes what the graph represents.

**Label** → Describes an axis.

**Legend** → Helps identify different data series.

**Marker** → A symbol representing an individual data point.

**Line** → Connects data points in a line plot.

<br>


<h5 style="color:#78B89A; font-weight:bold;">Figure vs Axes</h5>

Think of a Matplotlib visualization like this:

```text
Figure
┌───────────────────────────────┐
│                               │
│        Axes / Plot            │
│       ┌──────────────┐        │
│       │              │        │
│       │    Graph     │        │
│       │              │        │
│       └──────────────┘        │
│                               │
└───────────────────────────────┘
```

Remember:

```text
Figure → complete canvas
Axes   → actual plotting area
```

A single Figure can contain multiple Axes.

<h5 style="color:#78B89A; font-weight:bold;">Matplotlib and EDA</h5>

**EDA = Exploratory Data Analysis**

EDA means exploring a dataset before building a model or making conclusions.

Visualization is a major part of EDA.

For example:

```text
What does the data look like?
        ↓
How are values distributed?
        ↓
Are there outliers?
        ↓
Are variables related?
        ↓
Are there visible patterns?
```

Matplotlib helps us answer these questions visually.
