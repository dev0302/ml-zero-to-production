<h3 style="color:#6FA8DC; font-weight:bold">2. Matplotlib → Types of Data and Data Analysis</h3>

<h5 style="color:#78B89A; font-weight:bold;">Why do we need to know the type of data first?</h5>

Matplotlib has many different graphs. As we learn more graphs, one common confusion comes:

> **Which graph should I use for my data?**

Before choosing a graph, we should first understand **what type of data we have** and **how many variables we are studying**.

```text
Understand the data
       ↓
Identify data type
       ↓
Identify number of variables
       ↓
Decide what we want to understand
       ↓
Choose an appropriate visualization
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">1. Types of Data</h5>

At a basic level, data can be divided into:

```text
Data
├── Numerical Data
└── Categorical Data
```

These two categories are very important when working with datasets and choosing visualizations.


<h5 style="color:#78B89A; font-weight:bold;">2. Numerical Data</h5>

**Numerical data** is data represented using numbers where the numerical value has a meaningful mathematical interpretation.

Examples:

- Age → `18, 21, 25`
- Height → `165.5, 170.2`
- Salary → `30000, 45000`
- Runs → `20, 45, 102`

Numerical data is mainly of two types:

```text
Numerical
├── Discrete
└── Continuous
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">3. Discrete Numerical Data</h5>

**Discrete data** consists of countable values.

Examples:

- Number of students → `40`
- Number of goals → `3`
- Number of cars → `12`
- Number of children → `2`

These values are normally obtained by **counting**.

⭐ **Remember:**

```text
Discrete → Count
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">4. Continuous Numerical Data</h5>

**Continuous data** can take any value within a range.

It usually represents something that we **measure**.

Examples:

- Height → `172.5 cm`
- Weight → `68.7 kg`
- Temperature → `36.6°C`
- Time → `12.53 seconds`

Between two values, many possible values can exist.

⭐ **Remember:**

```text
Continuous → Measure
```

<br>

### Discrete vs Continuous

| Discrete | Continuous |
|---|---|
| Counted | Measured |
| Separate/countable values | Any value within a range |
| Number of students | Height |
| Number of goals | Weight |
| Number of products | Temperature |



<h5 style="color:#78B89A; font-weight:bold;">5. Categorical Data</h5>

**Categorical data** represents groups, labels or categories rather than numerical quantities.

Examples:

- Gender → Male, Female
- City → Delhi, Mumbai, Kolkata
- Department → CSE, ECE, IT
- Blood Group → A, B, AB, O
- Product Type → Mobile, Laptop, Tablet

Even if categories are represented using numbers, those numbers may simply be **labels**.

For example:

```text
CSE → 1
ECE → 2
IT  → 3
```

Here `3` does not mean that IT is mathematically greater than ECE. The numbers are simply category codes.

Categorical data is mainly divided into:

```text
Categorical
├── Nominal
└── Ordinal
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">6. Nominal Categorical Data</h5>

**Nominal data** contains categories with **no natural order or ranking**.

Examples:

- City → Delhi, Mumbai, Chennai
- Blood Group → A, B, AB, O
- Color → Red, Blue, Green
- Department → CSE, ECE, IT

There is no meaningful ranking between these categories.

⭐ **Remember:**

```text
Nominal → Names / Labels
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">7. Ordinal Categorical Data</h5>

**Ordinal data** contains categories that have a **meaningful order or ranking**.

Examples:

- Satisfaction → Poor, Average, Good, Excellent
- Education → School, Bachelor's, Master's, PhD
- Rating → 1 Star, 2 Stars, 3 Stars, 4 Stars, 5 Stars
- Size → Small, Medium, Large

There is an order:

```text
Poor < Average < Good < Excellent
```

But the difference between categories is not necessarily numerically equal.

⭐ **Remember:**

```text
Ordinal → Ordered Categories
```

<br>

### Nominal vs Ordinal

| Nominal | Ordinal |
|---|---|
| No natural order | Has natural order |
| City | Satisfaction |
| Blood group | Education level |
| Color | Rating |
| Department | Size |



<h5 style="color:#78B89A; font-weight:bold;">8. Complete Data Classification</h5>

We can remember the structure as:

```text
Data
│
├── Numerical
│   │
│   ├── Discrete → Count
│   │
│   └── Continuous → Measure
│
└── Categorical
    │
    ├── Nominal → No order
    │
    └── Ordinal → Has order
```

⭐⭐⭐⭐⭐ **VVVV Important**

This classification helps us understand what type of analysis and visualization may be suitable.


<h5 style="color:#78B89A; font-weight:bold;">9. Univariate Data</h5>

**Univariate** means we are studying **one variable**.

```text
Uni → One
```

Example:

```text
Age
18, 19, 20, 18, 21, 20, 19
```

Here we are studying only one variable:

```text
Age
```

Questions could be:

- What is the distribution of age?
- What is the average age?
- What is the minimum/maximum age?
- Which value occurs most often?

The focus is on understanding **one variable**.

Examples:

```text
Age
Salary
Marks
Gender
City
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">10. Bivariate Data</h5>

**Bivariate** means we are studying **two variables**.

```text
Bi → Two
```

Example:

```text
Height + Weight
```

We may want to understand whether height and weight are related.

Other examples:

- Hours studied + Marks
- Age + Salary
- Advertising cost + Sales
- Temperature + Ice cream sales

The main purpose of bivariate analysis is often to understand the **relationship or comparison between two variables**.

⭐ **Remember:**

```text
Univariate → 1 variable
Bivariate  → 2 variables
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">11. Multivariate Data</h5>

**Multivariate** means we are studying **more than two variables**.

Example:

```text
Age
Salary
Experience
Education
```

We are considering multiple variables together.

A Machine Learning dataset might contain:

```text
Age
Height
Weight
Salary
Experience
City
```

Multivariate analysis helps us understand how several variables interact with or relate to one another.

⭐ **Remember:**

```text
Univariate   → 1 variable
Bivariate    → 2 variables
Multivariate → More than 2 variables
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">12. Why Univariate, Bivariate and Multivariate Matter?</h5>

The number of variables changes the question we are asking.

### Univariate

```text
One variable
   ↓
Understand that variable
```

Example:

> What is the distribution of student marks?

### Bivariate

```text
Two variables
   ↓
Understand their relationship / comparison
```

Example:

> Does study time relate to marks?

### Multivariate

```text
Multiple variables
   ↓
Understand interactions among several variables
```

Example:

> How do study time, attendance and previous marks together relate to final marks?

This is why identifying the number of variables is important before choosing a visualization.

<br>


<h5 style="color:#78B89A; font-weight:bold;">13. Data Type + Number of Variables</h5>

Before choosing a graph, think about **two things**:

### 1. What type of data do I have?

```text
Numerical?
Categorical?
```

### 2. How many variables am I studying?

```text
One?
Two?
More than two?
```

Examples:

```text
Age
→ Numerical
→ Univariate
```

```text
Gender
→ Categorical
→ Univariate
```

```text
Height + Weight
→ Numerical + Numerical
→ Bivariate
```

```text
Department + Salary
→ Categorical + Numerical
→ Bivariate
```

This gives us a much better idea of what kind of visualization may be appropriate.

<h5 style="color:#78B89A; font-weight:bold;">15. Main Idea Before Learning Graphs</h5>

Do **not** try to memorize graphs randomly.

First ask:

```text
What is my data?
       ↓
Numerical or Categorical?
       ↓
If numerical → Discrete or Continuous?
If categorical → Nominal or Ordinal?
       ↓
How many variables?
       ↓
Univariate / Bivariate / Multivariate
       ↓
What do I want to understand?
       ↓
Choose the graph
```

⭐⭐⭐⭐⭐ **VVVV Important**

```text
Data Type + Number of Variables + Goal
                    ↓
          Appropriate Visualization
```

Now we can start learning the actual Matplotlib graphs one by one.
