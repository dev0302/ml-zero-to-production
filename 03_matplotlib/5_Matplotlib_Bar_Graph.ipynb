<h3 style="color:#6FA8DC; font-weight:bold">1. Bar Graph</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a bar graph?</h5>

A **bar graph** is used to compare numerical values across different **categories**.

Each category is represented by a rectangular bar.

The height of the bar represents the value.

Example:

- X-axis → Batter
- Y-axis → Runs
- Bar → Runs scored by that batter

<br>

<h5 style="color:#78B89A; font-weight:bold;">when to use a bar graph?</h5>

Use a bar graph when:

- You want to **compare categories**
- Your X-axis is generally **categorical**
- You want to compare quantities
- You want to find the largest/smallest category
- You want to show rankings

<br>

<h5 style="color:#78B89A; font-weight:bold;">why use a bar graph?</h5>

Bar graphs make comparison very easy.

⭐ **VVVV Important**

**Bar Graph → Compare numerical values across categories**


<h5 style="color:#78B89A; font-weight:bold;">our dataset → batter.csv</h5>

We will use `batter.csv`.

Columns:

- `batter` → batter name
- `runs` → total runs
- `avg` → batting average
- `strike_rate` → strike rate

We will mainly use `batter` as the category and `runs` as the numerical value.


````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("datasets/batter.csv")

df.head()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Creating a Basic Bar Graph</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.bar()</h5>

Basic syntax:

```python
plt.bar(x, height)
```

- `x` → categories
- `height` → numerical values


````python
top10 = df.head(10)

plt.bar(
    top10["batter"],
    top10["runs"]
)

plt.show()
````

Use:

````python
plt.bar(top10["batter"], top10["runs"])

plt.xlabel("Batter")
plt.ylabel("Runs")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">4. Title</h3>

<h3 style="color:#6FA8DC; font-weight:bold">5. Providing Color</h3>

<h5 style="color:#78B89A; font-weight:bold;">color parameter</h5>

We can change bar color using `color`.


````python
plt.bar(
    top10["batter"],
    top10["runs"],
    color="skyblue"
)

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Top 10 IPL Batters by Runs")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">8. Aligning Bars</h3>

<h5 style="color:#78B89A; font-weight:bold;">align parameter</h5>

`align` controls how the bar is positioned around its X coordinate.

Common values:

- `"center"`
- `"edge"`


````python
plt.bar(
    players,
    runs,
    align="center"
)

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Bar Alignment")
plt.xticks(rotation=30, ha="right")

plt.show()
````

`align="center"` → category position is at the center of the bar.

`align="edge"` → category position is aligned with the edge of the bar.

<h3 style="color:#6FA8DC; font-weight:bold">13. Label and Legend</h3>

<h5 style="color:#78B89A; font-weight:bold;">label parameter</h5>

`label` gives a name to the plotted bar series.

````python
plt.bar(
    players,
    runs,
    color="skyblue",
    label="Runs"
)

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Runs by Batter")

plt.legend()

plt.xticks(rotation=30, ha="right")
plt.show()
````


- `loc` → legend position
- `fontsize` → legend text size

````python
plt.bar(
    players,
    runs,
    color="skyblue",
    label="Runs"
)

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Runs by Batter")

plt.legend(
    loc="upper right",
    fontsize=10,
    title="Metric"
)

plt.xticks(rotation=30, ha="right")
plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">14. Adding Values on Top of Bars</h3>

<h5 style="color:#78B89A; font-weight:bold;">bar objects + plt.text()</h5>

`plt.bar()` returns bar objects.

We can loop through these objects and write the exact value above each bar.


````python
bars = plt.bar(
    players,
    runs,
    color="skyblue"
)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Bar Chart with Values")
plt.xticks(rotation=30, ha="right")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">15. Grouped Bar Graph</h3>

<h5 style="color:#78B89A; font-weight:bold;">multiple bars for each category</h5>

A grouped bar graph places multiple bars next to each other for the same category.

Example:

- One bar → Runs
- One bar → Average

We shift the X positions slightly so the bars appear side by side.


````python
players = ["V Kohli", "S Dhawan", "DA Warner", "RG Sharma", "SK Raina"]

data = df.set_index("batter")

runs = data.loc[players, "runs"]
avg = data.loc[players, "avg"]

x = list(range(len(players)))
width = 0.35

plt.bar(
    [i - width/2 for i in x],
    runs / 100,
    width=width,
    label="Runs / 100"
)

plt.bar(
    [i + width/2 for i in x],
    avg,
    width=width,
    label="Average"
)

plt.xticks(x, players, rotation=30, ha="right")
plt.ylabel("Value")
plt.title("Grouped Bar Chart")
plt.legend()

plt.show()
````

````python
display(Image(filename="matplotlib_bargraph_images/grouped_bar.png"))
````

### Why divide runs by 100?

Runs and average have very different scales.

For this demonstration, `runs / 100` makes the two values easier to compare visually.

In real analysis, variables with very different units/scales may be better shown in separate graphs.

<h3 style="color:#6FA8DC; font-weight:bold">16. Stacked Bar Graph</h3>

<h5 style="color:#78B89A; font-weight:bold;">showing parts of a total</h5>

A stacked bar graph places one bar on top of another.

It is useful when we want to show:

- Total value
- Contribution of different parts

The `bottom` parameter is the key idea.


````python
categories = ["A", "B", "C", "D", "E"]

part_a = [40, 55, 70, 60, 80]
part_b = [30, 25, 20, 35, 15]
part_c = [20, 20, 10, 5, 5]

plt.bar(categories, part_a, label="Part A")

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    label="Part B"
)

bottom2 = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_c,
    bottom=bottom2,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Stacked Bar Chart")
plt.legend()

plt.show()
````

````python
display(Image(filename="matplotlib_bargraph_images/stacked_bar.png"))
````

### `bottom` parameter

`bottom` tells Matplotlib where the next bar should start.

This is what allows multiple values to be stacked.

<h3 style="color:#6FA8DC; font-weight:bold">17. Hatch Pattern</h3>

<h5 style="color:#78B89A; font-weight:bold;">hatch → pattern inside bars</h5>

We can use patterns inside bars.

Common hatch patterns:

```text
/    \\    |    -    +    x    o    O    .    *
```


````python
plt.bar(
    players,
    runs,
    color="skyblue",
    edgecolor="black",
    hatch="/"
)

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Bar Graph with Hatch Pattern")

plt.xticks(rotation=30, ha="right")
plt.show()
````

Hatching can be useful when a graph should remain understandable without depending only on color.

<h3 style="color:#6FA8DC; font-weight:bold">20. Bar Graph vs Scatter Plot vs Line Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">choosing the correct graph</h5>

| Graph | Main purpose | Example |
|---|---|---|
| Bar | Compare categories | Runs of players |
| Scatter | Relationship between 2 numerical variables | Average vs strike rate |
| Line | Trend/progression | Sales over months |

### Simple rule

**Category comparison → Bar**

**Numerical relationship → Scatter**

**Trend over ordered/time data → Line**

<h3 style="color:#6FA8DC; font-weight:bold">21. Complete Bar Graph Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">putting important concepts together</h5>


````python
top10 = df.head(10)

plt.bar(
    top10["batter"],
    top10["runs"],
    color="skyblue",
    edgecolor="black",
    width=0.7,
    alpha=0.85,
    label="Runs"
)

plt.xlabel("Batter")
plt.ylabel("Runs")
plt.title("Top 10 IPL Batters by Runs")

plt.legend(
    loc="upper right",
    fontsize=10,
    title="Metric"
)

plt.grid(axis="y")

plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()

plt.show()
````
