<h3 style="color:#6FA8DC; font-weight:bold">1. Pie Chart</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a pie chart?</h5>

A **pie chart** is used to show how different categories contribute to a **whole**.

The complete circle represents **100%**.

Each slice represents one category's share of that total.

Example:

```text
Total = 100%

Category A → 40%
Category B → 30%
Category C → 20%
Category D → 10%
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">when to use?</h5>

Use a pie chart when:

- You want to show **part-to-whole** relationships
- There are only a few categories
- All values belong to the same meaningful total
- Percentages/shares are more important than exact comparison

Examples:

- Market share
- Budget distribution
- Website traffic sources
- Student distribution
- Expense distribution

<br>

<h5 style="color:#78B89A; font-weight:bold;">when NOT to use?</h5>

Avoid pie charts when:

- There are too many categories
- Exact comparison is important
- Values do not represent parts of one meaningful whole
- You are showing change over time

In these cases, a **bar chart** is often easier to read.

⭐ **VVVV Important**

**Pie Chart → Part of a whole**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Creating a Basic Pie Chart</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.pie()</h5>

Basic syntax:

```python
plt.pie(values)
```

We can also provide labels:

```python
plt.pie(values, labels=labels)
```


````python
labels = ["V Kohli", "S Dhawan", "DA Warner", "RG Sharma", "SK Raina"]
values = [25, 20, 18, 22, 15]

plt.pie(
    values,
    labels=labels
)

plt.show()
````

````python
from IPython.display import Image, display

display(Image(filename="matplotlib_pie_chart_images/basic_pie.png"))
````

The values are automatically converted into slices.

Here:

```text
25 + 20 + 18 + 22 + 15 = 100
```

So each value directly represents its percentage.

If the total is not 100, Matplotlib still calculates the percentage automatically.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. labels → Naming Each Slice</h3>

<h5 style="color:#78B89A; font-weight:bold;">labels parameter</h5>

Without labels, it can be difficult to know what each slice represents.

Use:

```python
labels=["A", "B", "C"]
```


````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels
)

plt.title("Category Distribution")

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">4. autopct → Showing Percentages</h3>

<h5 style="color:#78B89A; font-weight:bold;">autopct parameter ⭐</h5>

`autopct` displays the percentage inside each slice.

Example:

```python
autopct="%1.1f%%"
```

Meaning:

- `%1.1f` → show one decimal place
- `%%` → display `%`


````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%"
)

plt.title("Category Distribution")

plt.show()
````

Other useful formats:

```python
autopct="%1.0f%%"
```

→ no decimal places

```python
autopct="%1.2f%%"
```

→ two decimal places

⭐ **VVVV Important**

`autopct` is very commonly used in pie charts.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">5. startangle → Rotating the Pie Chart</h3>

<h5 style="color:#78B89A; font-weight:bold;">startangle parameter</h5>

`startangle` controls where the first slice starts.



````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Rotated Pie Chart")

plt.show()
````

For example:

```python
startangle=90
```

rotates the starting position by 90 degrees.

This is useful when you want the chart to have a cleaner orientation.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">6. explode → Highlighting a Slice</h3>

<h5 style="color:#78B89A; font-weight:bold;">explode parameter ⭐</h5>

`explode` moves one or more slices slightly away from the center.

Example:

```python
explode=[0.1, 0, 0, 0]
```

The first slice is separated from the rest.


````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    explode=[0.1, 0, 0, 0],
    startangle=90
)

plt.title("Exploded Pie Chart")

plt.show()
````

For multiple slices:

```python
explode=[0.1, 0, 0.1, 0]
```

This separates the first and third slices.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">7. shadow → Shadow Effect</h3>

<h5 style="color:#78B89A; font-weight:bold;">shadow parameter</h5>

Use:

```python
shadow=True
```

to add a shadow beneath the pie chart.


````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    shadow=True
)

plt.title("Pie Chart with Shadow")

plt.show()
````

The shadow is mainly a visual customization. It is not necessary for data analysis.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">8. Colors</h3>

<h5 style="color:#78B89A; font-weight:bold;">colors parameter</h5>

We can provide a list of colors for the slices.



````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    colors=["skyblue", "orange", "lightgreen", "pink"],
    autopct="%1.1f%%"
)

plt.title("Colored Pie Chart")

plt.show()
````

Each color corresponds to one slice.

The order is:

```text
A → first color
B → second color
C → third color
D → fourth color
```


<br>

<h3 style="color:#6FA8DC; font-weight:bold">9. wedgeprops → Borders</h3>

<h5 style="color:#78B89A; font-weight:bold;">wedgeprops parameter</h5>

`wedgeprops` allows us to customize the individual pie slices.

For example, we can add borders.


````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1
    }
)

plt.title("Pie Chart with Borders")

plt.show()
````

````python
display(Image(filename="matplotlib_pie_chart_images/pie_borders.png"))
````

The important properties here are:

```python
"edgecolor"
"linewidth"
```

They create a visible separation between slices.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. pctdistance and labeldistance</h3>

<h5 style="color:#78B89A; font-weight:bold;">controlling label positions</h5>

`pctdistance` controls the position of percentage text.

`labeldistance` controls the position of category labels.


````python
labels = ["A", "B", "C", "D"]
values = [40, 30, 20, 10]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    pctdistance=0.7,
    labeldistance=1.1
)

plt.title("Pie Chart with Label Positions")

plt.show()
````

````python
display(Image(filename="matplotlib_pie_chart_images/pie_distances.png"))
````

Generally:

- smaller `pctdistance` → percentage moves closer to center
- larger `pctdistance` → percentage moves toward edge
- smaller `labeldistance` → labels move closer to pie
- larger `labeldistance` → labels move farther away


<br>

<h3 style="color:#6FA8DC; font-weight:bold">11. Complete Pie Chart Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">putting important parameters together</h5>


````python
labels = ["V Kohli", "S Dhawan", "DA Warner", "RG Sharma", "SK Raina"]
values = [25, 20, 18, 22, 15]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.05, 0, 0, 0, 0],
    shadow=True,
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1
    }
)

plt.title("Top 5 Batters Distribution")

plt.show()
````
