<h3 style="color:#6FA8DC; font-weight:bold">1. Stacked Bar Chart</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a stacked bar chart?</h5>

A **stacked bar chart** is a special type of bar graph where multiple values are placed **on top of each other**.

Instead of having only one value for each category, one bar can contain multiple parts.

Example:

```text
Category A
Part A + Part B + Part C
```

So one complete bar represents the **total**, while the individual sections represent the **contribution of each part**.

<br>

<h5 style="color:#78B89A; font-weight:bold;">when to use?</h5>

Use a stacked bar chart when:

- You want to show a **total**
- You also want to show how that total is divided into **parts**
- You want to compare the composition of different categories
- You want to see how different components contribute to a total

Examples:

- Monthly expenses → food + rent + travel
- Sales → product A + product B + product C
- Students → different categories within each department
- Website traffic → different traffic sources
- Company revenue → different products

<br>

<h5 style="color:#78B89A; font-weight:bold;">why use?</h5>

A normal bar graph tells us:

**"How much is the total?"**

A stacked bar graph can tell us:

**"How much is the total AND where did that total come from?"**

⭐ **VVVV Important**

**Stacked Bar → Total + contribution of different parts**


<br>

<h3 style="color:#6FA8DC; font-weight:bold">2. Basic Idea</h3>

Suppose we have:

```text
Category A → Part A = 40, Part B = 30, Part C = 20
```

Total:

```text
40 + 30 + 20 = 90
```

The stacked bar will have:

```text
        ┌─────────┐  Part C = 20
        ├─────────┤  Part B = 30
        ├─────────┤  Part A = 40
        └─────────┘
```

The key parameter that makes stacking possible is:

```python
bottom=
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">basic syntax</h5>

```python
plt.bar(x, height)
plt.bar(x, height, bottom=previous_height)
```

The second bar starts where the first bar ends.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">3. Creating a Basic Stacked Bar Chart</h3>

Let's create some simple data first.


````python
categories = ["A", "B", "C", "D", "E"]

part_a = [40, 55, 70, 60, 80]
part_b = [30, 25, 20, 35, 15]
part_c = [20, 20, 10, 5, 5]
````

Now create the first bar.



````python
plt.bar(
    categories,
    part_a,
    label="Part A"
)

plt.show()
````

Now we add Part B.

The important thing is:

```python
bottom=part_a
```

This tells Matplotlib:

**Start Part B from the top of Part A.**


````python
plt.bar(
    categories,
    part_a,
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    label="Part B"
)

plt.show()
````

Now we add Part C.

Part C must start after both Part A and Part B.

Therefore:

```python
bottom = Part A + Part B
```


````python
bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_a,
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    label="Part B"
)

plt.bar(
    categories,
    part_c,
    bottom=bottom_c,
    label="Part C"
)

plt.show()
````

### Complete basic stacked bar


````python
plt.bar(categories, part_a, label="Part A")

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    label="Part B"
)

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_c,
    bottom=bottom_c,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Basic Stacked Bar Chart")
plt.legend()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">4. Understanding `bottom` Parameter</h3>

<h5 style="color:#78B89A; font-weight:bold;">bottom → where the next bar starts</h5>

This is the **most important parameter** in a stacked bar chart.

Suppose:

```python
part_a = [40, 55, 70]
part_b = [30, 25, 20]
```

For Part B:

```python
bottom=part_a
```

means:

```text
Part A
──────
Part B starts here
```

For Part C:

```python
bottom = part_a + part_b
```

So Part C starts after both previous parts.

<br>

⭐ **VVVV Important**

```python
plt.bar(x, part_b, bottom=part_a)
```

is the basic pattern for stacking one bar on another.

````python
plt.bar(categories, part_a, label="Part A")
plt.bar(categories, part_b, bottom=part_a, label="Part B")

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(categories, part_c, bottom=bottom_c, label="Part C")

plt.xlabel("Category")
plt.ylabel("Value")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">7. Providing Color</h3>

<h5 style="color:#78B89A; font-weight:bold;">color → identify different parts</h5>

Different colors make the different sections of a stacked bar easier to distinguish.


````python
plt.bar(
    categories,
    part_a,
    color="skyblue",
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    color="orange",
    label="Part B"
)

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_c,
    bottom=bottom_c,
    color="lightgreen",
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Colored Stacked Bar Chart")
plt.legend()

plt.show()
````

You can use either one color for a series or different colors for different series.

In a stacked bar chart, **different colors for different parts** are usually more useful because they show which part contributes to the total.

<h3 style="color:#6FA8DC; font-weight:bold">8. Edge Color and Line Width</h3>

<h5 style="color:#78B89A; font-weight:bold;">edgecolor → border of each section</h5>

<h5 style="color:#78B89A; font-weight:bold;">linewidth → thickness of border</h5>


````python
plt.bar(
    categories,
    part_a,
    color="skyblue",
    edgecolor="black",
    linewidth=1.5,
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    color="orange",
    edgecolor="black",
    linewidth=1.5,
    label="Part B"
)

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_c,
    bottom=bottom_c,
    color="lightgreen",
    edgecolor="black",
    linewidth=1.5,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Stacked Bar with Borders")
plt.legend()

plt.show()
````

The border helps visually separate adjacent sections of the same bar.

<h3 style="color:#6FA8DC; font-weight:bold">9. Alpha → Transparency</h3>

<h5 style="color:#78B89A; font-weight:bold;">alpha parameter</h5>

`alpha` controls transparency.

- `alpha=1` → fully visible
- `alpha=0.5` → semi-transparent
- `alpha=0` → invisible



````python
plt.bar(
    categories,
    part_a,
    alpha=0.7,
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    alpha=0.7,
    label="Part B"
)

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_c,
    bottom=bottom_c,
    alpha=0.7,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Transparent Stacked Bar Chart")
plt.legend()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">10. Width of Stacked Bars</h3>

<h5 style="color:#78B89A; font-weight:bold;">width parameter</h5>

All sections of the same stacked bar should normally have the **same width**.



````python
width = 0.6

plt.bar(
    categories,
    part_a,
    width=width,
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    width=width,
    bottom=part_a,
    label="Part B"
)

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_c,
    width=width,
    bottom=bottom_c,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Stacked Bar with Custom Width")
plt.legend()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">11. Legend</h3>

<h5 style="color:#78B89A; font-weight:bold;">label + plt.legend()</h5>

A legend is especially important in stacked bar charts because it tells us what each section represents.


````python
plt.bar(categories, part_a, label="Part A")
plt.bar(categories, part_b, bottom=part_a, label="Part B")

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(categories, part_c, bottom=bottom_c, label="Part C")

plt.legend(
    loc="upper right",
    fontsize=10,
    title="Parts"
)

plt.show()
````

Useful legend parameters:

- `loc` → position
- `fontsize` → text size
- `title` → legend title

<h3 style="color:#6FA8DC; font-weight:bold">13. Horizontal Stacked Bar Chart</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.barh() + left=</h5>

For a horizontal stacked bar chart, we use `plt.barh()`.

The equivalent of `bottom` is:

```python
left=
```

because the bars grow horizontally.


````python
plt.barh(
    categories,
    part_a,
    label="Part A"
)

plt.barh(
    categories,
    part_b,
    left=part_a,
    label="Part B"
)

left_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.barh(
    categories,
    part_c,
    left=left_c,
    label="Part C"
)

plt.xlabel("Value")
plt.ylabel("Category")
plt.title("Horizontal Stacked Bar Chart")
plt.legend()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">14. 100% / Percentage Stacked Bar Chart</h3>

<h5 style="color:#78B89A; font-weight:bold;">showing percentage contribution</h5>

Sometimes the actual totals are not important.

We only want to know:

**What percentage of each category comes from each part?**

Then we convert every part into a percentage of its category total.

For every category:

```text
Part A % + Part B % + Part C % = 100%
```


````python
totals = [
    part_a[i] + part_b[i] + part_c[i]
    for i in range(len(categories))
]

a_pct = [
    part_a[i] / totals[i] * 100
    for i in range(len(categories))
]

b_pct = [
    part_b[i] / totals[i] * 100
    for i in range(len(categories))
]

c_pct = [
    part_c[i] / totals[i] * 100
    for i in range(len(categories))
]

print(a_pct)
print(b_pct)
print(c_pct)
````

Now we stack the percentages.


````python
plt.bar(
    categories,
    a_pct,
    label="Part A"
)

plt.bar(
    categories,
    b_pct,
    bottom=a_pct,
    label="Part B"
)

bottom_c_pct = [
    a_pct[i] + b_pct[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    c_pct,
    bottom=bottom_c_pct,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Percentage")
plt.title("100% Stacked Bar Chart")

plt.ylim(0, 100)

plt.legend()

plt.show()
````

### When should we use 100% stacked bars?

Use them when the **composition** is more important than the absolute total.

Example:

If Department A has 100 students and Department B has 500 students, a normal stacked bar shows their different totals.

A 100% stacked bar instead focuses on:

**What percentage of students belong to each category?**

⭐ **VVVV Important**

**Normal stacked bar → compare totals + parts**

**100% stacked bar → compare proportions**

<h3 style="color:#6FA8DC; font-weight:bold">17. Stacked Bar vs Grouped Bar vs Normal Bar</h3>

<h5 style="color:#78B89A; font-weight:bold;">which one should we use?</h5>

| Graph | Main purpose |
|---|---|
| Normal Bar | Compare one value across categories |
| Grouped Bar | Compare multiple values side-by-side |
| Stacked Bar | Show total + parts of that total |
| 100% Stacked Bar | Compare percentage composition |

### Example

If we have:

```text
Product A → Online sales + Offline sales
Product B → Online sales + Offline sales
```

**Grouped bar:**

The two values appear side by side.

**Stacked bar:**

The two values are placed on top of each other and show the total.

**100% stacked bar:**

We compare the percentage split between online and offline sales.


<br>

<h3 style="color:#6FA8DC; font-weight:bold">18. Complete Stacked Bar Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">putting everything together</h5>


````python
categories = ["A", "B", "C", "D", "E"]

part_a = [40, 55, 70, 60, 80]
part_b = [30, 25, 20, 35, 15]
part_c = [20, 20, 10, 5, 5]

bottom_c = [
    part_a[i] + part_b[i]
    for i in range(len(categories))
]

plt.bar(
    categories,
    part_a,
    color="skyblue",
    edgecolor="black",
    linewidth=1,
    alpha=0.9,
    label="Part A"
)

plt.bar(
    categories,
    part_b,
    bottom=part_a,
    color="orange",
    edgecolor="black",
    linewidth=1,
    alpha=0.9,
    label="Part B"
)

plt.bar(
    categories,
    part_c,
    bottom=bottom_c,
    color="lightgreen",
    edgecolor="black",
    linewidth=1,
    alpha=0.9,
    label="Part C"
)

plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Complete Stacked Bar Chart")

plt.legend(
    loc="upper right",
    fontsize=10,
    title="Parts"
)

plt.grid(axis="y")

plt.tight_layout()

plt.show()
````

bottom = part_a + part_b
