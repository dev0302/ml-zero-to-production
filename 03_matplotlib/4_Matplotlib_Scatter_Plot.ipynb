<h3 style="color:#6FA8DC; font-weight:bold">1. Scatter Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is a scatter plot?</h5>

A **scatter plot** is used to show the relationship between **two numerical variables**.

Each point represents **one observation**.

For example:

- X-axis → Batting Average
- Y-axis → Strike Rate
- One point → One batter

A scatter plot helps us visually understand whether two variables are related or not.

<br>

<h5 style="color:#78B89A; font-weight:bold;">when to use scatter plot?</h5>

Use a scatter plot when:

- You have **two numerical variables**
- You want to check the **relationship** between them
- You want to see **patterns**
- You want to identify **outliers**
- You want to understand whether values move together

Example:

**Batting Average vs Strike Rate**

This can help us see whether batters with a higher average generally have a higher strike rate.

<br>

<h5 style="color:#78B89A; font-weight:bold;">why use scatter plot?</h5>

A scatter plot can help answer questions like:

- Is there a relationship between X and Y?
- Is the relationship positive or negative?
- Are the points randomly distributed?
- Are there clusters?
- Are there unusual/outlier points?

In **Machine Learning**, scatter plots are very useful during **EDA (Exploratory Data Analysis)** before building a model.

<h3 style="color:#6FA8DC; font-weight:bold">6. Linestyle in Scatter Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">linestyle → important difference</h5>

For a normal scatter plot, points are **not connected**, so `linestyle` normally has no visible effect on the points.

`linestyle` is mainly useful with `plt.plot()` because `plt.plot()` draws lines.

For example:



````python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 10]

plt.plot(
    x,
    y,
    linestyle="--",
    marker="o"
)

plt.show()
````

Common line styles:

- `"-"` → solid
- `"--"` → dashed
- `":"` → dotted
- `"-."` → dash-dot

⭐ In scatter plots, focus more on **marker, color, size and transparency** rather than `linestyle`.



<br>

<h3 style="color:#6FA8DC; font-weight:bold">7. Marker</h3>

<h5 style="color:#78B89A; font-weight:bold;">marker → shape of each point</h5>

The `marker` parameter controls the shape used for each observation.



````python
plt.scatter(
    df["avg"],
    df["strike_rate"],
    marker="o"
)

plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.title("Batting Average vs Strike Rate")

plt.show()
````

| `"^"` | Triangle |
| `"*"` | Star |
| `"x"` | X |
| `"+"` | Plus |
| `"D"` | Diamond |

<h3 style="color:#6FA8DC; font-weight:bold">8. Label and Legend</h3>

<h5 style="color:#78B89A; font-weight:bold;">label → name of the plotted data</h5>

We can give a name to our scatter data using `label`.

Then we use:

```python
plt.legend()
```

to display that label on the graph.



````python
plt.scatter(
    df["avg"],
    df["strike_rate"],
    color="red",
    marker="o",
    label="Batter"
)

plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.title("Batting Average vs Strike Rate")

plt.legend()

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">legend parameters</h5>

`plt.legend()` has many parameters.

Some important ones:

### 1. loc → location of legend

Examples:

```python
plt.legend(loc="upper right")
plt.legend(loc="upper left")
plt.legend(loc="lower right")
plt.legend(loc="lower left")
```

### 2. fontsize → text size

```python
plt.legend(fontsize=10)
```

### 3. title → title of legend

```python
plt.legend(title="Players")
```

We can combine them:



````python
plt.scatter(
    df["avg"],
    df["strike_rate"],
    color="red",
    marker="o",
    label="Batter"
)

plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.title("Batting Average vs Strike Rate")

plt.legend(
    loc="upper left",
    fontsize=10,
    title="Players"
)

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">9. Scatter Plot Parameters</h3>

<h5 style="color:#78B89A; font-weight:bold;">important parameters of plt.scatter()</h5>

The basic syntax can be written as:

```python
plt.scatter(
    x,
    y,
    color=...,
    marker=...,
    s=...,
    alpha=...,
    label=...
)
```

Important parameters:

| Parameter | Use |
|---|---|
| `x` | X-axis values |
| `y` | Y-axis values |
| `color` | Color of points |
| `marker` | Shape of points |
| `s` | Size of points |
| `alpha` | Transparency |
| `label` | Name shown in legend |

<br>

⭐ **VVVV Important**

For scatter plots, remember:

**x, y, color, marker, s, alpha, label**



<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. Size of Points → s parameter</h3>

<h5 style="color:#78B89A; font-weight:bold;">s → size of scatter points</h5>

The `s` parameter controls the size of each point.



````python
plt.scatter(
    df["avg"],
    df["strike_rate"],
    s=100
)

plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.title("Batting Average vs Strike Rate")

plt.show()
````

If we increase `s`, the points become larger.

Example:

```python
s=20
s=50
s=100
s=200
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">different size for different points</h5>

`s` can also receive an array/Series.

This is useful when we want point size to represent a **third numerical variable**.

For example:

- X → batting average
- Y → strike rate
- Point size → runs



````python
plt.scatter(
    df["avg"].head(20),
    df["strike_rate"].head(20),
    s=df["runs"].head(20) / 10
)

plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.title("Batting Average vs Strike Rate")

plt.show()
````

Here, batters with more runs get larger points.

This gives us information about **three variables in one graph**.

<h3 style="color:#6FA8DC; font-weight:bold">13. plt.scatter() vs plt.plot()</h3>

<h5 style="color:#78B89A; font-weight:bold;">main difference</h5>

Both can display points, but their purpose is different.

### `plt.scatter()`

Used mainly to show the **relationship between two numerical variables**.

Points are normally **not connected**.

```python
plt.scatter(x, y)
```

### `plt.plot()`

Used mainly for **trends and ordered data**.

Points can be connected by a line.

```python
plt.plot(x, y)
```

<br>

| Feature | `plt.scatter()` | `plt.plot()` |
|---|---|---|
| Main purpose | Relationship | Trend / progression |
| Points | Yes | Can show points with marker |
| Lines | Not normally connected | Connected by default |
| Point size `s` | Directly supported | Not used this way |
| Marker | Supported | Supported |
| Color | Supported | Supported |
| Linestyle | Not useful for normal scatter points | Very important |
| Best for correlation | ⭐⭐⭐⭐⭐ | ⭐⭐ |
| Best for time trend | ⭐⭐ | ⭐⭐⭐⭐⭐ |

<br>

### Visual comparison



````python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 10]

# plt.plot() → points connected
plt.plot(x, y, marker="o")
plt.title("plt.plot()")
plt.show()

# plt.scatter() → points not connected
plt.scatter(x, y)
plt.title("plt.scatter()")
plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">14. Complete Scatter Plot Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">putting everything together</h5>

Now let's create a more complete scatter plot using our batter dataset.



````python
plt.scatter(
    df["avg"],
    df["strike_rate"],
    color="red",
    marker="o",
    s=60,
    alpha=0.7,
    label="Batter"
)

plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.title("Batting Average vs Strike Rate")

plt.legend(
    loc="upper left",
    fontsize=10,
    title="Players"
)

plt.grid(True)

plt.show()
````
