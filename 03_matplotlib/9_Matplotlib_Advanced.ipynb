<h3 style="color:#6FA8DC; font-weight:bold">2. cmap → Colormap</h3>

<h5 style="color:#78B89A; font-weight:bold;">what is cmap?</h5>

`cmap` means **colormap**.

A colormap maps numerical values to different shades/colors.

It is especially useful when the **color itself represents another numerical value**.

For example:

```python
plt.scatter(
    x,
    y,
    c=y,
    cmap="viridis"
)
```

Here:

- `c=y` → color is based on the values of `y`
- `cmap="viridis"` → tells Matplotlib which colormap to use

⭐ **VVVV Important**

`cmap` is different from:

```python
color="blue"
```

`color` gives a fixed color.

`cmap` creates a color mapping based on numerical values.


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.scatter(
    x,
    y,
    c=y,
    cmap="viridis",
    s=100
)

plt.colorbar(label="Y value")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Using cmap")

plt.show()
````

### colorbar()

When color represents a numerical variable, `plt.colorbar()` helps us understand what the colors mean.

```python
plt.colorbar()
```

We can also give it a label:

```python
plt.colorbar(label="Y value")
```

Common colormaps include:

```text
viridis
plasma
inferno
magma
cividis
coolwarm
```

You do not need to memorize all of them. The important idea is:

**cmap → numerical value → color**

<h3 style="color:#6FA8DC; font-weight:bold">3. Plot Size</h3>

<h5 style="color:#78B89A; font-weight:bold;">figsize → control graph size</h5>

Sometimes the default graph is too small.

We can control the size using:

```python
plt.figure(figsize=(width, height))
```

Example:

```python
plt.figure(figsize=(10, 5))
```

The values are in **inches**.

```text
figsize=(10, 5)

10 → width
5  → height
```


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.figure(figsize=(10, 5))

plt.plot(x, y, marker="o")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Large Plot")

plt.show()
````

### Smaller graph

```python
plt.figure(figsize=(5, 3))
```

### Larger graph

```python
plt.figure(figsize=(12, 7))
```

⭐ **VVVV Important**

The order is:

```python
plt.figure(figsize=(10, 5))
plt.plot(x, y)
```

Set the figure size **before creating the plot**.

<h3 style="color:#6FA8DC; font-weight:bold">4. dpi → Resolution</h3>

<h5 style="color:#78B89A; font-weight:bold;">dots per inch</h5>

`dpi` controls the resolution of a figure.

For example:

```python
plt.figure(figsize=(10, 5), dpi=120)
```

Higher DPI generally means a sharper image, especially when saving figures.

For example:

```python
plt.savefig("graph.png", dpi=300)
```

is commonly used when a high-resolution image is required.

⭐ `figsize` → physical size

⭐ `dpi` → resolution


````python
plt.figure(
    figsize=(8, 5),
    dpi=120
)

plt.plot(x, y, marker="o")

plt.title("Figure Size and DPI")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">5. Annotations</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are annotations?</h5>

Annotations allow us to **write information next to a particular point** on a graph.

This is useful when we want to highlight:

- Important data points
- Maximum/minimum values
- Outliers
- Specific observations
- Important events

The main function is:

```python
plt.annotate()
```


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker="o")

plt.annotate(
    "Important Point",
    xy=(4, 30)
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Basic Annotation")

plt.show()
````

### xy

```python
xy=(4, 30)
```

specifies the point where the annotation is attached.

We can also move the text away from the point using `xytext`.

```python
plt.annotate(
    "Important Point",
    xy=(4, 30),
    xytext=(5, 35)
)
```


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker="o")

plt.annotate(
    "Maximum",
    xy=(4, 30),
    xytext=(4.5, 34),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Annotation with Arrow")

plt.show()
````

### Important `annotate()` parameters

| Parameter | Use |
|---|---|
| `text` | Text to display |
| `xy` | Point being highlighted |
| `xytext` | Position of text |
| `arrowprops` | Customize arrow |

⭐ **VVVV Important**

Basic pattern:

```python
plt.annotate(
    "text",
    xy=(x, y)
)
```


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.scatter(x, y)

for xi, yi in zip(x, y):
    plt.annotate(
        f"({xi}, {yi})",
        (xi, yi),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Annotating Multiple Points")

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">6. Horizontal Line</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.axhline()</h5>

A horizontal line can be used to show a:

- Threshold
- Average
- Target
- Limit
- Reference value

Syntax:

```python
plt.axhline(y=value)
```


````python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y, marker="o")

plt.axhline(y=20)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Horizontal Reference Line")

plt.show()
````

### Customize the line

```python
plt.axhline(
    y=20,
    linestyle="--",
    label="Target"
)
```

Then:

```python
plt.legend()
```

can be used to identify it.


````python
plt.plot(x, y, marker="o")

plt.axhline(
    y=20,
    linestyle="--",
    label="Target = 20"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Horizontal Line with Label")

plt.legend()

plt.show()
````

<br>

<h3 style="color:#6FA8DC; font-weight:bold">7. Vertical Line</h3>

<h5 style="color:#78B89A; font-weight:bold;">plt.axvline()</h5>

A vertical line is useful for showing a particular:

- X value
- Event
- Boundary
- Cutoff
- Important position

Syntax:

```python
plt.axvline(x=value)
```


````python
plt.plot(x, y, marker="o")

plt.axvline(
    x=3,
    linestyle="--",
    label="X = 3"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Vertical Reference Line")

plt.legend()

plt.show()
````

### Remember

```python
plt.axhline(y=20)
```

→ horizontal line

```python
plt.axvline(x=3)
```

→ vertical line


````python
plt.plot(x, y, marker="o")

plt.axhline(
    y=25,
    linestyle="--",
    label="Y = 25"
)

plt.axvline(
    x=3,
    linestyle=":",
    label="X = 3"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Horizontal and Vertical Lines")

plt.legend()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">9. Subplots</h3>

<h5 style="color:#78B89A; font-weight:bold;">what are subplots?</h5>

Subplots allow us to place **multiple graphs inside one figure**.

For example:

```text
┌─────────────┬─────────────┐
│   Plot 1    │   Plot 2    │
├─────────────┼─────────────┤
│   Plot 3    │   Plot 4    │
└─────────────┴─────────────┘
```

This is very useful when comparing multiple visualizations together.

The main function is:

```python
plt.subplots()
```


<br>

<h3 style="color:#6FA8DC; font-weight:bold">10. Creating 2 Subplots</h3>

<h5 style="color:#78B89A; font-weight:bold;">1 row × 2 columns</h5>

Syntax:

```python
fig, axes = plt.subplots(1, 2)
```

Here:

```text
1 → number of rows
2 → number of columns
```


````python
fig, axes = plt.subplots(1, 2)

axes[0].plot(
    [1, 2, 3, 4],
    [10, 20, 15, 30]
)

axes[0].set_title("Line Plot")

axes[1].scatter(
    [1, 2, 3, 4],
    [10, 20, 15, 30]
)

axes[1].set_title("Scatter Plot")

plt.tight_layout()

plt.show()
````

Notice that with subplots we use:

```python
axes[0].plot()
```

instead of:

```python
plt.plot()
```

And for labels:

```python
axes[0].set_xlabel()
axes[0].set_ylabel()
axes[0].set_title()
```


<br>

<h3 style="color:#6FA8DC; font-weight:bold">11. 2 × 2 Subplots</h3>

<h5 style="color:#78B89A; font-weight:bold;">4 graphs in one figure</h5>

We can create:

```python
fig, axes = plt.subplots(2, 2)
```

This gives:

```text
axes[0,0]    axes[0,1]

axes[1,0]    axes[1,1]
```


````python
fig, axes = plt.subplots(
    2,
    2,
    figsize=(9, 7)
)

axes[0, 0].plot(
    [1, 2, 3, 4],
    [10, 20, 15, 30]
)
axes[0, 0].set_title("Line Plot")

axes[0, 1].scatter(
    [1, 2, 3, 4],
    [10, 20, 15, 30]
)
axes[0, 1].set_title("Scatter Plot")

axes[1, 0].bar(
    ["A", "B", "C", "D"],
    [20, 35, 25, 45]
)
axes[1, 0].set_title("Bar Chart")

axes[1, 1].hist(
    [1, 5, 7, 8, 10, 12, 14, 15, 18, 20,
     21, 24, 25, 27, 30, 32, 35, 37, 40, 42]
)
axes[1, 1].set_title("Histogram")

plt.tight_layout()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">12. Understanding `fig` and `axes`</h3>

When we write:

```python
fig, axes = plt.subplots(2, 2)
```

we get:

### `fig`

The **entire figure/container**.

### `axes`

The individual plotting areas.

For a 2 × 2 layout:

```python
axes[0, 0]
axes[0, 1]
axes[1, 0]
axes[1, 1]
```

Each one is an individual **Axes** object.

So:

```python
axes[0, 0].plot()
```

plots only inside the first subplot.

<h3 style="color:#6FA8DC; font-weight:bold">13. Subplot Size and Spacing</h3>

We can control the overall figure size:

```python
fig, axes = plt.subplots(
    2,
    2,
    figsize=(10, 7)
)
```

If labels overlap, use:

```python
plt.tight_layout()
```

⭐ **VVVV Important**

```python
plt.tight_layout()
```

automatically adjusts spacing between subplots so that titles, labels and graphs fit better.

<h3 style="color:#6FA8DC; font-weight:bold">14. `plt.subplot()` vs `plt.subplots()`</h3>

These two are different.

### `plt.subplot()`

Used to select/create a subplot one at a time.

```python
plt.subplot(2, 2, 1)
```

means:

```text
2 rows
2 columns
1st position
```

### `plt.subplots()`

Creates the figure and axes together.

```python
fig, axes = plt.subplots(2, 2)
```

This is generally easier when working with multiple plots because we can directly access each Axes object.

⭐ **For our notes, focus mainly on `plt.subplots()`.**


````python
plt.subplot(2, 2, 1)
plt.plot([1, 2, 3], [10, 20, 15])
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.scatter([1, 2, 3], [10, 20, 15])
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.bar(["A", "B", "C"], [10, 20, 15])
plt.title("Plot 3")

plt.subplot(2, 2, 4)
plt.hist([1, 2, 2, 3, 3, 3, 4, 5])
plt.title("Plot 4")

plt.tight_layout()

plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">15. Complete Advanced Example</h3>

<h5 style="color:#78B89A; font-weight:bold;">cmap + annotations + reference lines + subplots</h5>

These concepts can also be combined in real EDA work.


````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("datasets/batter.csv")

fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5)
)

# Scatter plot with cmap
scatter = axes[0].scatter(
    df["avg"],
    df["strike_rate"],
    c=df["runs"],
    cmap="viridis",
    alpha=0.8
)

axes[0].set_xlabel("Average")
axes[0].set_ylabel("Strike Rate")
axes[0].set_title("Average vs Strike Rate")

axes[0].axvline(
    df["avg"].mean(),
    linestyle="--",
    label="Mean Average"
)

axes[0].legend()
fig.colorbar(scatter, ax=axes[0], label="Runs")

# Histogram
axes[1].hist(
    df["runs"],
    bins=20,
    edgecolor="black"
)

axes[1].set_xlabel("Runs")
axes[1].set_ylabel("Frequency")
axes[1].set_title("Distribution of Runs")

axes[1].axvline(
    df["runs"].mean(),
    linestyle="--",
    label="Mean Runs"
)

axes[1].legend()

plt.tight_layout()

plt.show()
````
