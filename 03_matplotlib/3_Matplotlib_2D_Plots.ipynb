<h3 style="color:#6FA8DC; font-weight:bold">3. Matplotlib → 2D Plot</h3>

<h5 style="color:#78B89A; font-weight:bold;">What is a 2D Plot?</h5>

A **2D plot** represents data using two axes:

```text
X-axis → horizontal
Y-axis → vertical
```

A point on the graph is represented using:

```text
(x, y)
```

For example:

```text
(1, 2)
(2, 5)
(3, 4)
```

A 2D plot is one of the most basic and important ways of visualizing data.

<br>

<h5 style="color:#78B89A; font-weight:bold;">When do we use a 2D plot?</h5>

A 2D plot is useful when we want to understand how one variable changes with respect to another.

Common examples:

- time vs sales
- day vs temperature
- study hours vs marks
- match number vs runs
- age vs salary

For **ordered data or trends**, a line plot is often useful.

For two numerical variables where the main goal is to see their relationship, a scatter plot may be more appropriate.

The important idea is:

```text
Two variables
     ↓
Need to understand their relationship / change
     ↓
2D visualization
```

<br>


<h5 style="color:#78B89A; font-weight:bold;">Basic structure of a Matplotlib graph</h5>

A typical graph contains:

```text
              Title
                ↓
        ┌─────────────────┐
        │                 │
 Y-axis │      Graph      │
        │                 │
        │                 │
        └─────────────────┘
                ↑
             X-axis
```

We can customize:

- X-axis label
- Y-axis label
- title
- color
- line style
- markers
- legend
- grid
- axis limits

<br>


<h5 style="color:#78B89A; font-weight:bold;">1. Import Matplotlib</h5>

````python
import matplotlib.pyplot as plt
````

<h5 style="color:#78B89A; font-weight:bold;">2. Creating a Basic 2D Plot</h5>

The most basic function for creating a line plot is:

```python
plt.plot(x, y)
```

`x` contains the X-axis values and `y` contains the corresponding Y-axis values.


````python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 10]

plt.plot(x, y)
plt.show()
````

![Basic 2D Plot](matplotlib_graph_images/basic_2d_plot.png)

Here the points are connected with lines.

<br>


<h5 style="color:#78B89A; font-weight:bold;">3. Multiple Plots</h5>

We can draw multiple lines on the same graph by calling `plt.plot()` multiple times.

This is useful when we want to compare two or more datasets.


````python
x = [1, 2, 3, 4, 5]

sales_a = [10, 15, 12, 20, 25]
sales_b = [8, 12, 16, 18, 22]

plt.plot(x, sales_a)
plt.plot(x, sales_b)

plt.show()
````

Both lines are shown on the same graph, making comparison easier.

<br>


<h5 style="color:#78B89A; font-weight:bold;">4. X-axis and Y-axis Labels</h5>

We use:

```python
plt.xlabel()
plt.ylabel()
```

These tell the reader what each axis represents.


````python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 10]

plt.plot(x, y)

plt.xlabel("Days")
plt.ylabel("Sales")

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">5. Adding a Title</h5>

Use:

```python
plt.title()
```

A title should clearly describe what the graph represents.


````python
plt.plot(x, y)

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales over 5 Days")

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">6. Providing Color</h5>

We can change the line color using the `color` parameter.

```python
plt.plot(x, y, color="red")
```

Some commonly used color names include:

```text
red
blue
green
black
orange
purple
```


````python
plt.plot(x, y, color="red")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales over 5 Days")

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">7. Linestyle</h5>

The `linestyle` parameter controls how the line looks.

Common styles:

```text
"-"   → solid
"--"  → dashed
":"   → dotted
"-."  → dash-dot
```


````python
plt.plot(x, y, linestyle="--")

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">8. Marker</h5>

A marker shows the individual data points.

Use the `marker` parameter.

Common markers:

```text
"o" → circle
"s" → square
"^" → triangle
"*" → star
"x" → x
```


````python
plt.plot(x, y, marker="o")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales with Data Point Markers")

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">9. Combining Color + Linestyle + Marker</h5>

These parameters can be used together.


````python
plt.plot(
    x,
    y,
    color="red",
    linestyle="--",
    marker="o"
)

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Customized 2D Plot")

plt.show()
````

![Customized Plot](matplotlib_graph_images/customized_plot.png)

<br>


<h5 style="color:#78B89A; font-weight:bold;">10. Legend</h5>

When multiple plots are present, we need to tell the viewer what each line represents.

We give each plot a `label`:

```python
plt.plot(x, y, label="Sales")
```

Then display the legend using:

```python
plt.legend()
```


````python
x = [1, 2, 3, 4, 5]

sales_a = [10, 15, 12, 20, 25]
sales_b = [8, 12, 16, 18, 22]

plt.plot(x, sales_a, label="Product A")
plt.plot(x, sales_b, label="Product B")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Product Sales Comparison")

plt.legend()
plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">11. Legend Parameters</h5>

`plt.legend()` has several useful parameters.

### `loc`

Controls where the legend is placed.

Examples:

```python
plt.legend(loc="upper left")
plt.legend(loc="upper right")
plt.legend(loc="lower left")
plt.legend(loc="lower right")
plt.legend(loc="best")
```

`"best"` lets Matplotlib choose a suitable location.

### `fontsize`

Controls the text size.

```python
plt.legend(fontsize=10)
```

### `title`

Adds a title to the legend.

```python
plt.legend(title="Products")
```

These parameters can also be combined.


````python
plt.plot(x, sales_a, label="Product A")
plt.plot(x, sales_b, label="Product B")

plt.legend(
    loc="upper left",
    fontsize=10,
    title="Products"
)

plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">12. Grid</h5>

A grid makes it easier to read values from a graph.

Use:

```python
plt.grid()
```


````python
plt.plot(x, y, marker="o")

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales with Grid")

plt.grid()
plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">13. Outlier Problem</h5>

An **outlier** is a value that is unusually far away from the other observations.

For example:

```text
20, 25, 30, 27, 1000
```

Here `1000` is much larger than the other values.

When we plot this data, the large value can stretch the Y-axis and make the normal values difficult to see.



````python
x = [1, 2, 3, 4, 5]
y = [20, 25, 30, 27, 1000]

plt.plot(x, y, marker="o")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Outlier affecting the Y-axis")

plt.show()
````

![Outlier Problem](matplotlib_graph_images/outlier_problem.png)

The smaller values are compressed near the bottom because of the large outlier.

<br>


<h5 style="color:#78B89A; font-weight:bold;">14. Controlling the Y-axis using ylim()</h5>

If we want to focus on a particular Y-axis range, we can use:

```python
plt.ylim(lower, upper)
```

For example:

```python
plt.ylim(0, 40)
```

This changes the visible Y-axis range.

⚠️ `ylim()` does **not remove the outlier from the data**. It only controls the visible plotting range.

This is useful when we want to inspect the normal range more clearly.



````python
x = [1, 2, 3, 4, 5]
y = [20, 25, 30, 27, 1000]

plt.plot(x, y, marker="o", label="Data")

plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Using ylim() to control the visible Y-axis")

plt.ylim(0, 40)
plt.legend()
plt.grid()

plt.show()
````

![ylim Solution](matplotlib_graph_images/ylim_solution.png)

⭐ **Important:** Before using `ylim()` to hide a value, understand whether the value is a genuine observation or an actual data error. In data analysis, we should not simply hide an outlier without investigating it.

<br>


<h5 style="color:#78B89A; font-weight:bold;">15. plt.show()</h5>

`plt.show()` displays the graph.

A common structure is:

```python
plt.plot(x, y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("My Graph")
plt.grid()
plt.legend()

plt.show()
```

`plt.show()` is normally written at the end of the plotting code.

<br>


<h5 style="color:#78B89A; font-weight:bold;">16. Complete 2D Plot Example</h5>

Putting the important concepts together:



````python
x = [1, 2, 3, 4, 5]
sales_a = [10, 15, 12, 20, 25]
sales_b = [8, 12, 16, 18, 22]

plt.plot(
    x,
    sales_a,
    color="blue",
    linestyle="-",
    marker="o",
    label="Product A"
)

plt.plot(
    x,
    sales_b,
    color="red",
    linestyle="--",
    marker="s",
    label="Product B"
)

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Product Sales Comparison")

plt.legend(loc="best")
plt.grid()

plt.show()
````
