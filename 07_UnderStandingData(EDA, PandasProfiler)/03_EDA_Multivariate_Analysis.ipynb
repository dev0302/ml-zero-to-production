<h5 style="color:#78B89A; font-weight:bold;">Understanding relationships between multiple columns</h5>

In **Univariate Analysis**, we study one column at a time.

In **Multivariate Analysis**, we study the relationship between two or more columns.

The main purpose is to answer questions like:

- Does one numerical column increase when another increases?
- How does a numerical column change across categories?
- Are two categorical columns related?
- Can we identify patterns, groups, or trends?
- Which columns may be useful for Machine Learning?

## 1. Scatterplot (Numerical - Numerical)

A **scatterplot** is used when we want to study the relationship between two numerical columns.

Example:

- `total_bill` → numerical
- `tip` → numerical

Each dot represents one observation.

### What can we understand?

- Positive relationship
- Negative relationship
- No clear relationship
- Outliers
- Clusters or groups

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip'
)

plt.title('Total Bill vs Tip')
plt.show()
```

### Adding a categorical column using `hue`

`hue` adds different colors according to a categorical column.

Here, we are checking whether the relationship changes for smokers and non-smokers.

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex'
)

plt.title('Total Bill vs Tip according to Sex')
plt.show()
```

### Adding more information using `style` and `size`

- `hue` → changes color
- `style` → changes marker shape
- `size` → changes marker size

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',
    style='smoker',
    size='size'
)

plt.title('Multivariate Scatterplot')
plt.show()
```

## 2. Bar Plot (Numerical - Categorical)

A **bar plot** is used to compare a numerical value across different categories.

By default, Seaborn calculates the **mean** of the numerical column for each category.

Example:

- `Pclass` → categorical
- `Age` → numerical

We can compare the average age of passengers from different classes.

```python
sns.barplot(
    data=titanic,
    x='class',
    y='age'
)

plt.title('Average Age according to Passenger Class')
plt.show()
```

### Adding `hue`

Using `hue='sex'`, we can compare males and females inside every passenger class.

```python
sns.barplot(
    data=titanic,
    x='class',
    y='age',
    hue='sex'
)

plt.title('Average Age according to Class and Sex')
plt.show()
```

### Important point

A bar plot generally shows an **aggregated value**, usually the mean.

It is different from a countplot:

- `countplot()` → counts observations
- `barplot()` → compares an aggregate numerical value

## 3. Box Plot (Numerical - Categorical)

A **box plot** compares the distribution of a numerical column across categories.

It displays:

- Median
- First quartile (Q1)
- Third quartile (Q3)
- Interquartile Range (IQR)
- Whiskers
- Possible outliers

Example: Compare passenger age according to gender.

```python
sns.boxplot(
    data=titanic,
    x='sex',
    y='age'
)

plt.title('Age Distribution according to Sex')
plt.show()
```

### Adding `hue`

We can further divide each category according to survival status.

```python
sns.boxplot(
    data=titanic,
    x='sex',
    y='age',
    hue='survived'
)

plt.title('Age according to Sex and Survival')
plt.show()
```

### Why is a box plot useful?

It helps us compare:

- Which category has a higher median?
- Which category has more spread?
- Which category contains possible outliers?
- Whether two distributions are different?

<div style="border-top:black 2px solid"></div>

## 4. Distribution Plot (Numerical - Categorical)

We can compare the distribution of a numerical column for different categories.

Older tutorials commonly use `distplot()`, but it is deprecated.

Use:

- `kdeplot()` for smooth density curves
- `histplot()` for histograms
- `displot()` for figure-level distribution plots

Here, we compare the age distribution of survivors and non-survivors.

```python
sns.kdeplot(
    data=titanic[titanic['survived'] == 0],
    x='age',
    label='Did not survive',
    fill=True
)

sns.kdeplot(
    data=titanic[titanic['survived'] == 1],
    x='age',
    label='Survived',
    fill=True
)

plt.title('Age Distribution according to Survival')
plt.legend()
plt.show()
```

### Interpretation

A density plot helps us understand where values are concentrated.

For example, we can check whether the age distribution of survivors differs from the age distribution of passengers who did not survive.

## 5. Heatmap (Categorical - Categorical)

A **heatmap** represents values using different colors.

To study two categorical columns, we generally create a frequency table using `pd.crosstab()`.

Example:

- `class`
- `survived`

The cells show how many passengers belong to each combination.

```python
pd.crosstab(titanic['class'], titanic['survived'])
```

```python
sns.heatmap(
    pd.crosstab(titanic['class'], titanic['survived']),
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title('Passenger Class vs Survival')
plt.xlabel('Survived')
plt.ylabel('Passenger Class')
plt.show()
```

### Understanding `crosstab`

`pd.crosstab()` creates a frequency table.

- Rows represent one categorical column.
- Columns represent another categorical column.
- Values represent the number of observations.

`annot=True` writes the values inside the heatmap.

`fmt='d'` displays the values as integers.

### Percentage-based analysis

Counts can sometimes be misleading because categories may have different sizes.

We can calculate the survival percentage for each embarkation location.

```python
titanic.groupby('embark_town')['survived'].mean() * 100
```

## 6. ClusterMap (Categorical - Categorical)

A **clustermap** is similar to a heatmap, but it also groups similar rows and columns together using clustering.

It is useful when we want to identify patterns in a matrix.

Example: Parent/child count versus survival.

```python
sns.clustermap(
    pd.crosstab(titanic['parch'], titanic['survived']),
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.show()
```

### Heatmap vs ClusterMap

| Heatmap | ClusterMap |
|---|---|
| Shows values using colors | Shows values using colors |
| Does not automatically cluster rows/columns | Groups similar rows/columns |
| Simpler comparison | Useful for pattern discovery |

Clustermap is especially useful when the matrix is large.

## 7. Pairplot

A **pairplot** creates plots for every possible pair of numerical columns in a dataset.

It is useful for quickly understanding:

- Relationships between numerical columns
- Distribution of individual columns
- Clusters
- Differences between categories

We will use the Iris dataset.

```python
sns.pairplot(
    iris,
    hue='species'
)

plt.show()
```

### Important parameters

- `hue` → color according to a category
- `kind='scatter'` → scatterplots
- `kind='reg'` → regression plots
- `diag_kind='hist'` → histograms on diagonal
- `diag_kind='kde'` → KDE plots on diagonal
- `corner=True` → show only the lower triangle

```python
sns.pairplot(
    iris,
    hue='species',
    diag_kind='kde',
    corner=True
)

plt.show()
```

<div style="border-top:black 2px solid"></div>

## 8. Lineplot (Numerical - Numerical)

A **line plot** is mainly used to understand trends over an ordered variable, especially time.

Example:

- `year` → time
- `passengers` → numerical value

First, we aggregate the monthly data year-wise.

```python
new = flights.groupby('year')['passengers'].sum().reset_index()

new.head()
```

```python
sns.lineplot(
    data=new,
    x='year',
    y='passengers',
    marker='o'
)

plt.title('Total Passengers per Year')
plt.show()
```

### Why use a line plot?

It helps us identify:

- Increasing trends
- Decreasing trends
- Seasonal patterns
- Sudden changes
- Long-term movement

### Monthly passenger heatmap

We can convert the flights dataset into a matrix using `pivot_table()`.

- Rows → months
- Columns → years
- Values → passengers

```python
flight_pivot = flights.pivot_table(
    values='passengers',
    index='month',
    columns='year'
)

flight_pivot
```

```python
sns.heatmap(
    flight_pivot,
    annot=True,
    fmt='d',
    cmap='YlGnBu'
)

plt.title('Monthly Passengers across Years')
plt.show()
```

## Quick Revision

| Plot | Data Relationship | Main Purpose |
|---|---|---|
| Scatterplot | Numerical - Numerical | Relationship between two numerical columns |
| Barplot | Numerical - Categorical | Compare aggregate numerical values |
| Boxplot | Numerical - Categorical | Compare distributions and outliers |
| KDE/Distribution plot | Numerical - Categorical | Compare distributions |
| Heatmap | Categorical - Categorical / Matrix | Show values using colors |
| Clustermap | Matrix | Find similar groups using clustering |
| Pairplot | Multiple numerical columns | Explore all pairwise relationships |
| Lineplot | Numerical - Numerical / Time | Understand trends over an ordered variable |

### Final memory trick

**Scatter → Relationship**

**Bar → Average comparison**

**Box → Spread and outliers**

**KDE → Distribution comparison**

**Heatmap → Color-based matrix**

**ClusterMap → Heatmap + clustering**

**Pairplot → All numerical relationships**

**Lineplot → Trend over time**
