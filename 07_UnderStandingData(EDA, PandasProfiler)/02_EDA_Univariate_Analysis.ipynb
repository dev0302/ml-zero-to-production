<h3 style="color:#6FA8DC; font-weight:bold">01 — EDA (Exploratory Data Analysis): Univariate Analysis</h3>

**Univariate Analysis** means analysing **only one column/variable at a time**.

In this notebook, we will understand:
- What is Univariate Analysis?
- Categorical Data Analysis
- Numerical Data Analysis
- Important plots and Pandas/Seaborn functions
- How to interpret graphs
- Practical examples using Titanic dataset

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
````

````python
df = sns.load_dataset("titanic")
df.head()
````

<h5 style="color:#78B89A; font-weight:bold;">What is EDA?</h5>

**EDA = Exploratory Data Analysis**

EDA is the process of understanding a dataset before applying Machine Learning.

We try to find:
- Patterns
- Missing values
- Outliers
- Distribution
- Most common categories
- Data spread

<h5 style="color:#78B89A; font-weight:bold;">What is Univariate Analysis?</h5>

Univariate means **one variable at a time**.

Examples:
- Distribution of Age
- Number of male and female passengers
- Distribution of Fare
- Count of passengers from each class

We do not study relationships between two columns here.

<h3 style="color:#6FA8DC; font-weight:bold">1. Types of Data in Univariate Analysis</h3>

There are mainly two types:

1. **Categorical Data**
   - Sex
   - Embarked
   - Pclass (when treated as categories)
   - Survived

2. **Numerical Data**
   - Age
   - Fare
   - SibSp
   - Parch

<h3 style="color:#6FA8DC; font-weight:bold">2. Categorical Data</h3>

Categorical data contains values in the form of categories or groups.

For categorical data, we mainly ask:

- Which category occurs most?
- How many observations belong to each category?
- What percentage belongs to each category?

Important plots:
- Countplot
- Pie Chart
- Bar Plot

<h5 style="color:#78B89A; font-weight:bold;">a. Countplot → count of each category</h5>

A **countplot** shows how many times each category appears.

It is useful for checking category frequency.

````python
sns.countplot(x=df["Sex"])
plt.title("Number of Male and Female Passengers")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
````

### Using `value_counts()`

Before plotting, we can also calculate the frequency of each category.

````python
df["Sex"].value_counts()
````

### Countplot with another categorical column

````python
sns.countplot(x=df["Embarked"])
plt.title("Passengers by Embarkation Port")
plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">b. Pie Chart → percentage contribution</h5>

A pie chart shows how the total data is divided among categories.

`autopct` displays percentage values.

````python
df["Sex"].value_counts().plot(
    kind="pie",
    autopct="%.2f%%",
    startangle=90
)

plt.title("Male vs Female Percentage")
plt.ylabel("")
plt.show()
````

### Important note

- **Countplot** is generally easier for comparing category sizes.
- **Pie chart** is useful when showing percentage contribution.
- Too many categories make a pie chart difficult to understand.

<h3 style="color:#6FA8DC; font-weight:bold">3. Numerical Data</h3>

Numerical data contains numbers.

Examples:
- Age
- Fare
- SibSp
- Parch

For numerical data, we ask:

- What is the distribution?
- Where are most values concentrated?
- Is the data symmetric or skewed?
- Are there outliers?
- What is the spread of the data?

Important plots:
- Histogram
- KDE Plot
- Boxplot
- Distplot (old function, avoid using it in modern Seaborn)

<h5 style="color:#78B89A; font-weight:bold;">a. Histogram → distribution of numerical data</h5>

A histogram divides numerical values into intervals called **bins**.

It shows how many values fall inside each interval.

````python
plt.hist(df["Age"].dropna(), bins=10)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
````

### Histogram using Seaborn

````python
sns.histplot(data=df, x="Age", bins=10)

plt.title("Age Distribution")
plt.show()
````

### Important histogram parameters

- `bins` → number of intervals
- `binwidth` → width of each interval
- `kde=True` → adds a smooth density curve
- `stat="density"` → shows density instead of count
- `hue` → compare distributions of categories

````python
sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Age Distribution with KDE")
plt.show()
````

<h5 style="color:#78B89A; font-weight:bold;">b. KDE Plot → smooth distribution curve</h5>

**KDE = Kernel Density Estimate**

KDE shows a smooth curve representing the distribution of numerical values.

It helps us understand:
- Where values are concentrated
- Peaks in the distribution
- Shape of the data
- Possible skewness

````python
sns.kdeplot(data=df, x="Age", fill=True)

plt.title("KDE Plot of Age")
plt.show()
````

### Histogram vs KDE

| Histogram | KDE |
|---|---|
| Uses bins | Uses a smooth curve |
| Shows frequency/count | Shows estimated density |
| Shape depends on bin size | Shape depends on bandwidth |
| Easy for beginners | Useful for understanding distribution shape |

<h5 style="color:#78B89A; font-weight:bold;">c. Boxplot → spread and outliers</h5>

A boxplot summarizes numerical data using:

- Minimum
- Q1 (25th percentile)
- Median (50th percentile)
- Q3 (75th percentile)
- Maximum
- Possible outliers

**IQR = Q3 - Q1**

Possible outliers are usually values below:

`Q1 - 1.5 × IQR`

or above:

`Q3 + 1.5 × IQR`

````python
sns.boxplot(y=df["Age"])

plt.title("Boxplot of Age")
plt.show()
````

### Boxplot of Fare

````python
sns.boxplot(y=df["Fare"])

plt.title("Boxplot of Fare")
plt.show()
````

### Why is boxplot important?

Boxplot helps us quickly identify:

- Median
- Data spread
- Skewness
- Possible outliers

It is very useful during EDA and before ML preprocessing.

<h3 style="color:#6FA8DC; font-weight:bold">4. Numerical Summary using Pandas</h3>

````python
df["Age"].describe()
````

### Important statistical functions

````python
age = df["Age"]

print("Mean:", age.mean())
print("Median:", age.median())
print("Mode:", age.mode()[0])
print("Minimum:", age.min())
print("Maximum:", age.max())
print("Standard Deviation:", age.std())
print("Variance:", age.var())
````

### Interpretation

- **Mean** → average value
- **Median** → middle value
- **Mode** → most frequently occurring value
- **Minimum** → smallest value
- **Maximum** → largest value
- **Standard deviation** → how spread out values are
- **Variance** → squared measure of spread

<h3 style="color:#6FA8DC; font-weight:bold">5. Checking Skewness</h3>

Skewness tells us whether the distribution is balanced or pulled toward one side.

Types:

1. **Symmetric distribution**
2. **Right-skewed / positively skewed**
3. **Left-skewed / negatively skewed**

General interpretation:

- Mean ≈ Median → approximately symmetric
- Mean > Median → often right-skewed
- Mean < Median → often left-skewed

````python
print("Mean:", df["Fare"].mean())
print("Median:", df["Fare"].median())

sns.histplot(data=df, x="Fare", kde=True)
plt.title("Fare Distribution")
plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">6. Finding Outliers using IQR</h3>

````python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
````

````python
outliers = df[
    (df["Fare"] < lower_limit) |
    (df["Fare"] > upper_limit)
]

outliers[["Fare"]].head()
````

### Important

An outlier is not always an error.

It may represent:
- A genuine unusual observation
- A rare event
- A data entry mistake

We should investigate before deleting it.

<h3 style="color:#6FA8DC; font-weight:bold">7. Univariate Analysis of Multiple Columns</h3>

We can analyse all numerical columns using histograms.

````python
df.hist(
    figsize=(12, 8),
    bins=15
)

plt.suptitle("Univariate Analysis of Numerical Columns")
plt.show()
````

### Boxplots for all numerical columns

````python
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
````

<h3 style="color:#6FA8DC; font-weight:bold">8. Practical Univariate Analysis Questions</h3>

For every column, ask:

### For categorical columns

- What are the unique categories?
- Which category is most common?
- What is the count of every category?
- Are some categories very rare?
- Is the data balanced or imbalanced?

### For numerical columns

- What is the minimum and maximum?
- What is the mean and median?
- How is the distribution shaped?
- Are there missing values?
- Are there outliers?
- Is the data skewed?
- What is the standard deviation?
