<h3 style="color:#6FA8DC; font-weight:bold">01 — Understanding Your Data: Asking Basic Questions</h3>

Before performing EDA or building a Machine Learning model, we must first understand the dataset.

In this notebook, we will learn how to ask basic questions about a dataset using Pandas and Seaborn's Titanic dataset.

| dtype | Meaning |
|---|---|

<h5 style="color:#78B89A; font-weight:bold;">10. Detailed overview → `df.info()` ⭐⭐⭐⭐⭐</h5>

`info()` shows the number of rows, column names, non-null counts, data types and memory usage.

````python
df.info()
````

If a column has fewer non-null values than the total number of rows, it contains missing values.

<h5 style="color:#78B89A; font-weight:bold;">11. Are there missing values? → `isnull()`</h5>

````python
df.isnull().head()
````

````python
df.isnull().sum()
````

`True` means missing and `False` means present.

To calculate missing-value percentages:

````python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage.sort_values(ascending=False)
````

Formula:

\[
	ext{Missing Percentage} =
rac{	ext{Missing Values}}{	ext{Total Rows}} 	imes 100
\]

<h5 style="color:#78B89A; font-weight:bold;">12. Check non-missing values → `notnull()`</h5>

````python
df.notnull().sum()
````

`notnull()` is the opposite of `isnull()`.

<h5 style="color:#78B89A; font-weight:bold;">15. Are there duplicate rows? → `duplicated()`</h5>

````python
df.duplicated().head()
````

````python
df.duplicated().sum()
````

````python
duplicate_rows = df[df.duplicated()]

duplicate_rows.head()
````

````python
df_without_duplicates = df.drop_duplicates()

df_without_duplicates.shape
````

Do not remove duplicates blindly. Some repeated records may be genuine.

<h5 style="color:#78B89A; font-weight:bold;">17. How many unique values exist? → `nunique()`</h5>

````python
df.nunique()
````

````python
df["class"].nunique()
````

````python
df["class"].unique()
````

Difference:

- `unique()` shows distinct values.
- `nunique()` counts distinct values.

<h5 style="color:#78B89A; font-weight:bold;">18. How frequently does each category occur? → `value_counts()`</h5>

````python
df["class"].value_counts()
````

````python
df["sex"].value_counts()
````

````python
df["class"].value_counts(normalize=True) * 100
````

<h5 style="color:#78B89A; font-weight:bold;">19. Check unique values for every column</h5>

````python
for column in df.columns:
    print(column, "→", df[column].nunique(), "unique values")
````

This helps identify binary columns, categorical columns, identifier columns and high-cardinality columns.

<h5 style="color:#78B89A; font-weight:bold;">21. Select categorical columns</h5>

<h5 style="color:#78B89A; font-weight:bold;">22. Check the distribution of a numerical column</h5>

A histogram shows how numerical values are distributed.

````python
df["age"].plot(
    kind="hist",
    bins=20,
    title="Distribution of Passenger Age"
)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
````

A histogram helps identify:

- Common value ranges
- Skewness
- Spread
- Possible unusual values

<h5 style="color:#78B89A; font-weight:bold;">23. Check minimum, maximum and average</h5>

````python
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())
print("Average age:", df["age"].mean())
print("Median age:", df["age"].median())
print("Mode:", df["age"].mode().tolist())
````

Mean is the average, median is the middle value and mode is the most frequent value.

<h5 style="color:#78B89A; font-weight:bold;">24. Check correlation between columns ⭐⭐⭐⭐⭐</h5>

Correlation measures the strength and direction of a linear relationship between numerical variables.

- `+1` → Perfect positive relationship
- `0` → No linear relationship
- `-1` → Perfect negative relationship

Correlation does not prove causation.

````python
correlation_matrix = df.select_dtypes(include="number").corr()

correlation_matrix
````

<h5 style="color:#78B89A; font-weight:bold;">25. Visualize correlation using a heatmap</h5>

````python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
````

A heatmap makes positive and negative relationships easier to inspect.

<h5 style="color:#78B89A; font-weight:bold;">26. Visualize the relationship between two numerical columns</h5>

````python
sns.scatterplot(
    data=df,
    x="age",
    y="fare"
)

plt.title("Age vs Fare")
plt.show()
````

A scatter plot helps us visually inspect whether two numerical columns are related.

<h5 style="color:#78B89A; font-weight:bold;">27. Compare a numerical column across categories</h5>

````python
df.groupby("class", observed=True)["fare"].mean()
````

````python
sns.barplot(
    data=df,
    x="class",
    y="fare",
    estimator="mean",
    errorbar=None
)

plt.title("Average Fare by Passenger Class")
plt.show()
````

This answers:

> Does average fare differ between passenger classes?

<h5 style="color:#78B89A; font-weight:bold;">30. Create a basic data summary report</h5>

````python
summary = pd.DataFrame({
    "dtype": df.dtypes,
    "missing_values": df.isnull().sum(),
    "missing_percentage": (df.isnull().sum() / len(df)) * 100,
    "unique_values": df.nunique()
})

summary
````

<h5 style="color:#78B89A; font-weight:bold;">31. Questions every data analyst should ask</h5>

1. What does one row represent?
2. What does each column mean?
3. How many rows and columns are present?
4. Which columns are numerical?
5. Which columns are categorical?
6. Are there missing values?
7. Are there duplicate rows?
8. Are there impossible or suspicious values?
9. Are there outliers?
10. Which columns have very few unique values?
11. Which columns are identifiers?
12. Is the target variable balanced?
13. Are some columns strongly correlated?
14. Is the dataset large enough for the intended task?
15. Which columns may need cleaning or transformation?

<h5 style="color:#78B89A; font-weight:bold;">32. Create a reusable inspection function</h5>

````python
def basic_data_report(dataframe):
    print("Shape:", dataframe.shape)

    print("\nColumn names:")
    print(dataframe.columns.tolist())

    print("\nData types:")
    print(dataframe.dtypes)

    print("\nMissing values:")
    print(dataframe.isnull().sum())

    print("\nDuplicate rows:")
    print(dataframe.duplicated().sum())

    print("\nUnique values:")
    print(dataframe.nunique())

    print("\nNumerical summary:")
    display(dataframe.describe())

basic_data_report(df)
````

<h3 style="color:#6FA8DC; font-weight:bold">Final Revision</h3>

| Question | Function |
|---|---|
| How big is the data? | `df.shape` |
| How does it look? | `df.head()` |
| Random rows | `df.sample()` |
| Last rows | `df.tail()` |
| Column names | `df.columns` |
| Data types | `df.dtypes` |
| Complete overview | `df.info()` |

| Missing percentage | `df.isnull().mean() * 100` |

| Duplicate rows | `df.duplicated().sum()` |
| Unique count | `df.nunique()` |
| Unique values | `df["column"].unique()` |
| Category frequency | `df["column"].value_counts()` |
| Correlation | `df.corr(numeric_only=True)` |
| Memory usage | `df.memory_usage(deep=True)` |

### Memory trick

```text
SIZE → LOOK → TYPES → MISSING → SUMMARY
→ DUPLICATES → UNIQUE → FREQUENCY → CORRELATION
→ BUSINESS QUESTIONS
```
