<h3 style="color:#6FA8DC; font-weight:bold">Broadcasting</h3>

Broadcasting is a NumPy feature which allows us to perform operations on arrays of different shapes.

Instead of manually changing the shape of arrays, NumPy automatically adjusts them so that the operation can be performed.

<h5 style="color:#78B89A; font-weight:bold;">initially creating arrays → used for examples</h5>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<h5 style="color:#78B89A; font-weight:bold;">broadcasting example → scalar with array</h5>

If we add a single number to an array, NumPy adds that number to every element of the array.

The scalar value is automatically broadcast to the complete array.

```python
a = np.array([1,2,3,4])

print(a + 10)
print(a * 2)
```

Here, `10` and `2` are scalars.

NumPy automatically applies the operation to every element:

`[1,2,3,4] + 10` → `[11,12,13,14]`

This is the simplest example of broadcasting.

<h5 style="color:#78B89A; font-weight:bold;">broadcasting example → 1D array with 2D array</h5>

A smaller array can also be broadcast across the rows of a larger array when their dimensions are compatible.

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([10,20,30])

print(a + b)
```

Here `a` has shape `(2,3)` and `b` has shape `(3,)`.

The 1D array `[10,20,30]` is applied to each row:

`[1,2,3] + [10,20,30]`

`[4,5,6] + [10,20,30]`

So broadcasting allows this operation without manually copying `b`.

<h5 style="color:#78B89A; font-weight:bold;">broadcasting example → column-wise broadcasting</h5>

We can also broadcast a column array across the columns of a 2D array.

```python
a = np.array([[1,2,3],
              [4,5,6]])

b = np.array([[10],
              [20]])

print(a + b)
```

Here `a` has shape `(2,3)` and `b` has shape `(2,1)`.

`10` is added to the first row and `20` is added to the second row.

The single column is automatically expanded across all 3 columns.

<h3 style="color:#6FA8DC; font-weight:bold">Broadcasting Rules</h3>

When NumPy performs an operation on two arrays, it compares their shapes **from right to left**.

Two dimensions are compatible when:

1. They are equal.
2. One of them is `1`.

If these conditions are not satisfied, broadcasting is not possible and NumPy gives an error.

<h5 style="color:#78B89A; font-weight:bold;">rule 1 → dimensions are equal</h5>

Example:

`(2,3)` and `(2,3)` → compatible

Both arrays have the same shape.

```python
a = np.ones((2,3))
b = np.ones((2,3))

print(a + b)
```

Both arrays have shape `(2,3)`, so normal element-wise addition is possible.

<h5 style="color:#78B89A; font-weight:bold;">rule 2 → one dimension is 1</h5>

If one of the dimensions is `1`, NumPy can stretch that dimension to match the other array.

```python
a = np.ones((2,3))
b = np.ones((2,1))

print(a + b)
```

Shapes:

`(2,3)`

`(2,1)`

The last dimensions are `3` and `1`, so they are compatible.

The `1` is broadcast to `3`.

<h5 style="color:#78B89A; font-weight:bold;">rule 3 → missing dimensions are treated as 1</h5>

If one array has fewer dimensions, NumPy treats the missing dimensions as `1` while comparing shapes from right to left.

```python
a = np.ones((2,3))
b = np.ones(3)

print((a + b).shape)
print(a + b)
```

Here:

`a` → `(2,3)`

`b` → `(3,)`

While comparing from the right:

`(2,3)`

`(1,3)`

The dimensions are compatible, so broadcasting is possible.

<h5 style="color:#78B89A; font-weight:bold;">when broadcasting fails → incompatible dimensions</h5>

If the dimensions are different and neither dimension is `1`, NumPy cannot broadcast the arrays.

```python
a = np.ones((2,3))
b = np.ones((2,2))

# print(a + b)  # ValueError: shapes are not compatible

print("Shapes:", a.shape, b.shape)
```

Shapes:

`(2,3)`

`(2,2)`

Compare from right to left:

`3` vs `2` → not equal

Neither is `1`, so broadcasting is **not possible**.

The addition would therefore give a `ValueError`.

<h3 style="color:#6FA8DC; font-weight:bold">Quick Revision</h3>

**Broadcasting** → NumPy automatically adjusts compatible array shapes for element-wise operations.

**Main rules:**
- Compare dimensions from **right to left**.
- Dimensions are compatible if they are **equal**.
- Dimensions are also compatible if **one of them is 1**.
- Missing dimensions are treated as `1`.
- If the dimensions are incompatible → **broadcasting fails**.

Broadcasting is very important in ML because we frequently perform operations between arrays, feature vectors, matrices, and scalar values.
