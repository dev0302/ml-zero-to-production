<h3 style="color:#6FA8DC; font-weight:bold">Plotting Graphs in NumPy</h3>

For plotting graphs, we generally use **Matplotlib** along with NumPy.

NumPy is used to create and calculate the values, while Matplotlib is used to visualize them.

First we will see:

- Straight line
- Parabolic curve
- Sin wave
- Scatter plot

<h5 style="color:#78B89A; font-weight:bold;">initially importing libraries → NumPy and Matplotlib</h5>

````python
import numpy as np
import matplotlib.pyplot as plt
````

<h3 style="color:#6FA8DC; font-weight:bold">Straight Line</h3>

A straight-line equation is generally written as:

$$
y = mx + c
$$

where:

- `m` → slope of the line
- `c` → y-intercept
- `x` → input
- `y` → output

For example:

$$
y = 2x + 3
$$

We first generate the `x` values and then calculate the corresponding `y` values.

````python
x = np.arange(10)
y = 2*x + 3

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Straight Line")
plt.show()
````

Here:

`x = [0,1,2,...,9]`

and:

`y = 2x + 3`

`plt.plot(x, y)` joins the `(x,y)` points and produces a straight line.

The equation `y = 2x + 3` is commonly used to understand **linear relationships**.

<h3 style="color:#6FA8DC; font-weight:bold">Parabolic Curve</h3>

A basic parabola can be represented by:

$$
y = x^2
$$

When we square the `x` values, the resulting graph forms a **parabolic curve**.

````python
x = np.arange(-10, 11)
y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Parabolic Curve")
plt.show()
````

Here:

`y = x**2`

means every `x` value is squared.

For example:

`x = -2` → `y = 4`

`x = 0` → `y = 0`

`x = 2` → `y = 4`

Because both positive and negative values of `x` can produce the same `y`, the graph is symmetric around the y-axis.

<h3 style="color:#6FA8DC; font-weight:bold">Sin Wave</h3>

The sine function is:

$$
y = \sin(x)
$$

NumPy provides `np.sin()` to calculate the sine of values.

**Important:** `np.sin()` expects the input in **radians**.

````python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Sin Wave")
plt.show()
````

Here we use:

`np.linspace(0, 2*np.pi, 100)`

which creates **100 evenly spaced values** from `0` to `2π`.

Then:

`np.sin(x)`

calculates the sine value for every `x`.

The graph goes smoothly between `-1` and `1`, producing the familiar **sin wave**.

<h5 style="color:#78B89A; font-weight:bold;">np.linspace() → generating evenly spaced values</h5>

`np.linspace(start, stop, num)`

creates `num` evenly spaced values between `start` and `stop`.

Example:

`np.linspace(0, 10, 5)`

gives:

`[0, 2.5, 5, 7.5, 10]`

````python
x = np.linspace(0, 10, 5)

print(x)
````

`linspace()` is very useful for graph plotting because we usually need many closely spaced `x` values to get a smooth curve.

<h3 style="color:#6FA8DC; font-weight:bold">Scatter Plot</h3>

A scatter plot represents individual data points using dots.

It is useful for understanding the **relationship between two variables**.

For example, suppose we have:

- `x` → hours studied
- `y` → marks obtained

We can plot each student's data as a point.

````python
hours = np.array([1, 2, 3, 4, 5, 6, 7])
marks = np.array([35, 40, 50, 55, 65, 70, 80])

plt.scatter(hours, marks)
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours Studied vs Marks")
plt.show()
````

Here each point represents one observation.

For example:

`(1, 35)` → 1 hour studied and 35 marks

`(5, 65)` → 5 hours studied and 65 marks

A scatter plot is especially useful in **Machine Learning and data analysis** for visually checking relationships or patterns between variables.

<h5 style="color:#78B89A; font-weight:bold;">plot() vs scatter() → important difference</h5>

`plt.plot(x, y)` → mainly used for lines and continuous functions.

`plt.scatter(x, y)` → used to show individual data points.

Example:

For `y = x²` → use `plot()`

For `hours studied vs marks` → use `scatter()`
