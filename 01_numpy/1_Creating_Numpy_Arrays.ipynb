<h3 style="color:#6FA8DC; font-weight:bold">Creating Numpy Arrays</h3>

```python
# step-1 is to import numpy
import numpy as np
```

```python
# np.array -> to create numpy array
np.array([1,2,3,4])
```

```python
# to store this array in variable
a = np.array([1,2,3,4,5,6])
print(a)
```

```python
# to see its type
print(type(a))
```

```python
#dtype
np.array([1,2,3,4], dtype=float)
```
