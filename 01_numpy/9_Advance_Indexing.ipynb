<h3 style="color:#6FA8DC; font-weight:bold">Advance Indexing</h3>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Fancy Indexing → selecting multiple specific indexes at once</h5>

```python
print(a2)

# selecting rows 0, 2 and 1
print(a2[[0,2,1]])
```

<b>Description:</b><br>
Fancy indexing is used when we want to select multiple specific rows or columns at the same time.<br><br>
Instead of selecting a continuous range like <b>0:3</b>, we can directly give the indexes we want inside another list.<br><br>
For example, <b>a2[[0,2,1]]</b> selects row 0, row 2 and row 1 in that order.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Fancy Indexing → selecting specific columns</h5>

```python
print(a2)

# selecting columns 0, 2 and 3
print(a2[:,[0,2,3]])
```

<b>Description:</b><br>
Here <b>:</b> means select all rows, while <b>[0,2,3]</b> selects only columns 0, 2 and 3.<br><br>
So <b>a2[:,[0,2,3]]</b> means → take all rows but only the selected columns.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Boolean Indexing → filtering data based on conditions (VVVV Important)</h5>

```python
print(a1)

print(a1 > 5)

print(a1[a1 > 5])
```

<b>Description:</b><br>
Boolean indexing is <b>very important</b> because it allows us to filter data based on conditions.<br><br>
First we create a condition such as <b>a1 > 5</b>. This gives a Boolean array containing True and False.<br><br>
When we put the condition inside <b>a1[...]</b>, NumPy returns only the elements where the condition is True.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Filtering even numbers</h5>

```python
print(a1[a1 % 2 == 0])
```

<b>Description:</b><br>
To find even numbers, we check whether the remainder after dividing by 2 is equal to 0.<br><br>
So <b>a1 % 2 == 0</b> gives True for even numbers, and boolean indexing returns only those values.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Filtering odd numbers</h5>

```python
print(a1[a1 % 2 != 0])
```

<b>Description:</b><br>
For odd numbers, the remainder after dividing by 2 is not 0.<br><br>
Therefore, <b>a1 % 2 != 0</b> filters and returns only the odd numbers.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Filtering using multiple conditions → AND</h5>

```python
print(a1[(a1 > 2) & (a1 < 8)])
```

<b>Description:</b><br>
We can combine more than one condition using <b>&</b> for AND.<br><br>
Here we want values that are <b>greater than 2 AND less than 8</b>.<br><br>
Only the values satisfying both conditions are returned.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Filtering using multiple conditions → OR</h5>

```python
print(a1[(a1 < 2) | (a1 > 7)])
```

<b>Description:</b><br>
We can use <b>|</b> for OR when either one of the conditions can be True.<br><br>
Here we select values that are <b>less than 2 OR greater than 7</b>.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Filtering values that are greater than 5 and even</h5>

```python
print(a1[(a1 > 5) & (a1 % 2 == 0)])
```

<b>Description:</b><br>
Here we combine two conditions together.<br><br>
First condition → value should be <b>greater than 5</b>.<br>
Second condition → value should be <b>even</b>.<br><br>
The <b>&</b> means both conditions must be True. Therefore only values satisfying both conditions are returned.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Boolean Indexing in 2D array</h5>

```python
print(a2)

print(a2[a2 > 5])
```

<b>Description:</b><br>
Boolean indexing also works with 2D arrays.<br><br>
Here <b>a2 > 5</b> checks every element of the 2D array and returns only the elements that are greater than 5.<br><br>
The result is a 1D array containing the matching values.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Boolean Indexing in ML → filtering dataset values</h5>

```python
marks = np.array([45, 78, 32, 91, 66, 28, 85])

print(marks[marks >= 50])
```

<b>Description:</b><br>
This is a simple example of how boolean indexing can be used in real data.<br><br>
Here we have marks of different students and want to filter students who scored at least 50.<br><br>
The condition <b>marks >= 50</b> creates True/False values and boolean indexing returns only the matching marks.<br><br>
The same idea can be used in ML to filter rows based on feature values or conditions.
