<h3 style="color:#6FA8DC; font-weight:bold">Array Functions</h3>

so many functions are there in numpy, we will be learning important ones as per required for ML

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">max() → tells the maximum value present in an array</h5>

```python
print(a1.max())
print(a2.max())
print(a3.max())

# output = 9, 11 and 7 respectively
```

<b>Description:</b><br>Here max() checks the complete array and returns the largest value present in it.<br><br>For example, in <b>a1</b> the maximum value is 9, in <b>a2</b> it is 11 and in <b>a3</b> it is 7.

<br>

<h5 style="color:#78B89A; font-weight:bold;">min() → tells the minimum value present in an array</h5>

```python
print(a1.min())
print(a2.min())
print(a3.min())

# output = 0, 0.0 and 0 respectively
```

<b>Description:</b><br>min() works opposite to max(). It checks the complete array and returns the smallest value present in it.<br><br>For example, all three arrays start from 0, so the minimum value is 0.

<br>

<h5 style="color:#78B89A; font-weight:bold;">sum() → adds all the elements of an array</h5>

```python
print(a1.sum())
print(a2.sum())
print(a3.sum())
```

<b>Description:</b><br>sum() adds all the elements present in the array and returns a single value when no axis is given.<br><br>For example, <b>a1.sum()</b> adds 0 + 1 + 2 + ... + 9 and gives 45.

<br>

<h5 style="color:#78B89A; font-weight:bold;">prod() → multiplies all the elements of an array</h5>

```python
b1 = np.arange(1,6)
print(b1)
print(b1.prod())
```

<b>Description:</b><br>prod() multiplies all the elements of an array and returns the final product.<br><br>For example, here <b>b1</b> contains 1, 2, 3, 4 and 5, so prod() gives <b>1 × 2 × 3 × 4 × 5 = 120</b>.

<br>

<h5 style="color:#78B89A; font-weight:bold;">max() of each row → use axis=1</h5>

```python
print(a2)
print(a2.max(axis=1))

# output = [ 3.  7. 11.]
```

<b>Description:</b><br>When <b>axis=1</b> is used with a 2D array, NumPy works across each row and gives one result for every row.<br><br>For example, the first row of a2 is <b>[0, 1, 2, 3]</b>, so its maximum is 3. Similarly, the maximum values of the second and third rows are 7 and 11.

<br>

<h5 style="color:#78B89A; font-weight:bold;">max() column wise → use axis=0</h5>

```python
print(a2)
print(a2.max(axis=0))

# output = [ 8.  9. 10. 11.]
```

<b>Description:</b><br>When <b>axis=0</b> is used with a 2D array, NumPy works down each column and gives one result for every column.<br><br>For example, the first column contains <b>0, 4, 8</b>, so its maximum is 8. The same process is repeated for all the columns.

<br>

<h5 style="color:#78B89A; font-weight:bold;">sum() row wise and column wise</h5>

```python
# row wise
print(a2.sum(axis=1))

# column wise
print(a2.sum(axis=0))
```

<b>Description:</b><br><b>axis=1</b> gives the sum of each row, while <b>axis=0</b> gives the sum of each column.<br><br>So axis decides the direction in which NumPy performs the operation.

<br>

<h5 style="color:#78B89A; font-weight:bold;">prod() row wise and column wise</h5>

```python
# row wise
print(a2.prod(axis=1))

# column wise
print(a2.prod(axis=0))
```

<b>Description:</b><br>Just like sum(), prod() can also work row wise or column wise using the axis parameter.<br><br><b>axis=1</b> multiplies the elements of every row, while <b>axis=0</b> multiplies the elements of every column.<br><br>This same axis concept is used with many NumPy functions and is very important when working with ML datasets.
