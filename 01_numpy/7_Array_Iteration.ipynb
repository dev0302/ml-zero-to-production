<h3 style="color:#6FA8DC; font-weight:bold">Array Iteration</h3>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Normal Iteration → using for loop to access elements of an array</h5>

```python
# 1D array
for i in a1:
    print(i)
```

<b>Description:</b><br>
In a 1D array, a normal for loop directly gives us each element one by one.<br><br>
So if a1 contains 0 to 9, the loop will print 0, then 1, then 2 and so on.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Normal Iteration in 2D array</h5>

```python
for i in a2:
    print(i)
```

<b>Description:</b><br>
When we normally iterate over a 2D array, the loop gives us one row at a time.<br><br>
So here each value of <b>i</b> is itself a 1D array representing one complete row.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Normal Iteration in 2D array → accessing every element</h5>

```python
for i in a2:
    for j in i:
        print(j)
```

<b>Description:</b><br>
Since a normal loop over a 2D array gives one row at a time, we use another loop inside it to access every element.<br><br>
The outer loop accesses the rows and the inner loop accesses the elements inside each row.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Normal Iteration in 3D array</h5>

```python
for i in a3:
    for j in i:
        for k in j:
            print(k)
```

<b>Description:</b><br>
A 3D array has three levels, so to access every individual element using normal iteration, we use three nested loops.<br><br>
The first loop goes through the 2D arrays, the second goes through their rows and the third goes through the individual elements.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">np.nditer()</h3>

np.nditer() is used to iterate over all the elements of an array, regardless of its dimensions.

```python
for i in np.nditer(a3):
    print(i)
```

<b>Description:</b><br>
Here np.nditer() directly gives us every individual element of the array one by one.<br><br>
The main advantage is that we do not need to write separate nested loops for a 2D or 3D array.<br><br>
So for higher dimensional arrays, <b>np.nditer()</b> makes iteration much simpler.

<br>

<h5 style="color:#78B89A; font-weight:bold;">np.nditer() with 2D array</h5>

```python
for i in np.nditer(a2):
    print(i)
```

<b>Description:</b><br>
Even though a2 is a 2D array, np.nditer() goes through every element directly instead of giving us complete rows.
