<h3 style="color:#6FA8DC; font-weight:bold">Working with Missing Values</h3>

In real-world datasets, some values may be missing.

NumPy represents a missing numerical value using **`NaN`**.

`NaN` means **Not a Number**.

Missing values are very common in real-world ML datasets, so we need to identify and handle them before training a model.

<h5 style="color:#78B89A; font-weight:bold;">example → how missing values look like</h5>

We can create an array containing `np.nan` to represent missing values.

```python
import numpy as np

a = np.array([1, 2, np.nan, 4, np.nan, 6])

print(a)
```

Here `np.nan` represents a missing value.

So the array:

`[1, 2, np.nan, 4, np.nan, 6]`

contains **2 missing values**.

Now let's understand step-by-step how NumPy helps us find and filter these values.

<h5 style="color:#78B89A; font-weight:bold;">step 1 → np.isnan(a)</h5>

`np.isnan(a)` checks every element of the array.

- If the element is `NaN` → `True`
- If the element is not `NaN` → `False`

It returns a **Boolean array**.

```python
a = np.array([1, 2, np.nan, 4, np.nan, 6])

print(np.isnan(a))
```

Output:

`[False False True False True False]`

So we can understand it as:

- `1` → `False` → not missing
- `2` → `False` → not missing
- `np.nan` → `True` → missing
- `4` → `False` → not missing
- `np.nan` → `True` → missing
- `6` → `False` → not missing

**Important:** `np.isnan(a)` does not return the missing values themselves. It returns a Boolean mask telling us **where the missing values are**.

<h5 style="color:#78B89A; font-weight:bold;">step 2 → a[np.isnan(a)]</h5>

Now we use the Boolean array created by `np.isnan(a)` for **Boolean indexing**.

`a[np.isnan(a)]` means:

> Select the elements of `a` where `np.isnan(a)` is `True`.

```python
a = np.array([1, 2, np.nan, 4, np.nan, 6])

print(a[np.isnan(a)])
```

Output:

`[nan nan]`

So this gives us **only the missing values** from the array.

Break it down:

`np.isnan(a)` → `[False False True False True False]`

Then:

`a[False False True False True False]`

→ selects the positions where the value is `True`

→ `[nan nan]`

<h5 style="color:#78B89A; font-weight:bold;">step 3 → ~np.isnan(a)</h5>

`~` is the **NOT operator** for a Boolean array.

So:

`np.isnan(a)` → finds missing values

`~np.isnan(a)` → finds values which are **NOT missing**

```python
a = np.array([1, 2, np.nan, 4, np.nan, 6])

print(~np.isnan(a))
```

Output:

`[ True  True False  True False  True]`

The `True` and `False` values are reversed.

`np.isnan(a)`:

`[False False True False True False]`

`~np.isnan(a)`:

`[ True  True False  True False  True]`

<h5 style="color:#78B89A; font-weight:bold;">step 4 → a[~np.isnan(a)]</h5>

Now we use this inverted Boolean mask to select only the values which are **not missing**.

```python
a = np.array([1, 2, np.nan, 4, np.nan, 6])

print(a[~np.isnan(a)])
```

Output:

`[1. 2. 4. 6.]`

So:

`a[np.isnan(a)]`

→ gives **only missing values**

while:

`a[~np.isnan(a)]`

→ gives **only non-missing values**.

This is a very important pattern when working with missing data in NumPy.

<h5 style="color:#78B89A; font-weight:bold;">complete flow → understand all 3 steps together</h5>

For:

`a = [1, 2, NaN, 4, NaN, 6]`

### 1. Find missing positions

`np.isnan(a)`

→ `[False False True False True False]`

### 2. Extract missing values

`a[np.isnan(a)]`

→ `[nan nan]`

### 3. Extract non-missing values

`a[~np.isnan(a)]`

→ `[1. 2. 4. 6.]`

So the basic pattern is:

`np.isnan(a)` → **find**

`a[np.isnan(a)]` → **get missing values**

`a[~np.isnan(a)]` → **get non-missing values**

<h5 style="color:#78B89A; font-weight:bold;">step 5 → count missing values</h5>

Once we have the Boolean mask, we can also count how many values are missing.

In NumPy, `True` behaves like `1` and `False` behaves like `0` when we perform a sum.

```python
a = np.array([1, 2, np.nan, 4, np.nan, 6])

missing = np.isnan(a)

print("Missing values:", np.sum(missing))
```

There are **2 missing values** in the array.

`np.sum(np.isnan(a))`

is a common way to count the number of `NaN` values.
