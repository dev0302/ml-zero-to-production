<h3 style="color:#6FA8DC; font-weight:bold">Statistical Functions</h3>

numpy has many statistical functions, we will be learning important ones as per required for ML

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">mean() → tells the average value of all elements in an array</h5>

```python
print(a1.mean())
print(a2.mean())
print(a3.mean())
```

<b>Description:</b><br>mean() calculates the average of all the elements present in the array.<br><br>For example, a1 contains values from 0 to 9, so its mean is <b>4.5</b>.

<br>

<h5 style="color:#78B89A; font-weight:bold;">median() → tells the middle value of the data</h5>

```python
print(np.median(a1))
print(np.median(a2))
print(np.median(a3))
```

<b>Description:</b><br>median() gives the middle value when the elements are arranged in sorted order.<br><br>For example, a1 contains 0 to 9. The middle values are 4 and 5, so the median is <b>4.5</b>.

<br>

<h5 style="color:#78B89A; font-weight:bold;">std() → tells the standard deviation of the data</h5>

```python
print(a1.std())
print(a2.std())
print(a3.std())
```

<b>Description:</b><br>std() tells us how much the values are spread out from their mean value.<br><br>A smaller standard deviation means the values are closer to the mean, while a larger value means the data is more spread out.<br><br>Standard deviation is commonly used in ML while understanding and preprocessing datasets.

<br>

<h5 style="color:#78B89A; font-weight:bold;">var() → tells the variance of the data</h5>

```python
print(a1.var())
print(a2.var())
print(a3.var())
```

<b>Description:</b><br>var() tells how much the values are spread around their mean by calculating the average squared difference from the mean.<br><br>Variance is related to standard deviation. Standard deviation is the square root of variance.

<br>

<h3 style="color:#6FA8DC; font-weight:bold">Array Dot Product</h3>

```python
b1 = np.arange(1,5)
b2 = np.arange(5,9)

print(b1)
print(b2)

print(np.dot(b1,b2))
```

<b>Description:</b><br>Dot product multiplies the corresponding elements of two arrays and then adds all the results.<br><br>For example, <b>b1 = [1,2,3,4]</b> and <b>b2 = [5,6,7,8]</b>.<br><br>So the dot product is <b>(1×5) + (2×6) + (3×7) + (4×8) = 70</b>.<br><br>Dot product is very important in ML because it is used in calculations involving vectors, weights and features.

<br>

<h5 style="color:#78B89A; font-weight:bold;">round() → rounds values to the nearest number</h5>

```python
a4 = np.array([1.2, 2.5, 3.7, 4.4, 5.8])

print(np.round(a4))
```

<b>Description:</b><br>round() rounds the decimal values to the nearest integer by default.<br><br>For example, <b>1.2</b> becomes 1, <b>3.7</b> becomes 4 and <b>5.8</b> becomes 6.

<br>

<h5 style="color:#78B89A; font-weight:bold;">floor() → rounds values down to the nearest integer</h5>

```python
print(np.floor(a4))
```

<b>Description:</b><br>floor() always rounds a value down to the nearest integer.<br><br>For example, <b>2.9</b> becomes 2 and <b>5.8</b> becomes 5.

<br>

<h5 style="color:#78B89A; font-weight:bold;">ceil() → rounds values up to the nearest integer</h5>

```python
print(np.ceil(a4))
```

<b>Description:</b><br>ceil() always rounds a value up to the nearest integer.<br><br>For example, <b>1.2</b> becomes 2, <b>3.7</b> becomes 4 and <b>5.8</b> becomes 6.<br><br>round(), floor() and ceil() are useful when we need to control decimal values during data processing.
