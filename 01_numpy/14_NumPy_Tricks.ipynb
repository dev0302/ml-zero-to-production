<h3 style="color:#6FA8DC; font-weight:bold">NumPy Tricks</h3>

NumPy has many useful functions which make array manipulation and data analysis easier.

These functions are especially useful while working with **Machine Learning datasets**.

<h5 style="color:#78B89A; font-weight:bold;">initially creating arrays → used for examples</h5>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<h5 style="color:#78B89A; font-weight:bold;">np.sort() → sorting an array</h5>

`np.sort()` returns the elements of an array in sorted order.

By default, it sorts in **ascending order**.

```python
a = np.array([5, 2, 8, 1, 4])

print(np.sort(a))
```

The original array is not changed.

`np.sort(a)` → `[1 2 4 5 8]`

For a 2D array, we can also sort along an axis.

```python
a = np.array([[5, 2, 8],
              [1, 7, 3]])

print(np.sort(a))
print(np.sort(a, axis=0))
```

<h5 style="color:#78B89A; font-weight:bold;">np.append() → adding elements</h5>

`np.append()` adds elements to the end of an array.

It returns a **new array**.

```python
a = np.array([1, 2, 3])

a = np.append(a, 4)

print(a)
```

We can also append multiple values:

`np.append(a, [4,5])`

For a 1D array, the new values are added at the end.

<h5 style="color:#78B89A; font-weight:bold;">np.concatenate() → joining arrays</h5>

`np.concatenate()` joins two or more arrays along an existing axis.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate((a, b)))
```

For 2D arrays, we can choose the axis:

- `axis=0` → join row-wise
- `axis=1` → join column-wise

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

print("axis=0:")
print(np.concatenate((a, b), axis=0))

print("axis=1:")
print(np.concatenate((a, b), axis=1))
```

<h5 style="color:#78B89A; font-weight:bold;">np.unique() → removing duplicate values</h5>

`np.unique()` returns the unique values from an array.

It is very useful when we want to know the different categories or values present in a dataset.

```python
a = np.array([1, 2, 2, 3, 4, 4, 4, 5])

print(np.unique(a))
```

<h5 style="color:#78B89A; font-weight:bold;">np.expand_dims() → adding a new dimension</h5>

`np.expand_dims()` adds a new axis/dimension to an array.

This is useful when the shape of an array needs to be changed before passing it to an ML model or performing another operation.

```python
a = np.array([1, 2, 3, 4])

print("Original shape:", a.shape)

b = np.expand_dims(a, axis=0)

print("New shape:", b.shape)
print(b)
```

Original shape:

`(4,)`

After `axis=0`:

`(1,4)`

So we have converted a 1D array into a **2D row array**.

```python
a = np.array([1, 2, 3, 4])

b = np.expand_dims(a, axis=1)

print(b)
print("Shape:", b.shape)
```

With `axis=1`, the shape becomes `(4,1)`.

So it becomes a **2D column array**.

<h5 style="color:#78B89A; font-weight:bold;">np.where() → finding positions based on condition</h5>

`np.where()` can be used to find the **indices/positions** where a condition is true.

```python
a = np.array([10, 25, 5, 40, 15])

print(np.where(a > 20))
```

The values greater than `20` are at indices `1` and `3`.

So:

`np.where(a > 20)` → positions where the condition is `True`.

It is very useful for filtering and finding positions in datasets.

<h5 style="color:#78B89A; font-weight:bold;">np.argmax() and np.argmin() → position of maximum/minimum</h5>

`np.argmax()` returns the **index of the largest value**.

`np.argmin()` returns the **index of the smallest value**.

```python
a = np.array([10, 50, 20, 80, 30])

print("Index of maximum:", np.argmax(a))
print("Index of minimum:", np.argmin(a))
```

Here:

Maximum value = `80` → index `3`

Minimum value = `10` → index `0`

Important: these functions return the **position**, not the value itself.

<h5 style="color:#78B89A; font-weight:bold;">np.cumsum() → cumulative sum</h5>

Cumulative sum means:

> Keep adding the current value to all the values that came before it.

Example:

| Value | Cumulative Sum |
|---:|---:|
| 2 | 2 |
| 3 | 5 |
| 4 | 9 |
| 1 | 10 |

Calculation:

`2`

`2 + 3 = 5`

`2 + 3 + 4 = 9`

`2 + 3 + 4 + 1 = 10`

```python
a = np.array([2, 3, 4, 1])

print(np.cumsum(a))
```

Output:

`[ 2  5  9 10]`

So every position stores the sum of all values **up to that position**.

This can be useful for cumulative totals, running counts, financial data, and many data-analysis tasks.

<h5 style="color:#78B89A; font-weight:bold;">np.percentile() → finding a percentile</h5>

A percentile tells us the value below which a certain percentage of observations fall.

For example:

- 50th percentile → median
- 25th percentile → Q1
- 75th percentile → Q3

```python
a = np.array([10, 20, 30, 40, 50])

print("25th percentile:", np.percentile(a, 25))
print("50th percentile:", np.percentile(a, 50))
print("75th percentile:", np.percentile(a, 75))
```

The **50th percentile** is the median.

Percentiles are commonly used in statistics and ML to understand the distribution of data and identify unusually high or low values.

<h5 style="color:#78B89A; font-weight:bold;">np.histogram() → frequency distribution</h5>

`np.histogram()` divides numerical data into **bins** and counts how many values fall inside each bin.

It returns two arrays:

- `hist` → number of values in each bin
- `bins` → boundaries of the bins

```python
a = np.array([1, 2, 2, 3, 4, 4, 4, 5, 6, 7])

hist, bins = np.histogram(a, bins=3)

print("Frequency:", hist)
print("Bins:", bins)
```

For example, if a bin covers a certain range, `hist` tells us how many values fall inside that range.

This is useful for understanding the **distribution of a dataset** before applying ML algorithms.

<h5 style="color:#78B89A; font-weight:bold;">np.corrcoef() → correlation between variables</h5>

Correlation tells us how strongly two variables are related.

`np.corrcoef()` returns a **correlation coefficient** between `-1` and `+1`.

### What the value tells us:

- `+1` → perfect positive relationship
- `0` → no linear relationship
- `-1` → perfect negative relationship

Positive correlation:

> As one variable increases, the other generally increases.

Negative correlation:

> As one variable increases, the other generally decreases.

Important: **correlation does not mean causation.**

```python
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([35, 45, 55, 65, 75])

correlation = np.corrcoef(hours, marks)

print(correlation)
```

Here, hours studied and marks increase together.

So the correlation is close to `+1`.

This tells us that there is a **strong positive linear relationship** between hours studied and marks in this example.

It does NOT prove that studying more is the only reason marks increase — correlation only tells us about the relationship between the variables.

```python
temperature = np.array([10, 20, 30, 40, 50])
ice_cream_sales = np.array([90, 80, 60, 40, 20])

print(np.corrcoef(temperature, ice_cream_sales))
```

Here temperature increases while ice-cream sales decrease, so the correlation is close to `-1`.

This represents a **strong negative linear relationship** in this example.

<h5 style="color:#78B89A; font-weight:bold;">np.isin() → checking whether values belong to a collection</h5>

`np.isin()` checks whether each element of an array is present in another set/list of values.

It returns a Boolean array.

```python
a = np.array([1, 2, 3, 4, 5, 6])

print(np.isin(a, [2, 4, 6]))
```

Output:

`[False True False True False True]`

So `2`, `4`, and `6` are present in the given list.

We can also use it for filtering:

```python
a = np.array([1, 2, 3, 4, 5, 6])

print(a[np.isin(a, [2, 4, 6])])
```

<h5 style="color:#78B89A; font-weight:bold;">np.flip() → reversing an array</h5>

`np.flip()` reverses the order of elements in an array.

```python
a = np.array([1, 2, 3, 4, 5])

print(np.flip(a))
```

For a 2D array, `np.flip()` can reverse all elements or we can specify an axis.

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

print(np.flip(a))
print(np.flip(a, axis=0))
print(np.flip(a, axis=1))
```

<h5 style="color:#78B89A; font-weight:bold;">np.put() → replacing values at specific positions</h5>

`np.put()` replaces values at specified **indices**.

It modifies the original array.

```python
a = np.array([10, 20, 30, 40, 50])

np.put(a, [1, 3], [200, 400])

print(a)
```

Here:

index `1` → `20` is replaced by `200`

index `3` → `40` is replaced by `400`

Result:

`[10 200 30 400 50]`

<h5 style="color:#78B89A; font-weight:bold;">np.delete() → removing elements</h5>

`np.delete()` removes elements at the specified indices.

It returns a **new array**; the original array is not changed.

```python
a = np.array([10, 20, 30, 40, 50])

b = np.delete(a, [1, 3])

print("Original:", a)
print("After delete:", b)
```

Here indices `1` and `3` are removed.

So:

`[10, 20, 30, 40, 50]`

becomes:

`[10, 30, 50]`

<h3 style="color:#6FA8DC; font-weight:bold">Set Functions</h3>

NumPy also provides functions for performing **set operations** on arrays.

These are useful when comparing two datasets or finding common/different values.

Important set functions:

- `np.union1d()`
- `np.intersect1d()`
- `np.setdiff1d()`
- `np.setxor1d()`

<h5 style="color:#78B89A; font-weight:bold;">np.union1d() → all unique values from both arrays</h5>

Union combines the values from both arrays and removes duplicates.

```python
a = np.array([1, 2, 3, 4])
b = np.array([3, 4, 5, 6])

print(np.union1d(a, b))
```

<h5 style="color:#78B89A; font-weight:bold;">np.intersect1d() → common values</h5>

Intersection returns the values which are present in **both arrays**.

```python
a = np.array([1, 2, 3, 4])
b = np.array([3, 4, 5, 6])

print(np.intersect1d(a, b))
```

<h5 style="color:#78B89A; font-weight:bold;">np.setdiff1d() → values present in first array but not second</h5>

`np.setdiff1d(a, b)` gives values which are in `a` but **not in `b`**.

```python
a = np.array([1, 2, 3, 4])
b = np.array([3, 4, 5, 6])

print(np.setdiff1d(a, b))
print(np.setdiff1d(b, a))
```

Order matters here.

`setdiff1d(a, b)` → values in `a` but not `b`

`setdiff1d(b, a)` → values in `b` but not `a`

<h5 style="color:#78B89A; font-weight:bold;">np.setxor1d() → values present in only one of the arrays</h5>

Symmetric difference returns values which are in **either array, but not in both**.

```python
a = np.array([1, 2, 3, 4])
b = np.array([3, 4, 5, 6])

print(np.setxor1d(a, b))
```
