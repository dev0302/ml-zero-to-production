<h3 style="color:#6FA8DC; font-weight:bold">Array Operations</h3>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Scalar Operations → performing an operation between an array and a single value</h5>

```python
# addition
print(a1 + 2)

# subtraction
print(a1 - 2)

# multiplication
print(a1 * 2)

# division
print(a1 / 2)

# power
print(a1 ** 2)
```

<b>Description:</b><br>Here we are performing scalar operations on a1. A single value is applied to every element of the array.<br><br>For example, <b>a1 + 2</b> adds 2 to every element, <b>a1 * 2</b> multiplies every element by 2 and <b>a1 ** 2</b> squares every element.<br><br>This is useful in ML when we want to scale, shift or transform all values in an array.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Relational Operations → comparing each element of an array with a value</h5>

```python
# greater than
print(a1 > 5)

# less than
print(a1 < 5)

# greater than or equal to
print(a1 >= 5)

# less than or equal to
print(a1 <= 5)

# equal to
print(a1 == 5)

# not equal to
print(a1 != 5)
```

<b>Description:</b><br>Relational operations compare every element of the array and return <b>True</b> or <b>False</b> for each element.<br><br>For example, <b>a1 > 5</b> checks which elements are greater than 5. Only the element 5 itself is <b>False</b> for this condition, while 6, 7, 8 and 9 are <b>True</b>.<br><br>These operations are very useful for filtering data and creating conditions in ML.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Vector Operations → performing operations between two arrays element by element</h5>

```python
# creating another array of same size as a1
b1 = np.arange(10, 20)

# addition
print(a1 + b1)

# subtraction
print(a1 - b1)

# multiplication
print(a1 * b1)

# division
print(b1 / (a1 + 1))

# power
print(a1 ** 2)
```

<b>Description:</b><br>Vector operations are performed between two arrays element by element.<br><br>For example, in <b>a1 + b1</b>, the first element of a1 is added to the first element of b1, the second element is added to the second element, and so on.<br><br>The two arrays generally need compatible shapes for element-wise operations. NumPy performs these operations directly without writing a separate loop for every element.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Relational Operations between Arrays → comparing corresponding elements of two arrays</h5>

```python
b1 = np.arange(10, 20)

print(a1 > b1)
print(a1 < b1)
print(a1 == b1)
print(a1 != b1)
```

<b>Description:</b><br>Here we compare two arrays element by element. For example, <b>a1 > b1</b> checks whether each element of a1 is greater than the corresponding element of b1.<br><br>The result is a Boolean array containing True or False values for every comparison.
