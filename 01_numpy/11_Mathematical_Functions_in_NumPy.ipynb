<h3 style="color:#6FA8DC; font-weight:bold">Sigmoid Function</h3>

Sigmoid is an important mathematical function used in **Machine Learning**, especially in **binary classification**.

It takes any real number as input and converts it into a value between **0 and 1**.

<h5 style="color:#78B89A; font-weight:bold;">sigmoid formula → mathematical representation</h5>

The formula of sigmoid is:

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

where:

- `x` → input value
- `e` → Euler's number, approximately `2.718`
- `e⁻ˣ` → exponential term

The important property is:

$$
0 < \sigma(x) < 1
$$

<h5 style="color:#78B89A; font-weight:bold;">sigmoid → how it works</h5>

The sigmoid function converts different input values into values between `0` and `1`.

- If `x` is a **large positive number** → sigmoid approaches `1`
- If `x = 0` → sigmoid is `0.5`
- If `x` is a **large negative number** → sigmoid approaches `0`

So sigmoid can be thought of as a function which **squashes** a large range of numbers into the range `0 to 1`.

This makes it useful when we want to interpret the output as a **probability**.

```python
x = np.array([-10, -2, 0, 2, 10])

sigmoid = 1 / (1 + np.exp(-x))

print(sigmoid)
```

For the input values:

`[-10, -2, 0, 2, 10]`

the sigmoid values are approximately:

`[0.0000, 0.1192, 0.5000, 0.8808, 1.0000]`

Notice how every value is converted into the range `0 to 1`.

At `x = 0`:

$$
\sigma(0) = \frac{1}{1 + e^0} = \frac{1}{2} = 0.5
$$

<h5 style="color:#78B89A; font-weight:bold;">np.exp() → exponential function</h5>

`np.exp()` calculates the exponential value $e^x$ for each element.

We use `np.exp()` in the sigmoid formula to calculate the term $e^{-x}$.

```python
x = np.array([-2, -1, 0, 1, 2])

print(np.exp(x))
```

Here NumPy calculates $e^x$ element-wise.

This is useful because NumPy can perform the exponential operation on the **whole array at once**, instead of using a loop.

<h5 style="color:#78B89A; font-weight:bold;">sigmoid → using a function</h5>

Instead of writing the formula again and again, we can create our own sigmoid function.

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.array([-3, -1, 0, 1, 3])

print(sigmoid(x))
```

This is one of the important benefits of NumPy.

Because of **vectorization**, the same sigmoid function can directly work on an entire NumPy array.

<h5 style="color:#78B89A; font-weight:bold;">sigmoid → uses in machine learning</h5>

Sigmoid is mainly used when the output needs to represent a value between `0 and 1`.

### Common use:

**Binary classification**

For example, predicting whether an email is spam or not spam.

The model can produce a sigmoid output such as:

`0.92`

This can be interpreted as a high probability of the positive class.

Another example:

`0.15`

This represents a low probability of the positive class.

Sigmoid was also traditionally used as an **activation function in neural networks**, although modern deep networks commonly use other activations such as ReLU in hidden layers.

<h5 style="color:#78B89A; font-weight:bold;">sigmoid → threshold for binary classification</h5>

A common classification rule is to use `0.5` as the threshold.

- sigmoid output `>= 0.5` → class `1`
- sigmoid output `< 0.5` → class `0`

```python
probabilities = np.array([0.2, 0.7, 0.4, 0.9])

predictions = (probabilities >= 0.5).astype(int)

print(predictions)
```

Here:

`[0.2, 0.7, 0.4, 0.9]`

becomes:

`[0, 1, 0, 1]`

because we are using `0.5` as the classification threshold.

<h3 style="color:#6FA8DC; font-weight:bold">Mean Squared Error (MSE)</h3>

Mean Squared Error is a very important **loss function** used in Machine Learning.

It measures how different the model's predictions are from the actual values.

In simple words:

**MSE tells us how much error our predictions have.**

Lower MSE → better predictions.

Higher MSE → larger prediction errors.

<h5 style="color:#78B89A; font-weight:bold;">MSE formula → mathematical representation</h5>

The formula is:

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2
$$

where:

- `n` → number of data points
- $y_i$ → actual value
- $\hat{y}_i$ → predicted value
- $(y_i - \hat{y}_i)$ → error
- square of error → makes all errors positive
- average of squared errors → MSE

<h5 style="color:#78B89A; font-weight:bold;">MSE → simple example</h5>

Suppose actual values are:

`[3, 5, 7]`

and our model predicted:

`[2, 6, 8]`

Errors:

`[3-2, 5-6, 7-8]`

→ `[1, -1, -1]`

Squared errors:

`[1², (-1)², (-1)²]`

→ `[1, 1, 1]`

Mean:

`(1 + 1 + 1) / 3 = 1`

So MSE = **1**.

```python
actual = np.array([3, 5, 7])
predicted = np.array([2, 6, 8])

mse = np.mean((actual - predicted) ** 2)

print("MSE:", mse)
```

Here we use three NumPy operations:

- `actual - predicted` → calculates the error
- `** 2` → squares every error
- `np.mean()` → calculates the average

Therefore:

`MSE = 1.0`

This is exactly the formula implemented using NumPy.

<h5 style="color:#78B89A; font-weight:bold;">MSE → why do we square the error?</h5>

Suppose the errors are:

`[2, -2]`

If we simply take the mean:

`(2 + (-2)) / 2 = 0`

The positive and negative errors cancel each other.

Instead, we square them:

`[2², (-2)²]` → `[4, 4]`

Now the errors cannot cancel each other.

Also, squaring gives **more importance to larger errors**.

```python
actual = np.array([10, 20])
predicted = np.array([12, 18])

errors = actual - predicted
squared_errors = errors ** 2
mse_value = np.mean(squared_errors)

print("Errors:", errors)
print("Squared Errors:", squared_errors)
print("MSE:", mse_value)
```

Here the errors are `[-2, 2]`.

After squaring:

`[4, 4]`

and their mean is:

`4`

So the MSE is `4`.

This step-by-step calculation makes it easier to understand what happens internally when we use the MSE formula.

<h5 style="color:#78B89A; font-weight:bold;">MSE → function in NumPy</h5>

We can create a reusable function for calculating MSE.

```python
def mse(actual, predicted):
    return np.mean((actual - predicted) ** 2)

actual = np.array([10, 20, 30, 40])
predicted = np.array([12, 18, 33, 35])

print("MSE:", mse(actual, predicted))
```

The function can now be used with any two arrays containing the actual and predicted values, as long as their shapes are compatible.

This is useful in ML when evaluating the performance of a regression model.

### Sigmoid

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

- Converts values into the range `0 to 1`
- `sigmoid(0) = 0.5`
- Mainly useful for binary classification
- `np.exp()` is used to calculate the exponential term

### Mean Squared Error

$$
MSE = \frac{1}{n}\sum(y - \hat{y})^2
$$

- Measures prediction error
- Used mainly for regression problems
- Lower MSE → better predictions
- Errors are squared so positive and negative errors do not cancel
- Large errors are penalized more because of squaring

Both are important mathematical concepts for understanding **Machine Learning algorithms and model training**.
