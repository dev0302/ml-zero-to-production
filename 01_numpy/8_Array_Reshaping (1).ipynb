<h3 style="color:#6FA8DC; font-weight:bold">Array Reshaping</h3>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">reshape() → changes the shape of an array</h5>

```python
print(a1)

print(a1.reshape(2,5))

print(a1.reshape(5,2))
```

<b>Description:</b><br>
reshape() changes the shape of an array without changing its elements.<br><br>
a1 has 10 elements, so we can reshape it into <b>2 × 5</b> or <b>5 × 2</b> because both shapes contain 10 elements.<br><br>
The total number of elements must remain the same.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Transpose → converting rows into columns and columns into rows</h5>

```python
print(a2)

print(a2.T)
```

<b>Description:</b><br>
Transpose changes the arrangement of a 2D array by converting rows into columns and columns into rows.<br><br>
For example, the first row of a2 becomes the first column after transpose.<br><br>
Transpose is commonly used in ML when we need to change the orientation of a dataset or matrix.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Horizontal Stacking → joining arrays side by side</h5>

```python
b1 = np.arange(1,5).reshape(2,2)
b2 = np.arange(5,9).reshape(2,2)

print(b1)
print(b2)

print(np.hstack((b1,b2)))
```

<b>Description:</b><br>
hstack() joins arrays horizontally, meaning they are placed side by side.<br><br>
Here both arrays have 2 rows. After horizontal stacking, the number of rows remains 2 and the columns are combined.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Vertical Stacking → joining arrays one below another</h5>

```python
print(b1)
print(b2)

print(np.vstack((b1,b2)))
```

<b>Description:</b><br>
vstack() joins arrays vertically, meaning one array is placed below the other.<br><br>
Here both arrays have 2 columns. After vertical stacking, the columns remain 2 and the rows are combined.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Horizontal vs Vertical Stacking</h5>

```python
print("Horizontal stacking:")
print(np.hstack((b1,b2)))

print("Vertical stacking:")
print(np.vstack((b1,b2)))
```

<b>Description:</b><br>
<b>hstack()</b> → joins arrays side by side → columns increase.<br>
<b>vstack()</b> → joins arrays one below another → rows increase.<br><br>
Both arrays need compatible dimensions for the stacking operation.

<br>

<h5 style="color:#78B89A; font-weight:bold;">hsplit() → splits an array horizontally</h5>

```python
print(a2)

print(np.hsplit(a2, 2))
```

<b>Description:</b><br>
hsplit() splits a 2D array horizontally, so the columns are divided into separate arrays.<br><br>
Here a2 has 4 columns and we split it into 2 parts. Each part therefore contains 2 columns.

<br>

<h5 style="color:#78B89A; font-weight:bold;">vsplit() → splits an array vertically</h5>

```python
print(a2)

print(np.vsplit(a2, 3))
```

<b>Description:</b><br>
vsplit() splits a 2D array vertically, so the rows are divided into separate arrays.<br><br>
Here a2 has 3 rows and we split it into 3 parts, so each resulting array contains one row.

<br>

<h5 style="color:#78B89A; font-weight:bold;">hsplit vs vsplit</h5>

```python
# hsplit → split columns
print(np.hsplit(a2, 2))

# vsplit → split rows
print(np.vsplit(a2, 3))
```

<b>Description:</b><br>
<b>hsplit()</b> divides the array column-wise, while <b>vsplit()</b> divides the array row-wise.<br><br>
Easy way to remember:<br>
<b>h</b> → horizontal split → columns are split<br>
<b>v</b> → vertical split → rows are split
