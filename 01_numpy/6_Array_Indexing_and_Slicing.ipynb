<h3 style="color:#6FA8DC; font-weight:bold">Array Indexing and Slicing</h3>

```python
import numpy as np

# initially creating 3 arrays
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2) #3D matrix / 3D Tensor

print(a1)
print(a2)
print(a3)
```

<br>

<h5 style="color:#78B89A; font-weight:bold;">Indexing → accessing a particular element from an array</h5>

```python
# 1D array
print(a1[5])

# 2D array
print(a2[1,2])

# 3D array
print(a3[1,0,1])
```

<b>Description:</b><br>Indexing is used when we want to retrieve a particular element from an array. Indexing starts from <b>0</b>.<br><br>For a 1D array, we need <b>one input</b> because there is only one dimension.<br>For a 2D array, we need <b>two inputs</b> → row and column.<br>For a 3D array, we need <b>three inputs</b> → layer, row and column.

<br>

<h5 style="color:#78B89A; font-weight:bold;">How to approach indexing in a 3D array?</h5>

```python
print(a3)

# to find 5
print(a3[1,0,1])
```

<b>Description:</b><br>First notice that <b>a3</b> is a 3D array. So we need <b>3 inputs</b> to reach any particular element.<br><br>A 3D array is made of 2D arrays, and a 2D array is made of 1D arrays.<br><br>Now we want to find <b>5</b>.<br><br>• First, see where 5 is located. It is in the <b>second 2D array</b>, so we write <b>1</b> because indexing starts from 0.<br><br>• Now we are inside the second 2D array. It has two rows. 5 is in the <b>first row</b>, so we write <b>0</b>.<br><br>• Now we are inside the first row. 5 is the <b>second element</b>, so we write <b>1</b>.<br><br>Therefore, the index of 5 is <b>[1, 0, 1]</b>.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Negative Indexing → accessing elements from the end of an array</h5>

```python
print(a1[-1])
print(a1[-2])

print(a2[-1,-1])

print(a3[-1,-1,-1])
```

<b>Description:</b><br>Negative indexing starts from the end of the array.<br><br><b>-1</b> means the last element, <b>-2</b> means the second last element and so on.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Slicing → retrieving more than one element from an array</h5>

```python
# 1D array
print(a1[2:6])

# 2D array
print(a2[0:2, 1:3])

# 3D array
print(a3[0:2, 0:2, 0:1])
```

<b>Description:</b><br>Slicing is used when we need to retrieve a range of elements instead of only one element.<br><br>The basic syntax is <b>start : stop</b>.<br><br>The important thing to remember is that the <b>last index is not included</b>.<br><br>For example, <b>a1[2:6]</b> starts from index 2 and goes up to index 5. Index 6 is not included.<br><br>In a 2D or 3D array, we can use slicing on each dimension by separating the ranges using a comma.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Slicing in 1D array → start : stop</h5>

```python
print(a1)
print(a1[2:7])

# start from beginning
print(a1[:5])

# go till the end
print(a1[5:])
```

<b>Description:</b><br>In a 1D array, slicing uses one range.<br><br><b>a1[2:7]</b> gives elements from index 2 to index 6. Index 7 is not included.<br><br>If we leave the start empty, slicing starts from the beginning. If we leave the stop empty, slicing continues till the end.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Slicing in 2D array → row slicing, column slicing</h5>

```python
print(a2)

# first 2 rows
print(a2[0:2, :])

# first 3 columns
print(a2[:, 0:3])

# rows 0 to 1 and columns 1 to 2
print(a2[0:2, 1:3])
```

<b>Description:</b><br>In a 2D array, the first part before the comma is used for <b>rows</b> and the second part is used for <b>columns</b>.<br><br>For example, <b>a2[0:2, :]</b> selects rows 0 and 1 and all columns.<br><b>a2[:, 0:3]</b> selects all rows and columns 0, 1 and 2.<br>So the colon <b>:</b> means we want all elements in that particular direction.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Slicing in 3D array → layer, row and column slicing</h5>

```python
print(a3)

# first layer
print(a3[0:1, :, :])

# all layers, first row and all columns
print(a3[:, 0:1, :])

# all layers, all rows and first column
print(a3[:, :, 0:1])
```

<b>Description:</b><br>In a 3D array, we use three parts separated by commas → <b>layer, row, column</b>.<br><br>For example, <b>a3[0:1, :, :]</b> selects the first layer, all its rows and all its columns.<br><br>Similarly, <b>a3[:, 0:1, :]</b> selects all layers but only the first row from each layer.<br><br>The same <b>start : stop</b> rule is followed here too — the stop value is not included.

<br>

<h5 style="color:#78B89A; font-weight:bold;">Colon Syntax → start : stop : step</h5>

```python
print(a1[1:8:2])

# reverse array
print(a1[::-1])

# every second column
print(a2[:, ::2])
```

<b>Description:</b><br>Slicing can also use three values → <b>start : stop : step</b>.<br><br><b>start</b> tells where to start, <b>stop</b> tells where to stop and <b>step</b> tells how many positions to jump.<br><br>For example, <b>a1[1:8:2]</b> starts from index 1 and takes every second element until index 8 (index 8 is not included).<br><br><b>a1[::-1]</b> reverses the complete array because the step is -1.
